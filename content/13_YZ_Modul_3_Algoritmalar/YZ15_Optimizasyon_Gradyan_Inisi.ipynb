## 1. Sisli Dağdan İniş 🏔️🌫️

Sisli bir dağdasın, etrafını göremiyorsun ve vadiye (en alçak noktaya) inmek istiyorsun. Ne yaparsın?

1. Ayağının altındaki **eğime** bakarsın.
2. **Aşağı doğru** küçük bir adım atarsın.
3. Tekrar edersin… Zemin düzleşince (eğim ≈ 0) vadiye varmışsındır!

Yapay zekâ modelleri **hatalarını en aza indirmek** için tam olarak bunu yapar. Buna **gradyan inişi** denir.

## 2. Matematiği (sadece bu kadarı yeterli!)

Fonksiyonumuz: **f(x) = (x − 3)² + 2** → En düşük nokta x = 3'te.

Eğim (türev): **eğim = 2·(x − 3)**

Güncelleme kuralı:
```text
x_yeni = x - ogrenme_orani * egim
```
- Eğim **pozitifse** x azalır (sola gider), **negatifse** x artar (sağa gider).
- **Öğrenme oranı** adımın büyüklüğüdür.

### ✏️ Örnek 15.1: Fonksiyonu çiz

`f(x)` ve `egim(x)` fonksiyonlarını yaz. x = −2'den 8'e kadar `f(x)` grafiğini çiz (`range` yerine `[i / 10 for i in range(-20, 81)]` kullanabilirsin). En düşük nokta nerede?

### ✏️ Örnek 15.2: İlk 5 adım

x = 9'dan başla, öğrenme oranı 0.1 olsun. 5 adım boyunca `x`, `f(x)` ve `egim` değerlerini yazdır.

💡 **Beklenen çıktı:**
```text
Adım 1: x=7.800  f(x)=25.040  eğim=12.000
Adım 2: x=6.840  f(x)=16.746  eğim=9.600
...
```

### ✏️ Örnek 15.3: Vadiye in

Eğimin mutlak değeri 0.001'den küçük olana kadar devam eden bir `while` döngüsü yaz. Kaç adımda vardın, son x kaç?

💡 **Beklenen çıktı:**
```text
43 adımda vardım: x = 3.0004
```

### ✏️ Örnek 15.4: Yolu çiz

Her adımdaki x değerlerini bir listede tut. Fonksiyon grafiğinin üzerine bu noktaları kırmızı (`plt.scatter`) çiz.

## 3. Öğrenme Oranı Deneyleri 🧪

### ✏️ Örnek 15.5: Farklı öğrenme oranları

15.3'teki kodu bir fonksiyona çevir: `inis(baslangic, oran, max_adim=1000)` adım sayısını ve son x'i döndürsün. Öğrenme oranını **0.01, 0.1, 0.5, 0.9 ve 1.1** yaparak dene. Ne oluyor?

💡 **Beklenen çıktı:**
```text
oran=0.01 → 465 adım, x=3.0005
oran=0.1 → 43 adım, x=3.0004
oran=0.5 → 1 adım, x=3.0000
oran=0.9 → 43 adım, x=2.9996
oran=1.1 → vadiye inemedi! (patladı)
```
