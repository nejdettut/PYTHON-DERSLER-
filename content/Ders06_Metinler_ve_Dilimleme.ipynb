## 1. String (Metin) Veri Tipi

Tırnak içine yazılan her şey **str** tipindedir. Tek, çift veya üçlü tırnak fark etmez.

### ✏️ Örnek 6.1: Metnin tipini görelim

`"Merhaba"`, `'Merhaba'` ve `"25"` ifadelerinin tiplerini yazdır.

💡 **Beklenen çıktı:**
```text
<class 'str'>
<class 'str'>
<class 'str'>
```

### ✏️ Örnek 6.2: Metnin uzunluğu

`metin = "Python"` tanımla. `len(metin)` ile karakter sayısını yazdır.

💡 **Beklenen çıktı:**
```text
6
```

## 2. İndeks: Karakterin Adresi

Metindeki her karakterin bir **sıra numarası (indeks)** vardır. **Sayma 0'dan başlar!**

```text
 P   y   t   h   o   n
 0   1   2   3   4   5     ← soldan
-6  -5  -4  -3  -2  -1     ← sağdan
```

`metin[0]` → ilk karakter · `metin[-1]` → son karakter

### ✏️ Örnek 6.3: İlk ve son karakter

`metin = "Python Yazılım Dili Öğreniyorum"` tanımla. İlk karakteri, son karakteri ve sondan ikinci karakteri yazdır.

💡 **Beklenen çıktı:**
```text
P
m
u
```

### ✏️ Örnek 6.4: Var olmayan indeks

Aynı metinde `metin[100]` yazdırmayı dene.

> ⚠️ `IndexError` hatası alacaksın. Metin o kadar uzun değil!

## 3. Dilimleme: `metin[başlangıç : bitiş : adım]`

- **başlangıç** dahildir, **bitiş dahil değildir**.
- Boş bırakılan başlangıç = en baş, boş bırakılan bitiş = en son.
- **adım** kaçar kaçar ilerleneceğini gösterir. Negatif adım **geriye** doğru gider.

| Yazım | Anlamı |
|---|---|
| `metin[4:8]` | 4, 5, 6, 7. karakterler |
| `metin[:7]` | Baştan 7. indekse kadar |
| `metin[8:]` | 8. indeksten sona kadar |
| `metin[0:8:2]` | İlk 8 karakter, ikişer atlayarak |
| `metin[::-1]` | Tüm metin **tersten** |

> Aşağıdaki örneklerde **`metin = "Python Yazılım Dili Öğreniyorum"`** kullan. Önce bir hücrede tanımla.

### ✏️ Örnek 6.5: Aradan bir parça alalım

`metin[4:8]` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
on Y
```

### ✏️ Örnek 6.6: Baştan ve sondan parça

İlk 6 karakteri (`Python`) ve 7. indeksten sona kadar olan kısmı yazdır.

💡 **Beklenen çıktı:**
```text
Python
Yazılım Dili Öğreniyorum
```

### ✏️ Örnek 6.7: Atlayarak dilimleme

`metin[0:8:2]` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
Pto 
```

### ✏️ Örnek 6.8: Baş ve son karakter hariç

`metin[1:-1]` sonucunu yazdır.

### ✏️ Örnek 6.9: Metni tersten yazdıralım

`metin[::-1]` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
muroyinerğÖ iliD mılızaY nohtyP
```

### ✏️ Örnek 6.10: Gizli isim

`metin[8] + metin[17] + metin[18]` sonucunu yazdır. Hangi isim çıktı?

💡 **Beklenen çıktı:**
```text
ali
```

### 🎯 Kendini Dene 1: Kendi Adınla Dilimleme

Ad-soyadını bir değişkene ata. Sırasıyla: adının ilk harfini, sadece soyadını, tamamını tersten ve uzunluğunu yazdır.

### 🎯 Kendini Dene 2: Palindrom Dedektifi

`kelime = "kabak"` tanımla. Kelime tersten de aynı okunuyor mu? `==` ve `[::-1]` kullanarak sonucu yazdır. Sonra `"ankara"` ile dene.

💡 **Örnek çıktı:**
```text
kabak tersten de aynı mı? True
```

> 📌 **Not:** `upper()`, `replace()` gibi metin metotları **Listeler Modülü**nde ayrıca işlenecek.

## ✅ Bugün Öğrendiklerimiz
- Metindeki karakterler 0'dan başlayarak numaralanır, `-1` son karakterdir.
- `[başlangıç:bitiş:adım]` ile dilimleme yapılır, bitiş dahil değildir.
- `[::-1]` metni tersine çevirir, `len()` uzunluğu verir.
