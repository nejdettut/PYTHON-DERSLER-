## 1. Kelime Torbası (Bag of Words) 🛍️

Her metni, **sözlükteki her kelimenin kaç kez geçtiğini** gösteren bir sayı listesine çeviririz. Kelimelerin **sırası** önemsenmez, hepsi bir torbaya atılmış gibi.

```text
Cümle 1: "kedi süt içti"
Cümle 2: "köpek süt içmedi kedi kaçtı"

Sözlük:      içmedi içti kaçtı kedi köpek süt
Cümle 1  →  [  0     1     0     1    0     1 ]
Cümle 2  →  [  1     0     1     1    1     1 ]
```

### ✏️ Örnek 27.1: Elle kelime torbası

Yukarıdaki iki cümle için sözlüğü (`sorted(set(...))`) ve her cümlenin sayı vektörünü **kendin** oluştur.

💡 **Beklenen çıktı:**
```text
['içmedi', 'içti', 'kaçtı', 'kedi', 'köpek', 'süt']
[0, 1, 0, 1, 0, 1]
[1, 0, 1, 1, 1, 1]
```

## 2. scikit-learn ile: `CountVectorizer`

````python
cumleler = [
    "kedi bahçede fare kovaladı",
    "köpek bahçede top oynadı",
    "kedi ve köpek bahçede uyudu",
    "borsa bugün yükselişle kapandı",
    "dolar ve borsa düşüşle başladı",
]
print(len(cumleler), "cümle hazır.")
````

### ✏️ Örnek 27.2: Vektörleştir

`from sklearn.feature_extraction.text import CountVectorizer` ile `vek = CountVectorizer()` oluştur, `X = vek.fit_transform(cumleler)` yap. Sözlüğü (`vek.get_feature_names_out()`) ve `X.toarray()` tablosunu yazdır. Tabloyu pandas DataFrame'e çevirerek göster.

## 3. Metinler Ne Kadar Benzer? 📐

İki vektör arasındaki **kosinüs benzerliği** 0 ile 1 arasında bir sayıdır:
- **1** → aynı yönde (çok benzer)
- **0** → hiç ortak kelime yok

### ✏️ Örnek 27.3: Benzerlik matrisi

`from sklearn.metrics.pairwise import cosine_similarity` ile tüm cümleler arasındaki benzerliği hesapla, 2 basamağa yuvarlayıp yazdır. Hangi iki cümle en benzer? 1. ve 4. cümlenin benzerliği neden 0?

### ✏️ Örnek 27.4: Mini arama motoru 🔎

`ara(sorgu)` fonksiyonu yaz: Sorguyu aynı `vek` ile dönüştür (`vek.transform([sorgu])`), tüm cümlelerle benzerliğini hesapla ve en benzer cümleyi ve skorunu döndürsün. `"bahçede kedi"`, `"borsa ne oldu"` ve `"uzay roketi"` için dene.

💡 **Beklenen çıktı:**
```text
'bahçede kedi' → kedi bahçede fare kovaladı (0.71)
'borsa ne oldu' → borsa bugün yükselişle kapandı (0.5)
'uzay roketi' → sonuç bulunamadı
```

### ✏️ Örnek 27.5: Sıralı sonuçlar

`ara` fonksiyonunu ilk 3 sonucu benzerliğe göre sıralı döndürecek şekilde geliştir. İpucu: `sorted(..., reverse=True)`
