## 1. Dönüşüm Fonksiyonları

| Fonksiyon | Neye çevirir? | Örnek | Sonuç |
|---|---|---|---|
| `int()` | Tam sayıya | `int(4.9)` | `4` |
| `float()` | Ondalıklı sayıya | `float(4)` | `4.0` |
| `str()` | Metne | `str(36)` | `"36"` |
| `bool()` | True / False | `bool(0)` | `False` |

> ⚠️ `int()` **yuvarlamaz, keser**: `int(4.9)` → `4`
> ⚠️ Her şey her şeye dönüşmez: `int("elma")` hata verir.

### ✏️ Örnek 7.1: Float'tan int'e

`a = 4.9` tanımla. `a`'yı, tipini, `int(a)`'yı ve onun tipini yazdır.

💡 **Beklenen çıktı:**
```text
4.9
<class 'float'>
4
<class 'int'>
```

### ✏️ Örnek 7.2: Int'ten float'a

`c = 56` tanımla. `float(c)` sonucunu ve tipini yazdır.

💡 **Beklenen çıktı:**
```text
56.0
<class 'float'>
```

### ✏️ Örnek 7.3: Sayıyı metne çevirelim

`str(3) + str(6)` sonucunu yazdır. Sonuç neden 9 değil?

💡 **Beklenen çıktı:**
```text
36
```

### ✏️ Örnek 7.4: Metni sayıya çevirip işlem yapalım

`a = str(3) + str(6)` tanımla. Önce `int(a) + 5`, sonra `a + str(5)` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
41
365
```

### ✏️ Örnek 7.5: Aynı sayılar, farklı tipler

`sayi1 = "5"`, `sayi2 = "6"` ile toplama ve `* 3` işlemi yap. Sonra tırnaksız (`5` ve `6`) aynısını yap. Karşılaştır.

### ✏️ Örnek 7.6: Hangi dönüşüm hata verir?

Sırayla dene: `int("36")`, `int("elma")`, `int("4.5")`, `float("4.5")`. Hata verenleri not al.

> ⚠️ Bazı satırlar `ValueError` verecek. Her birini **ayrı hücrede** dene.

### ✏️ Örnek 7.7: Bool'u dönüştürelim

`yas18_buyuk_mu = True` tanımla. `int()` ve `str()` ile dönüştürüp yazdır. Sonra `False` ile tekrarla.

💡 **Beklenen çıktı:**
```text
1
True
```

## 2. Problem: Tren Saat Kaçta Varır?

> **Soru:** Saat **12:00**'de İstanbul'dan kalkan tren **255 dakika** sonra Ankara'ya varıyor. Tren Ankara'ya saat kaçta varır?

Önce algoritmayı düşün:
1. 255 dakikanın kaç **tam saat** olduğunu bul → `//`
2. Artan **dakikayı** bul → `%`
3. Kalkış saatine ekle

### ✏️ Örnek 7.8: 1. yol: `//` ve `%` ile çözelim

`kalkis_saati`, `yolculuk_dakikasi` değişkenlerini kullan.

💡 **Beklenen çıktı:**
```text
Varış saati: 16 : 15
```

### `divmod()` Fonksiyonu

`divmod(a, b)` → bölümü ve kalanı **birlikte** verir.

```text
divmod(255, 60)   →   (4, 15)
saat, dakika = divmod(255, 60)
```

### ✏️ Örnek 7.9: `divmod()` deneyelim

`divmod(17, 5)` ve `divmod(255, 60)` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
(3, 2)
(4, 15)
```

### ✏️ Örnek 7.10: 2. yol: `divmod()` ile çözelim

Kalkış saatini dakikaya çevir (`12 * 60`), yolculuk süresini ekle, sonra `divmod(toplam, 60)` ile saat ve dakikayı bul.

💡 **Beklenen çıktı:**
```text
Tren Ankara'ya saat 16:15'te varır.
```

💡 **Örnek çıktı:**
