## 1. Modül Nedir?

Python'da her `.py` dosyası bir **modüldür**. İçinde fonksiyonlar ve değişkenler bulunur; başka programlara **eklenerek** (import) kullanılır.

**Avantajları:** Kod tekrarını önler · Projeyi düzenli tutar · Bakımı ve güncellemesi kolaylaşır.

| Tür | Örnek |
|---|---|
| Hazır (standart) modüller | `math`, `random`, `time` |
| Kendi yazdığımız modüller | `selam.py`, `hesap.py` |

## 2. Defterde Modül Dosyası Oluşturmak

Hücrenin **en üst satırına** `%%writefile dosya_adi.py` yazarsak, hücrenin geri kalanı o isimde bir dosyaya kaydedilir.

```text
%%writefile selam.py
def cagir():
    print("Merhaba Python Grubu")
```

> 📌 Çalıştırınca `Writing selam.py` mesajı görürsün. Dosya defterle **aynı klasöre** kaydedilir.

### ✏️ Örnek 36.1: İlk modülümüz

Yukarıdaki gibi `selam.py` dosyasını oluştur. İçine `cagir()` fonksiyonunu yaz.

💡 **Beklenen çıktı:**
```text
Writing selam.py
```

### ✏️ Örnek 36.2: Modülü kullan

`import selam` ile modülü ekle ve `selam.cagir()` fonksiyonunu çağır. Sonra `print(dir(selam))` ile içini incele.

💡 **Beklenen çıktı:**
```text
Merhaba Python Grubu
```

### ✏️ Örnek 36.3: Değişken içeren modül

`bilgi.py` adında bir modül oluştur; içinde `ad`, `soyad`, `yas`, `meslek` değişkenleri olsun.

💡 **Beklenen çıktı:**
```text
Writing bilgi.py
```

## 3. import Çeşitleri

| Yazım | Kullanım | Ne zaman? |
|---|---|---|
| `import bilgi` | `bilgi.ad` | Varsayılan, en anlaşılır yol |
| `import bilgi as b` | `b.ad` | Modül adı uzunsa (takma ad) |
| `from bilgi import meslek` | `meslek` | Sadece bir şey lazımsa |
| `from bilgi import *` | `ad`, `soyad`... | Hepsini doğrudan (dikkatli kullan!) |

### ✏️ Örnek 36.4: import modul

`import bilgi` ile tüm bilgileri `Adı: ...` şeklinde yazdır.

💡 **Beklenen çıktı:**
```text
Adı: Nejdet
Soyadı: Tut
Yaşı: 36
Mesleği: Öğretmen
```

### ✏️ Örnek 36.5: from ... import

Sadece `meslek` değişkenini al ve yazdır. Sonra aynı hücrede `ad`'ı yazdırmayı dene.

> ⚠️ `ad` için `NameError` alacaksın; sadece `meslek` alındı.

### ✏️ Örnek 36.6: Takma ad

`import bilgi as b` ile aynı bilgileri yazdır.

### ✏️ Örnek 36.7: Hesap modülü

`hesap.py` modülünde `topla(a, b)`, `carp(a, b)` ve `kare(x)` fonksiyonlarını yaz. Başka bir hücrede `from hesap import topla, kare` ile ikisini kullan.

💡 **Beklenen çıktı:**
```text
Writing hesap.py
8
49
```

## ✅ Bugün Öğrendiklerimiz
- Her `.py` dosyası bir modüldür; `import` ile başka programa eklenir.
- `%%writefile ad.py` ile defterde modül dosyası oluşturulur.
- `import m`, `import m as x`, `from m import f`, `from m import *` farklı ekleme yollarıdır.
