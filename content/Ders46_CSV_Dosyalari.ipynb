## 1. CSV Nedir?

**CSV** (Comma Separated Values = Virgülle Ayrılmış Değerler), tabloları düz metin olarak saklayan bir biçimdir. Excel ve Google E-Tablolar CSV açabilir; yapay zekâ ve veri biliminde en çok kullanılan dosya türlerinden biridir.

```text
ad,sinif,not
Ali,9-A,80
Ayşe,9-A,95
```

İlk satır genelde **başlık** satırıdır.

| Yöntem | Ne yapar? |
|---|---|
| `csv.writer(dosya)` + `writerow([...])` | Bir satır yazar |
| `writerows([[...], [...]])` | Çok satır yazar |
| `csv.reader(dosya)` | Her satırı **liste** olarak verir |
| `csv.DictReader(dosya)` | Her satırı **sözlük** olarak verir |

> 🔹 CSV açarken `newline=""` ve `encoding="utf-8"` yazmak boş satır ve Türkçe karakter sorunlarını önler.

### ✏️ Örnek 46.1: CSV yaz

`import csv` ile `notlar.csv` dosyası oluştur. Başlık satırı `ad, sinif, not` olsun ve en az 4 öğrenci ekle (`writerows`).

### ✏️ Örnek 46.2: CSV oku (liste)

`csv.reader` ile dosyayı oku ve her satırı yazdır.

💡 **Beklenen çıktı:**
```text
['ad', 'sinif', 'not']
['Ali', '9-A', '80']
...
```

### ✏️ Örnek 46.3: CSV oku (sözlük)

`csv.DictReader` ile oku ve her öğrenciyi `Ali (9-A): 80` şeklinde yazdır.

💡 **Beklenen çıktı:**
```text
Ali (9-A): 80
Ayşe (9-A): 95
...
```

## 2. Veri Üzerinde Hesaplama

### ✏️ Örnek 46.4: Sınıf istatistiği

`notlar.csv`'den notları oku (`int()`!), ortalamayı, en yüksek notu ve o notu alan öğrenciyi bul.

## 3. pandas'a İlk Bakış 🐼

`pandas`, veri biliminin en popüler kütüphanesidir. CSV'yi tek satırda tablo olarak açar:

```text
import pandas as pd
tablo = pd.read_csv("notlar.csv")
```
