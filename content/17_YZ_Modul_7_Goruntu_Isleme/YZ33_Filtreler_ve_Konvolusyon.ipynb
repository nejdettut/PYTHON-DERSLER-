## 1. Kayan Pencere 🔍

**Konvolüsyon**, küçük bir sayı tablosunun (**filtre/çekirdek**, genelde 3×3) resmin üzerinde **kaydırılmasıdır**. Her konumda:
1. Filtre ile altındaki 3×3 piksel **eleman eleman çarpılır**,
2. Sonuçlar **toplanır**,
3. Toplam, yeni resimde o konumun değeri olur.

```text
Resim parçası     Filtre (ortalama)        Sonuç
[10 20 30]        [1/9 1/9 1/9]
[40 50 60]   ×    [1/9 1/9 1/9]    →   (10+20+...+90)/9 = 50
[70 80 90]        [1/9 1/9 1/9]
```

| Filtre | Etkisi |
|---|---|
| Ortalama (hepsi 1/9) | **Bulanıklaştırma** |
| Dikey kenar `[[-1,0,1],[-2,0,2],[-1,0,1]]` | **Dikey kenarları** bulur (Sobel) |
| Keskinleştirme `[[0,-1,0],[-1,5,-1],[0,-1,0]]` | Detayları **belirginleştirir** |

````python
import numpy as np
import matplotlib.pyplot as plt

# 40x40 gri test resmi: ortada beyaz bir kare ve bir çapraz çizgi
test = np.zeros((40, 40))
test[10:30, 10:30] = 255
for i in range(40):
    test[i, i] = 180
plt.imshow(test, cmap="gray")
plt.title("Test resmi")
plt.show()
````

## 2. Evrişimli Sinir Ağları (CNN) 🧠🖼️

YZ-24'te rakamları düzleştirip ağa vermiştik; ağ pikseller arasındaki **komşuluğu** bilmiyordu.

**CNN** (Convolutional Neural Network) ise:
1. İlk katmanlarda yüzlerce **filtre** uygular: Ama filtre sayılarını **biz yazmayız**, ağ **öğrenir**!
2. İlk katmanlar kenar ve köşe, sonrakiler göz, tekerlek gibi parçalar, son katmanlar yüz, araba gibi nesneler bulur.

```text
Resim → [Filtreler: kenarlar] → [Filtreler: şekiller] → [Filtreler: parçalar] → Sınıf: "kedi"
```
