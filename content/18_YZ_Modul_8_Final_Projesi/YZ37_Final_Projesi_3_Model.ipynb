Örnek projenin verisini hazırla:

```python
# Final projesi örnek veri seti: 300 öğrencinin bilgileri (hayalî). Sadece çalıştır!
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 300
calisma = rng.integers(0, 15, n)                 # haftalık çalışma saati
devamsizlik = rng.integers(0, 20, n)             # gün
uyku = rng.integers(5, 10, n)                    # saat
ekran = rng.integers(1, 9, n)                    # günlük ekran süresi (saat)
odev = rng.integers(0, 101, n)                   # ödev tamamlama yüzdesi
puan = (35 + 3.2 * calisma - 1.1 * devamsizlik + 2.0 * (uyku - 5) - 1.5 * ekran
        + 0.15 * odev + rng.normal(0, 7, n)).clip(0, 100).round()

proje = pd.DataFrame({"calisma_saati": calisma, "devamsizlik": devamsizlik, "uyku_saati": uyku,
                      "ekran_suresi": ekran, "odev_yuzdesi": odev, "yil_sonu_puani": puan})
proje.loc[rng.choice(n, 6, replace=False), "uyku_saati"] = np.nan
proje.to_csv("ogrenci_basari.csv", index=False)
print("ogrenci_basari.csv hazır!", proje.shape)

from sklearn.model_selection import train_test_split
df = pd.read_csv("ogrenci_basari.csv")
df["uyku_saati"] = df["uyku_saati"].fillna(df["uyku_saati"].median())
df["gecti"] = (df["yil_sonu_puani"] >= 50).astype(int)
X = df[["calisma_saati", "devamsizlik", "uyku_saati", "ekran_suresi", "odev_yuzdesi"]]
Xe, Xt, ye, yt = train_test_split(X, df["yil_sonu_puani"], test_size=0.25, random_state=0)
Xe2, Xt2, ye2, yt2 = train_test_split(X, df["gecti"], test_size=0.25, random_state=0, stratify=df["gecti"])
print("Hazır.")
```
