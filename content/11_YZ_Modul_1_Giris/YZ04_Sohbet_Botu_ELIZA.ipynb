## 1. ELIZA Nasıl Çalışıyordu?

1966'da Joseph Weizenbaum'un yazdığı **ELIZA** bir psikolog gibi konuşurdu. Ama aslında hiçbir şey **anlamıyordu**. Mesajda **anahtar kelime** arar, hazır cevaplardan birini verirdi:

```text
Kullanıcı: Annemle tartıştım.
ELIZA:     Bana ailenden biraz daha bahset.     ← "anne" kelimesini yakaladı
```

Birçok kullanıcı ELIZA'nın kendisini gerçekten anladığına inandı!

## 2. Botumuzun Planı

| Adım | Python bilgisi |
|---|---|
| Anahtar kelime → cevap listesi | Sözlük (Ders 32) |
| Mesajı küçük harfe çevir | `lower()` (Ders 30) |
| Kelime mesajda geçiyor mu? | `in` (Ders 16) |
| Cevaplardan birini seç | `random.choice` (Ders 38) |
| Kullanıcı "görüşürüz" diyene kadar sür | `while True` + `break` (Ders 18) |

### ✏️ Örnek 4.1: Cevap sözlüğü

Anahtarları kelime, değerleri **cevap listesi** olan `cevaplar` sözlüğünü oluştur. En az 5 anahtar kelime olsun: `merhaba`, `okul`, `anne`, `üzgün`, `oyun`…

### ✏️ Örnek 4.2: Cevap bulan fonksiyon

`cevap_ver(mesaj)` fonksiyonu mesajı küçük harfe çevirsin, sözlükteki anahtar kelimelerden biri geçiyorsa o kelimenin cevaplarından **rastgele** birini döndürsün. Hiçbiri yoksa genel bir cevap döndürsün (`"Bunu biraz açar mısın?"`).

💡 **Beklenen çıktı:**
```text
cevap_ver("Bugün okul çok yorucuydu") → Okulda en sevdiğin ders hangisi?
```

### ✏️ Örnek 4.3: Sohbet döngüsü

Botun adı `ELİZA` olsun. Kullanıcı `görüşürüz` yazana kadar mesaj alıp cevap versin.

💡 **Beklenen çıktı:**
```text
ELİZA: Merhaba! Ben ELİZA. Bugün nasılsın?
Sen: biraz üzgünüm
ELİZA: Neden üzgün hissediyorsun?
Sen: görüşürüz
ELİZA: Görüşmek üzere, kendine iyi bak!
```

### ✏️ Örnek 4.4: Botu geliştir

Botuna şu özelliklerden en az ikisini ekle:
- Kullanıcının adını sor, cevaplarda kullan
- `saat kaç` sorusuna gerçek saati söyle (`time`)
- Konuşulan mesaj sayısını tut, çıkışta söyle
- Mesaj soru işaretiyle bitiyorsa farklı bir genel cevap ver
