## 1. Sınıf ve Nesne

**Sınıf (class)**, nesneler üretmek için kullanılan bir **kalıp / şablondur**. **Nesne (object)** ise bu kalıptan üretilmiş gerçek bir örnektir.

| Sınıf (kalıp) | Nesneler (örnekler) |
|---|---|
| Araba | Kamyon, taksi, otobüs |
| Öğrenci | Ali, Ayşe, Can |
| Kurabiye kalıbı 🍪 | Fırından çıkan her kurabiye |

Her nesnenin:
- **Özellikleri** (değişkenler) vardır: renk, yaş, ad…
- **Davranışları** (metotlar) vardır: çalış, selamla, hızlan…

> 🔹 Sürpriz: Python'da her şey bir nesnedir! `int`, `str`, `list` aslında birer sınıftır. `"ali".upper()` yazarken bir **str nesnesinin** metodunu çağırıyordun.

### ✏️ Örnek 48.1: Python'daki sınıflar

`type(5)`, `type("ali")`, `type([1, 2])` ve `type(int)` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
<class 'int'>
<class 'str'>
<class 'list'>
<class 'type'>
```

## 2. `__init__` ve `self`

```text
class Kisi:
    def __init__(self, isim, yas):   # nesne oluşturulurken OTOMATİK çalışır
        self.isim = isim             # nesnenin özelliği
        self.yas = yas

ali = Kisi("Ali", 15)                # __init__ çalışır
print(ali.isim)                      # Ali
```

- **`__init__`** (yapıcı / constructor): Nesne oluşturulduğu anda **kendiliğinden** çalışır, başlangıç değerlerini verir.
- **`self`**: "Bu nesnenin kendisi" demektir. Sınıftaki her metodun **ilk parametresi** `self`'tir ama çağırırken yazılmaz.

### ✏️ Örnek 48.3: Yapıcı çalışıyor mu?

`__init__` içinde sadece `print("Yeni nesne oluşturuldu!")` yazan bir sınıf tanımla ve 2 nesne türet.

💡 **Beklenen çıktı:**
```text
Yeni nesne oluşturuldu!
Yeni nesne oluşturuldu!
```

### ✏️ Örnek 48.4: self'i unutursak

> ⚠️ `TypeError: __init__() takes 0 positional arguments but 1 was given`. Python `self`'i otomatik gönderir!

## 3. Metotlar

Sınıfın içindeki fonksiyonlara **metot** denir. Metotlar `self` sayesinde nesnenin özelliklerine ulaşır.

### ✏️ Örnek 48.6: selamla() metodu

`Kisi` sınıfına `selamla()` metodu ekle: `Merhaba, ben Ahmet. Yaşım 30.` yazsın. İki nesne için çağır.

💡 **Beklenen çıktı:**
```text
Merhaba, ben Ahmet. Yaşım 30.
Merhaba, ben Ayşe. Yaşım 25.
```

### ✏️ Örnek 48.7: Metin yazdırıcı

`MetinSinifi(ifade, tekrar, ayrac="\n")`: `__init__` içinde ifadeyi `tekrar` kez, aralarında `ayrac` olacak şekilde yazdırsın. Bir kez 2 argümanla, bir kez 3 argümanla nesne türet.
