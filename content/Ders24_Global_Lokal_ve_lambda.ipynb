## 1. Global ve Lokal Değişkenler

| Tür | Nerede tanımlanır? | Nerede geçerli? |
|---|---|---|
| **Lokal (yerel)** | Fonksiyonun **içinde** | Sadece o fonksiyonda |
| **Global (genel)** | Fonksiyonun **dışında** | Programın her yerinde |

> 🔹 Fonksiyon içindeki değişken, fonksiyon bitince **yok olur**. Dışarıya bilgi taşımanın doğru yolu `return`'dür.
> 🔹 Mecbur kalırsan, fonksiyon içinde `global degisken` yazarak onu global yapabilirsin.

### ✏️ Örnek 24.1: Lokal değişkene dışarıdan ulaşmak

`toplam_bul()` fonksiyonu 1'den 10'a kadar sayıları `topla` değişkeninde toplasın. Fonksiyonu çağır, sonra dışarıda `print(topla)` yaz.

> ⚠️ `NameError` alacaksın. `topla` sadece fonksiyonun içinde yaşıyor.

### ✏️ Örnek 24.2: Doğru yol: return

Aynı fonksiyonu `return` ile düzelt.

💡 **Beklenen çıktı:**
```text
Toplam: 55
```

### ✏️ Örnek 24.3: `global` ile

Bu sefer fonksiyonun içinde `global toplam` yazarak çöz.

💡 **Beklenen çıktı:**
```text
55
```

### 🐞 Hatayı Bul

```text
def carp():
    a = 2
    b = 3
    print("c:", c)
    return a * b

c = 5
print("a*b =", carp())
print("a:", a)
```

Bu kod hangi satırda hata verir? `c` neden sorun çıkarmıyor da `a` çıkarıyor? Kodu düzeltip aşağıya yaz.

### ✏️ Örnek 24.4: Hatayı düzelt

Yukarıdaki kodu `a` değeri de yazdırılabilecek şekilde düzelt.

💡 **Beklenen çıktı:**
```text
c: 5
a*b = 6
a: 2
```

## 2. `lambda` Fonksiyonları

**Tek satırlık, isimsiz** küçük fonksiyonlardır.

```text
lambda parametreler: döndürülecek_ifade
```

| `def` ile | `lambda` ile |
|---|---|
| `def kare(x):`<br>`    return x ** 2` | `kare = lambda x: x ** 2` |

> 🔹 `lambda` içinde `return` yazılmaz; ifadenin sonucu otomatik döner.
> 🔹 Kısa hesaplamalar için idealdir. Karmaşık işler için `def` kullan.

### ✏️ Örnek 24.5: Kare alan lambda

`kare = lambda x: x ** 2` tanımla ve `kare(5)` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
25
```

### ✏️ Örnek 24.6: İki parametreli lambda

İki sayıyı toplayan `topla` lambda'sını yaz, `topla(3, 4)` ile dene.

💡 **Beklenen çıktı:**
```text
7
```

### ✏️ Örnek 24.7: Koşullu lambda

`tek_cift = lambda x: "Çift" if x % 2 == 0 else "Tek"` tanımla. 5, 10 ve 7 için dene.

💡 **Beklenen çıktı:**
```text
5 için durum: Tek
10 için durum: Çift
7 için durum: Tek
```

### ✏️ Örnek 24.8: map() önizlemesi

`sayilar = [1, 2, 3, 4]` listesindeki her sayının karesini `list(map(lambda x: x ** 2, sayilar))` ile bul.

📌 `map()` ve listeler Modül 6'da ayrıntılı işlenecek.

💡 **Beklenen çıktı:**
```text
[1, 4, 9, 16]
```

### 🎯 Kendini Dene 1: lambda ile Dönüştürücüler

Üç lambda yaz: `km_mil` (km × 0.62), `c_f` (C × 1.8 + 32), `kdvli` (fiyat × 1.2). Her birini bir değerle dene ve sonuçları 2 basamakla yazdır.

💡 **Örnek çıktı:**
```text
10 km = 6.20 mil
25 °C = 77.00 °F
100 TL KDV'li = 120.00 TL
```
