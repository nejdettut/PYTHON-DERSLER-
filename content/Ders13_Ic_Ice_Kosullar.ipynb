## 1. İç İçe Koşullar

Bir `if` bloğunun **içine** başka bir `if`, `if – else` veya `if – elif – else` yazılabilir.

- Önce **dış koşul** kontrol edilir.
- Dış koşul sağlanırsa **iç koşula** geçilir.
- Dış koşul sağlanmazsa iç koşullar **hiç kontrol edilmez**.

```text
if dış_koşul:
    if iç_koşul:
        ...
    else:
        ...
else:
    ...
```

> 🔹 Her yeni seviye **4 boşluk daha** içeriden başlar. Girintiye dikkat!

### ✏️ Örnek 13.1: Harf notu (hatalı veri kontrollü)

0–100 arası not al. Önce **dış koşulda** notun bu aralıkta olup olmadığını kontrol et. Aralıktaysa:

85+ → AA · 70+ → BB · 55+ → CC · 40+ → DD · diğer → FF

Aralık dışındaysa `Hatalı veri girdiniz` yazdır.

💡 **Beklenen çıktı:**
```text
Notunuzu giriniz (0-100): 120
Hatalı veri girdiniz
```

### ✏️ Örnek 13.2: Adım adım giriş

Önce kullanıcı adını sor. Kullanıcı adı `admin` ise **ancak o zaman** şifreyi sor. Şifre `123456` ise giriş başarılı, değilse şifre hatalı. Kullanıcı adı yanlışsa şifre hiç sorulmasın.

💡 **Beklenen çıktı:**
```text
Kullanıcı adını giriniz: admin
Parolayı giriniz: 123456
Giriş başarılı
```

### ✏️ Örnek 13.3: Üçgenin türü (kenarlara göre)

Üç kenar uzunluğu al. Kenarlardan biri 0 veya negatifse `Hatalı veri`. Değilse: üçü eşitse eşkenar, ikisi eşitse ikizkenar, hiçbiri eşit değilse çeşitkenar.

İpucu: Python'da `a == b == c` yazabilirsin.

💡 **Beklenen çıktı:**
```text
Birinci kenar: 5
İkinci kenar: 5
Üçüncü kenar: 8
İkizkenar Üçgen
```

### ✏️ Örnek 13.4: Üçgenin türü (açılara göre)

Üç iç açı al. Açıların toplamı 180 değilse `Hatalı giriş`. 180 ise: hepsi 60 → eşkenar, ikisi eşit → ikizkenar, diğer → çeşitkenar.

💡 **Beklenen çıktı:**
```text
Birinci açı: 50
İkinci açı: 50
Üçüncü açı: 80
İkizkenar üçgen
```

## 2. Karar Ağacı Oyunu

Aşağıdaki elektronik aletlerden birini aklında tut: **Telefon, Tablet, Televizyon, Kamera, Akıllı Saat**

Bilgisayar sorular sorar, sen **E** (Evet) veya **H** (Hayır) cevabı verirsin. Bilgisayar aletini tahmin eder.

WhatsApp Image 2023-07-25 at 14.02.31.jpeg

### ✏️ Örnek 13.5: Elektronik alet tahmin oyunu

Aşağıdaki soru sırasını kullan:

1. Ekranı var mı? → H ise `Bulamadım`
2. Taşınabilir mi? → H ise **Televizyon**
3. Kola takılabilir mi? → E ise **Akıllı Saat**
4. Konuşmak için kullanılır mı? → E ise **Telefon**
5. Dokunmatik ekranı var mı? → E ise **Tablet**, H ise **Kamera**

İpucu: `cvp.lower() == "e"` yazarsan `E` ve `e` aynı kabul edilir.

💡 **Beklenen çıktı:**
```text
Ekranı var mı? [E/H]: e
Taşınabilir mi? [E/H]: e
Kola takılabilir mi? [E/H]: h
Konuşmak için kullanılır mı? [E/H]: e
Cevap: Telefon!
```

## 📝 Ödev 5 – İç İçe Koşullar

**Hayvanlar Alemi** karar ağacı oyununu iç içe koşullar kullanarak yazınız. Oyun şu hayvanları bulabilmeli: **Köpek, Kedi, Balık, Kuş**.

Örnek soru akışı: Memeli mi? → Havlıyor mu? → … / Suda yaşıyor mu? → Uçabiliyor mu? → …

💡 **Örnek çıktı:**
```text
Memeli bir hayvan mı? [E/H]: h
Suda yaşıyor mu? [E/H]: e
Cevap: Balık!
```
