## Proje Planı 🏨

Bir restoranın rezervasyonlarını `rezervasyon.txt` dosyasında tutan bir program yazacağız. Her satır bir rezervasyon (isim) olacak. Kapasite: **5** kişi.

| Menü | Fonksiyon | Görev |
|---|---|---|
| 1 | `rezervasyon_yap()` | Kapasite doluysa uyar, değilse ismi dosyaya **ekle** |
| 2 | `rezervasyon_kontrol()` | İsim dosyada var mı? |
| 3 | `rezervasyon_listele()` | Numaralı liste + kalan kapasite |
| 4 | `rezervasyon_sil()` | Numarası verilen kaydı sil |
| 5 | — | Çıkış |

Hepsinin ihtiyaç duyduğu ortak bir yardımcı fonksiyonla başlayalım.

### ✏️ Örnek 47.1: Yardımcı fonksiyon: kayıtları oku

`kayitlari_oku()` fonksiyonu dosyadaki isimleri **liste** olarak döndürsün (`strip()` ile). Dosya henüz yoksa (`FileNotFoundError`) **boş liste** döndürsün.

### ✏️ Örnek 47.2: Rezervasyon yap

`rezervasyon_yap()`: Kayıt sayısı kapasiteye ulaştıysa `Kapasite dolu!` yaz. Değilse isim al, `"a"` kipinde dosyaya ekle ve rezervasyon numarasını söyle.

### ✏️ Örnek 47.3: Kontrol ve listeleme

`rezervasyon_kontrol()` ve `rezervasyon_listele()` fonksiyonlarını yaz. Listelerken `enumerate` kullan ve kalan kapasiteyi göster.

### ✏️ Örnek 47.4: Silme

`rezervasyon_sil()`: Numara al (`try` ile!). Geçerliyse kaydı listeden `pop` ile çıkar ve **tüm listeyi** `"w"` kipinde dosyaya yeniden yaz.

### ✏️ Örnek 47.5: Ana menü

`while True` ile menüyü göster, seçime göre ilgili fonksiyonu çağır. 5 seçilince çık.

💡 **Beklenen çıktı:**
```text
1-Yap 2-Kontrol 3-Listele 4-Sil 5-Çıkış: 1
Ad Soyad: Ali Kaya
Rezervasyonunuz alındı. No: 1
1-Yap 2-Kontrol 3-Listele 4-Sil 5-Çıkış: 3
1. Ali Kaya
Kalan kapasite: 4
1-Yap 2-Kontrol 3-Listele 4-Sil 5-Çıkış: 5
Çıkış yapıldı...
```

### 🎯 Kendini Dene 1: Geliştir!

Projeye en az bir özellik ekle: aynı isimle ikinci rezervasyonu engelle **veya** rezervasyonla birlikte kişi sayısı ve saat de kaydet (CSV kullanarak!).

## 🏁 Modül 10 Kontrol Listesi

- [ ] `open()` ile dosyayı doğru kipte (`r`, `w`, `a`) açabiliyorum

## ✅ Bugün Öğrendiklerimiz
- Büyük programlar küçük fonksiyonlara bölünerek yazılır.
- Dosya yoksa oluşan hata `try–except` ile güvenle yönetilir.
- Kayıt silmek için: tümünü oku → listeden çıkar → tümünü yeniden yaz.
