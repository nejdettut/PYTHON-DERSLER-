## 4. İlk Kodlarımız: `print()`

`print()` fonksiyonu parantez içine yazılanı **ekrana yazdırır**. Metinler tırnak `" "` içinde yazılır.

### ✏️ Örnek 1.1: Ekrana "Merhaba Dünya" yazdıralım

💡 **Beklenen çıktı:**
```text
Merhaba Dünya
```

### ✏️ Örnek 1.2: Kendi adını ve sınıfını ekrana yazdıralım

İki ayrı `print()` satırı kullan.

💡 **Beklenen çıktı:**
```text
Adım: Ayşe
Sınıfım: 9-A
```

### ✏️ Örnek 1.3: Bilerek hata yapalım

`print` yerine **`Print`** (büyük P) yazıp çalıştır. Sonra tırnağı kapatmayı unut ve tekrar çalıştır.

> ⚠️ Kod hata verecek. Hata mesajının **son satırını** oku: Python sana neyin yanlış olduğunu söylüyor.

> 🔹 Python **büyük-küçük harfe duyarlıdır**: `print` ile `Print` farklı şeylerdir.
> 🔹 Hata mesajları düşmanın değil, **yol göstericindir**.

## 5. Yorum (Açıklama) Satırları

`#` ile başlayan satırları Python **çalıştırmaz**. Bunlar yazılımcının kendine veya başkasına bıraktığı notlardır.

Ne işe yarar?
- Kodun ne yaptığını açıklamak
- Bir satırı geçici olarak devre dışı bırakmak

Birden fazla satırlık açıklama için **üçlü tırnak** `""" ... """` kullanılabilir.

### ✏️ Örnek 1.4: Tek satırlık yorum yazalım

En üste `# İlk programım` yaz. Altına bir `print()` ekle ve satırın sonuna da bir yorum yaz.

💡 **Beklenen çıktı:**
```text
Python öğreniyorum
```

### ✏️ Örnek 1.5: Bir satırı devre dışı bırakalım

Üç ayrı `print()` satırı yaz. Sonra ortadakinin başına `#` koyup tekrar çalıştır. Ne değişti?

### ✏️ Örnek 1.6: Çok satırlı açıklama yazalım

En üste üçlü tırnakla 3 satırlık bir açıklama yaz, altına bir `print()` ekle.

### 🎯 Kendini Dene 1: Künye Kartı

Adını, sınıfını, okulunu ve en sevdiğin dersi **dört ayrı satırda** yazdır. Her satırın sonuna o satırın ne yaptığını anlatan bir yorum ekle.

💡 **Örnek çıktı:**
```text
Ad: Ayşe
Sınıf: 9-A
Okul: ...
Sevdiğim ders: Bilişim
```
