## 1. Yapay Zekânın Alt Dalları

```text
Yapay Zekâ
 └── Makine Öğrenmesi        (veriden öğrenme)
      └── Derin Öğrenme       (yapay sinir ağları)
Yapay Zekânın uygulama alanları:
 ├── Doğal Dil İşleme        (metin, konuşma)
 ├── Görüntü İşleme          (resim, video)
 └── Veri Bilimi             (veriden anlam çıkarma)
```

| Alt dal | Ne yapar? | Örnek |
|---|---|---|
| **Makine Öğrenmesi** | Açıkça programlanmadan veriden öğrenir | Spam filtresi |
| **Derin Öğrenme** | Beyindeki nöronlardan esinlenen sinir ağları kullanır | Yüz tanıma |

| **Görüntü İşleme** | Resim ve videoları anlar | Otonom araçta trafik işareti tanıma |
| **Veri Bilimi** | Veriden anlamlı bilgi çıkarır | Satış tahmini |

## 3. Temel Kavramlar

| Kavram | Anlamı | Benzetme |
|---|---|---|
| **Veri** | Öğrenme malzemesi | Ders kitabı |
| **Özellik (feature)** | Verinin ölçülen bilgileri | Çalışma saati, devamsızlık |
| **Etiket (label)** | Tahmin edilmek istenen cevap | Geçti / kaldı |
| **Eğitim (training)** | Modelin örneklerden öğrenmesi | Konuyu çalışmak |
| **Model** | Eğitimden çıkan "öğrenilmiş kurallar" | Öğrencinin kafasındaki bilgi |
| **Test** | Modelin **görmediği** verilerle sınanması | Sınav |

> ⚠️ Model, eğitimde gördüğü sorularla test edilirse gerçek başarısı ölçülmez. Sınav soruları ders kitabındakiyle birebir aynıysa ezberleyen de geçer!

## 4. İlk Modelimiz: Geçer mi, Kalır mı? 🎓

Elimizde geçen yılın öğrencilerinin **haftalık çalışma saati** ve **sonuçları** var. Bu veriden bir **eşik değeri** öğrenip yeni öğrencileri tahmin edeceğiz.

Önce şu hazır hücreyi çalıştır:

````python
# Eğitim verisi: (haftalık çalışma saati, sonuç)
egitim = [(1, "kaldı"), (2, "kaldı"), (2, "kaldı"), (3, "kaldı"), (4, "geçti"),
          (3, "geçti"), (5, "geçti"), (6, "geçti"), (7, "geçti"), (1, "kaldı")]

# Test verisi: model bunları eğitimde GÖRMEYECEK
test = [(2, "kaldı"), (6, "geçti"), (4, "geçti"), (3, "kaldı")]
print("Veri hazır:", len(egitim), "eğitim,", len(test), "test örneği")
````

### ✏️ Örnek 2.2: Veriyi incele

`egitim` listesindeki geçenlerin ve kalanların çalışma saatlerini iki ayrı listeye ayır, ikisinin ortalamasını yazdır.

💡 **Beklenen çıktı:**
```text
Geçenlerin ortalaması: 5.0
Kalanların ortalaması: 1.8
```

### ✏️ Örnek 2.3: Modeli eğit

Eşik değeri iki ortalamanın tam ortası olsun: `esik = (gecen_ort + kalan_ort) / 2`. Bu değer, modelimizin **öğrendiği kuraldır**.

💡 **Beklenen çıktı:**
```text
Öğrenilen eşik: 3.4 saat
```

### ✏️ Örnek 2.4: Tahmin fonksiyonu

`tahmin_et(saat)` fonksiyonu saat eşikten büyük veya eşitse `"geçti"`, değilse `"kaldı"` döndürsün. 0, 3 ve 8 saat için dene.

### ✏️ Örnek 2.5: Modeli test et

`test` listesindeki her öğrenci için tahmin yap, gerçek sonuçla karşılaştır. Doğruluk oranını hesapla: `doğru tahmin sayısı / test sayısı`.

💡 **Beklenen çıktı:**
```text
2 saat → tahmin: kaldı, gerçek: kaldı ✓
6 saat → tahmin: geçti, gerçek: geçti ✓
4 saat → tahmin: geçti, gerçek: geçti ✓
3 saat → tahmin: kaldı, gerçek: kaldı ✓
Doğruluk: %100.0
```
