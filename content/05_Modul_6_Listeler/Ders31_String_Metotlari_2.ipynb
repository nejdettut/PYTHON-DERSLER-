## 1. `strip()` Ailesi

Kullanıcılar çoğu zaman yanlışlıkla başa/sona boşluk koyar. `" ali "` ile `"ali"` Python için **farklıdır**!

| Metot | Ne siler? |
|---|---|
| `strip()` | Baştaki **ve** sondaki boşlukları |
| `lstrip()` | Sadece **baştakileri** (left) |
| `rstrip()` | Sadece **sondakileri** (right) |
| `strip("*")` | Baştaki ve sondaki `*` karakterlerini |

### ✏️ Örnek 31.1: Boşluk temizliği

`metin = "   Merhaba Dünya   "` için `strip`, `lstrip`, `rstrip` sonuçlarını köşeli parantez içinde yazdır ki boşlukları görebilesin: `print("[" + metin.strip() + "]")`

💡 **Beklenen çıktı:**
```text
[Merhaba Dünya]
[Merhaba Dünya   ]
[   Merhaba Dünya]
```

### ✏️ Örnek 31.2: Farklı karakterleri silmek

`"***Merhaba Dünya***".strip("*")` ve `"123Merhaba12312311132".strip("123")` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
Merhaba Dünya
Merhaba
```

### ✏️ Örnek 31.3: Boşluklu şifre

Şifre `python` olsun. Kullanıcı `"  python "` girerse de kabul edilsin.

💡 **Beklenen çıktı:**
```text
Şifre:   python 
Giriş başarılı
```

## 2. Hizalama

| Metot | Ne yapar? |
|---|---|
| `metin.center(20)` | 20 karakterlik alanın **ortasına** yerleştirir |
| `metin.ljust(20)` | **Sola** yaslar, sağı doldurur |
| `metin.rjust(20)` | **Sağa** yaslar, solu doldurur |
| `metin.center(20, "*")` | Boşluk yerine `*` ile doldurur |

### ✏️ Örnek 31.4: PYTHON'u hizalayalım

`"PYTHON"` kelimesini 20 karakterlik alanda ortala, sola yasla, sağa yasla ve `*` ile ortala. Görünür olması için başına ve sonuna `|` ekle.

💡 **Beklenen çıktı:**
```text
|       PYTHON       |
|PYTHON              |
|              PYTHON|
|*******PYTHON*******|
```

### ✏️ Örnek 31.5: Fiyat listesi

Ürün adlarını `ljust(12, ".")`, fiyatları `rjust(8)` ile hizalayarak düzgün bir fiyat listesi yazdır.

💡 **Beklenen çıktı:**
```text
Çay.........   10 TL
Simit.......   15 TL
Tost........   40 TL
```

### ✏️ Örnek 31.6: Eşkenar üçgen

Tabanda 19 yıldız olacak şekilde 10 satırlık bir üçgeni `center(20)` ile çiz. İpucu: yıldız sayısı 1'den başlayıp her satırda 2 artar.

💡 **Beklenen çıktı:**
```text
         *
        ***
       *****
...
*******************
```

### ✏️ Örnek 31.7: Ters üçgen

Aynı üçgeni ters çevir.

💡 **Beklenen çıktı:**
```text
*******************
 *****************
...
         *
```
