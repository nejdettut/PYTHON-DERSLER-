## 1. `while` Döngüsü

`while` Türkçede **"...olduğu sürece"** demektir. Koşul **True olduğu sürece** blok tekrar eder.

```text
sayac = 0              # 1) Başlangıç değeri
while sayac < 5:       # 2) Koşul
    print(sayac)
    sayac += 1         # 3) Değeri değiştir!
```

> ⚠️ **Sonsuz döngü tuzağı:** 3. adımı unutursan `sayac` hep 0 kalır, koşul hep `True` olur ve döngü **hiç bitmez**.
> Böyle olursa hücrenin yanındaki **■ (Durdur)** düğmesine bas.

## ✅ Bugün Öğrendiklerimiz
- `while koşul:` → koşul True olduğu sürece tekrar eder.
- Koşuldaki değişken döngü içinde **değişmelidir**, yoksa sonsuz döngü olur.
- Tekrar sayısı belli değilse `while`, belliyse `for` tercih edilir.
