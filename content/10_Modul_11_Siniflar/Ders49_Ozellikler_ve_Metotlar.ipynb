## 1. Nesnenin Durumu

Bir nesnenin özellikleri onun **durumunu** gösterir. Metotlar bu durumu değiştirir.

```text
lamba.acik_mi      → False   (durum)
lamba.ac()         → metot durumu değiştirir
lamba.acik_mi      → True
```

### ✏️ Örnek 49.1: Masa lambası

`MasaLambasi` sınıfı yaz:
- `__init__(self, siddet, renk="sarı")` → `siddet`, `renk` ve `acik_mi = False`
- `ac()` ve `kapat()` metotları `acik_mi`'yi değiştirsin ve mesaj **döndürsün**
- `isigi_artir(miktar)` şiddeti artırsın

### ✏️ Örnek 49.2: Lambayı kullan

Şiddeti 60 olan bir lamba oluştur. Durumunu yazdır, aç, şiddeti 5 artır, tekrar durumunu yazdır, kapat.

💡 **Beklenen çıktı:**
```text
Açık mı? False
Işık açıldı
Işık şiddeti: 65
Açık mı? True
Işık kapatıldı
```

### ✏️ Örnek 49.3: İki lamba, iki ayrı durum

İki lamba oluştur, sadece birini aç. İkisinin durumunu yazdır. Neden biri açık, diğeri kapalı?

💡 **Beklenen çıktı:**
```text
Lamba 1 açık mı? True
Lamba 2 açık mı? False
```

### ✏️ Örnek 49.4: Nesneyi sil

`del lamba1` yaz ve ardından `lamba1.acik_mi`'yi yazdırmayı dene.

> ⚠️ `NameError` alacaksın.

## 2. Sınıf Özelliği ve Nesne Özelliği

```text
class Ogrenci:
    okul = "Ankara Lisesi"           # SINIF özelliği → tüm nesneler ORTAK
    def __init__(self, ad):
        self.ad = ad                 # NESNE özelliği → her nesnede FARKLI
```

### ✏️ Örnek 49.5: Ortak ve özel

Yukarıdaki sınıfı yaz. İki öğrenci oluştur, ikisinin `ad` ve `okul` özelliğini yazdır. Sonra `Ogrenci.okul = "Bilişim Lisesi"` yap ve tekrar yazdır.

💡 **Beklenen çıktı:**
```text
Ali Ankara Lisesi
Ayşe Ankara Lisesi
Ali Bilişim Lisesi
Ayşe Bilişim Lisesi
```

### ✏️ Örnek 49.6: Nesne sayacı

`Ogrenci` sınıfına `sayi = 0` sınıf özelliği ekle. `__init__` her çalıştığında `Ogrenci.sayi += 1` yapsın. 3 öğrenci oluşturup toplam sayıyı yazdır.

💡 **Beklenen çıktı:**
```text
Toplam öğrenci: 3
```

## 3. Uygulama: Banka Hesabı 🏦

### ✏️ Örnek 49.7: Hesap sınıfı

`Hesap` sınıfı yaz: `sahip`, `bakiye=0` özellikleri; `para_yatir(miktar)`, `para_cek(miktar)` ve `bilgi()` metotları olsun. Bakiye yetersizse para çekilmesin, uyarı versin.

💡 **Beklenen çıktı:**
```text
Ali'nin hesabına 500 TL yatırıldı.
200 TL çekildi.
Bakiye yetersiz!
Hesap sahibi: Ali | Bakiye: 300 TL
```

### 🎯 Kendini Dene 1: Karakter Oluştur 🎮

Bir oyun için `Karakter` sınıfı yaz: `ad`, `can=100`, `seviye=1`. `hasar_al(miktar)`, `iyiles(miktar)` (can 100'ü geçmesin) ve `seviye_atla()` metotları olsun. Karakterle küçük bir senaryo oynat.

## ✅ Bugün Öğrendiklerimiz
- Metotlar `self.ozellik` üzerinden nesnenin durumunu değiştirir.
- Her nesnenin kendi özellik değerleri vardır; birini değiştirmek diğerini etkilemez.
- Sınıf özellikleri tüm nesnelerde ortaktır.
