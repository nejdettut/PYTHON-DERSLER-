## 1. Hangi Soruya Hangi Grafik?

| Soru | Grafik | matplotlib |
|---|---|---|
| Kimin değeri ne kadar? (karşılaştırma) | Çubuk | `plt.bar(x, y)` |
| Değerler nasıl dağılmış? | Histogram | `plt.hist(veri)` |
| Gruplar arasında dağılım farkı | Kutu grafiği | `df.boxplot(column=..., by=...)` |
| İki değişken arasında ilişki var mı? | Saçılım (scatter) | `plt.scatter(x, y)` |

Her grafikte: `plt.title()`, `plt.xlabel()`, `plt.ylabel()` → sonra `plt.show()`

## 2. İlişki Var mı? Saçılım Grafiği

Makine öğrenmesinin temel sorusu: **Bir özellik, sonucu tahmin etmeye yardım eder mi?**
Saçılım grafiğinde noktalar bir **çizgi** gibi diziliyorsa iki değişken arasında ilişki vardır.

### ✏️ Örnek 9.5: Çalışma saati ↔ puan

### ✏️ Örnek 9.7: Korelasyon

İlişkinin gücünü sayıyla ölç: `df["calisma_saati"].corr(df["puan"])` ve `df["devamsizlik"].corr(df["puan"])`.

+1'e yakın → güçlü **pozitif** ilişki · −1'e yakın → güçlü **negatif** ilişki · 0'a yakın → ilişki yok

💡 **Beklenen çıktı:**
```text
Çalışma saati - puan: 0.97
Devamsızlık - puan: -0.97
```
