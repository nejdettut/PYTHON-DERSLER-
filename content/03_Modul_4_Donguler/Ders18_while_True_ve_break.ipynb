## 1. `while True` + `break`

`while True:` koşulu hep doğru olan, **bilerek kurulmuş** bir sonsuz döngüdür.
Döngüden çıkmanın tek yolu **`break`** komutudur.

```text
while True:
    ...
    if çıkış_koşulu:
        break        # döngü burada biter
    ...
```

Menüler, oyunlar ve "çıkmak için 0 girin" türü programlar için idealdir.

### ✏️ Örnek 18.1: 0 girilene kadar kare (while True ile)

Ders 17'deki programı `while True` ve `break` ile yeniden yaz.

💡 **Beklenen çıktı:**
```text
Çıkmak için 0 giriniz!
Sayı Giriniz: 5
Girdiğiniz sayının karesi: 25
Sayı Giriniz: 0
Çıkış yapılıyor...
```

### ✏️ Örnek 18.2: Trafik lambası

Kullanıcıdan lamba rengini iste: `KIRMIZI` → BEKLE, `SARI` → HAZIRLAN, `YEŞİL` → GEÇ ve **döngüden çık**. Başka bir şey → `Hatalı giriş`.

💡 **Beklenen çıktı:**
```text
Lamba rengi [KIRMIZI, SARI, YEŞİL]: KIRMIZI
BEKLE
Lamba rengi [KIRMIZI, SARI, YEŞİL]: YEŞİL
GEÇ
```

## 3. Sayı Tahmin Oyunu 🎲

Rastgele sayı üretmek için `random` kütüphanesini kullanırız:

```text
import random
pc = random.randint(1, 100)    # 1 ile 100 arasında (ikisi dahil) rastgele tam sayı
```

### ✏️ Örnek 18.5: Bilgisayarın tuttuğu sayıyı bul

Bilgisayar 1–100 arası bir sayı tutsun. Kullanıcı bulana kadar tahmin istesin, her seferinde `Daha büyük` / `Daha küçük` desin. Sayı bulununca kaç tahminde bulunduğunu yazdır.
