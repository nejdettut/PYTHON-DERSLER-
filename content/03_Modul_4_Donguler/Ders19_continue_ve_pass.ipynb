## 1. Üç Komut, Üç Görev

| Komut | Ne yapar? | Benzetme |
|---|---|---|
| `break` | Döngüyü **tamamen bitirir** | Koşuyu bırakıp eve gitmek |
| `continue` | O turu **atlar**, sonraki tura geçer | Bir engeli atlayıp koşmaya devam etmek |
| `pass` | **Hiçbir şey yapmaz** | Yer tutucu, "şimdilik boş" |

### ✏️ Örnek 19.1: 12'yi atla (for)

0'dan 20'ye çift sayıları yan yana yazdır ama 12'yi atla.

💡 **Beklenen çıktı:**
```text
0 2 4 6 8 10 14 16 18
```

### ✏️ Örnek 19.2: 12'yi atla (while)

Aynısını `while` ile yaz.

> ⚠️ Artırma satırını `continue`'dan **önce** yazmazsan sonsuz döngüye girersin!

💡 **Beklenen çıktı:**
```text
0 2 4 6 8 10 14 16 18
```

### ✏️ Örnek 19.3: Birden fazla sayıyı atla

0–20 arası çift sayılardan 6 ve 12'yi atla. Önce `or` ile, sonra `in [6, 12]` ile yaz.

💡 **Beklenen çıktı:**
```text
0 2 4 8 10 14 16 18 20
```

## 2. `pass` Komutu

Python boş bir bloğa izin vermez. Bir bloğu **şimdilik boş** bırakmak istersen `pass` yazarsın:

```text
if sayi < 0:
    pass        # bunu sonra yazacağım
```

### ✏️ Örnek 19.4: `pass` hiçbir şey değiştirmez

Örnek 19.1'de `continue` yerine `pass` yaz. 12 atlandı mı? Sonra `pass` satırını tamamen silip çalıştır.

> ⚠️ İkinci denemede `IndentationError` alacaksın: `if` bloğu boş kalamaz.

## 3. Döngülü Uygulamalar

### ✏️ Örnek 19.5: Parola belirleme

Kullanıcıdan parola iste. Boşsa `Parola boş geçilemez`, 6 karakterden kısaysa `En az 6 karakter olmalı` deyip **tekrar sor** (`continue`). Uygunsa onayla ve çık (`break`).

💡 **Beklenen çıktı:**
```text
Parola giriniz: 
Parola boş geçilemez.
Parola giriniz: 123
En az 6 karakter olmalı.
Parola giriniz: python26
Parolanız oluşturuldu.
```

### ✏️ Örnek 19.6: Döngülü hesap makinesi

İki sayı ve bir operatör (`+ - * /`) al, sonucu yazdır. Sonra `Başka işlem? [E/H]` diye sor. `H` girilirse çık, `E` girilirse devam et. Bölmede ikinci sayı 0 ise uyarı ver.

💡 **Beklenen çıktı:**
```text
1. sayı: 8
2. sayı: 2
Operatör [+,-,*,/]: *
Sonuç: 16
Başka işlem? [E/H]: h
ÇIKIŞ
```

### 🎯 Kendini Dene 1: Bıktıran Program 😄

Program sürekli `Nasılsın?` diye sorsun. Boş bırakılırsa `Cevap yazar mısın?` desin. Kullanıcı `yeter` yazınca (büyük-küçük harf fark etmeden) çıksın.

💡 **Örnek çıktı:**
```text
Nasılsın? iyi
Nasılsın? 
Cevap yazar mısın?
Nasılsın? YETER
Çıkış yapılıyor...
```

## ✅ Bugün Öğrendiklerimiz
- `break` döngüyü bitirir, `continue` turu atlar, `pass` hiçbir şey yapmaz.
- `while` içinde `continue` kullanırken artırma satırının yerine dikkat et.
- `while True` + `continue` + `break` ile kullanıcıdan doğru veri gelene kadar sorulabilir.
