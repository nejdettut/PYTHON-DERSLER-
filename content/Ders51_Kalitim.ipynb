## 1. Kalıtım Nedir?

Bir sınıfın, başka bir sınıfın özelliklerini ve metotlarını **devralmasıdır**.

```text
        Kedi  (üst sınıf / ebeveyn)
        ├── ad, yas
        └── ozellikler(), miyavla()
       ↙                ↘
PersKedisi            VanKedisi   (alt sınıflar / çocuk)
+ tuyler()            + goz_rengi()
```

```text
class VanKedisi(Kedi):     # parantez içine üst sınıf yazılır
    ...
```

- Alt sınıf, üst sınıfın her şeyini **miras alır**; ayrıca kendi özelliklerini ekleyebilir.
- Aynı kodu tekrar tekrar yazmaktan kurtarır.

### ✏️ Örnek 51.1: Kedi üst sınıfı

`Kedi` sınıfı yaz: sınıf özelliği `tur = "memeli"`, `ad` ve `yas` özellikleri, `ozellikler()` ve `miyavla(ses)` metotları (ikisi de metin döndürsün).

### ✏️ Örnek 51.2: Alt sınıflar

`Kedi`'den miras alan `PersKedisi` (`tuyler(renk)` metodu) ve `VanKedisi` (`goz_rengi(renk)` metodu) alt sınıflarını yaz. Bu sınıflarda `__init__` yazmana gerek yok!

### ✏️ Örnek 51.3: Miras çalışıyor mu?

Her iki alt sınıftan birer kedi oluştur. Hem miras aldıkları `ozellikler()` ve `miyavla()` metotlarını, hem de kendi metotlarını çağır.

💡 **Beklenen çıktı:**
```text
Pisi adlı kedi 4 yaşında
Pisi adlı kedinin gözü mavi-sarı rengindedir.
Persi adlı kedi Miyavvv diye ses çıkarır
Persi adlı kedinin tüyü beyaz
```

### ✏️ Örnek 51.4: Kardeşin metodu

Pers kedisinin `goz_rengi()` metodunu çağırmayı dene.

> ⚠️ `AttributeError`: Kardeş sınıfın metodu miras alınmaz; sadece üst sınıfınkiler alınır.

## 2. Override ve `super()`

- **Override (geçersiz kılma):** Alt sınıfta, üst sınıftaki bir metodu **aynı adla yeniden yazmak**. Alt sınıfın sürümü geçerli olur.
- **`super()`**: Alt sınıfın içinden **üst sınıfın** metodunu çağırmak için kullanılır.

```text
class EKitap(Kitap):
    def __init__(self, baslik, yazar, format):
        super().__init__(baslik, yazar)    # üst sınıfın işini yaptır
        self.format = format               # sadece yeni olanı ekle
```

### ✏️ Örnek 51.5: Override

`PersKedisi`'ni yeniden yaz ve `ozellikler()` metodunu override ederek `Türü: Pers Kedisi | Adı: Leyla | Yaşı: 1` döndürmesini sağla.

💡 **Beklenen çıktı:**
```text
Türü: Pers Kedisi | Adı: Leyla | Yaşı: 1
```

### ✏️ Örnek 51.6: Kitap sınıfı

`baslik`, `yazar`, `sayfa` özellikleri ve `ozet()` metodu olan `Kitap` sınıfını yaz.

💡 **Beklenen çıktı:**
```text
Başlık: Nutuk
Yazar: Mustafa Kemal Atatürk
Sayfa: 600
```

### ✏️ Örnek 51.7: EKitap alt sınıfı

`Kitap`'tan miras alan `EKitap` sınıfı yaz; ek olarak `format` özelliği olsun. Hem `__init__`'te hem `ozet()`'te `super()` kullan.

💡 **Beklenen çıktı:**
```text
Başlık: Dune
Yazar: Frank Herbert
Sayfa: 600
Format: epub
```

## 📝 Ödev 18 – Sınıf ve Nesneler

Bir `Ogrenci` sınıfı oluşturunuz. `ad`, `soyad`, `numara` ve `notlar` (boş liste) özellikleri olsun. Şu metotları yazınız:
- `not_ekle(not_)` → notu listeye eklesin (0–100 dışını kabul etmesin)
- `ortalama()` → ortalamayı döndürsün (not yoksa 0)
- `bilgileri_yazdir()` → tüm bilgileri ve ortalamayı yazsın

**Bonus:** `Ogrenci`'den miras alan `LiseOgrencisi` sınıfı yazıp `alan` (Sayısal/Sözel) özelliği ekleyin ve `bilgileri_yazdir()`'ı override edin.

💡 **Örnek çıktı:**
```text
Ad: Yusuf
Soyad: Aslan
Numara: 4125
Notlar: [85, 90, 75]
Ortalama: 83.33
```

## ✅ Bugün Öğrendiklerimiz
