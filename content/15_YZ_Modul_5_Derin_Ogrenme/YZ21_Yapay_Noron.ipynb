## 1. Beyinden Bilgisayara 🧠➡️💻

**Derin öğrenme**, insan beynindeki sinir hücrelerinden (nöronlardan) esinlenen **yapay sinir ağlarını** kullanan makine öğrenmesi dalıdır. Yüz tanıma, sesli asistanlar, ChatGPT gibi sohbet botları derin öğrenmeyle çalışır.

| Biyolojik nöron | Yapay nöron |
|---|---|
| Dendritler sinyal **alır** | **Girdiler** (x₁, x₂, …) |
| Bazı bağlantılar daha **güçlüdür** | Her girdinin bir **ağırlığı** (w₁, w₂, …) var |
| Sinyal yeterince güçlüyse nöron **ateşlenir** | **Aktivasyon fonksiyonu** karar verir |

```text
 x₁ ──(w₁)──┐
            ├──▶  toplam = x₁·w₁ + x₂·w₂ + b  ──▶  aktivasyon  ──▶  çıktı
 x₂ ──(w₂)──┘                         ↑
                                 b (eşik / bias)
```

> 🔹 Bir nöron sadece **çarpar, toplar ve karar verir**. Derin öğrenmenin sihri, milyonlarca nöronun birlikte çalışmasından gelir.

### ✏️ Örnek 21.1: Ağırlıklı toplam

Bir öğrencinin dışarı çıkıp çıkmayacağına karar veren nöron düşün: `x1` = hava güzel mi (1/0), `x2` = ödev bitti mi (1/0). Ağırlıklar `w1 = 0.4`, `w2 = 0.7`, eşik `b = -0.5`. `toplam = x1*w1 + x2*w2 + b` değerini 4 durum için hesapla.

💡 **Beklenen çıktı:**
```text
hava=0 ödev=0 → toplam=-0.5
hava=0 ödev=1 → toplam=0.2
hava=1 ödev=0 → toplam=-0.1
hava=1 ödev=1 → toplam=0.6
```

### ✏️ Örnek 21.2: Basamak aktivasyonu

`basamak(t)` fonksiyonu toplam 0'dan büyükse 1, değilse 0 döndürsün. 21.1'deki 4 durum için nöronun kararını yazdır. Öğrenci hangi durumlarda dışarı çıkıyor?

💡 **Beklenen çıktı:**
```text
hava=0 ödev=0 → 0 (evde)
hava=0 ödev=1 → 1 (dışarı)
hava=1 ödev=0 → 0 (evde)
hava=1 ödev=1 → 1 (dışarı)
```

### ✏️ Örnek 21.3: Nöron fonksiyonu

`noron(girdiler, agirliklar, esik)` fonksiyonu, istediğimiz sayıda girdi alabilsin ve basamak aktivasyonunun sonucunu döndürsün. İpucu: `zip` ile girdileri ve ağırlıkları eşleştir.

## 2. Aktivasyon Fonksiyonları

| Fonksiyon | Formül | Çıktı | Kullanım |
|---|---|---|---|
| **Basamak** | t > 0 ise 1, değilse 0 | Sadece 0 veya 1 | İlk nöron modelleri |
| **Sigmoid** | 1 / (1 + e⁻ᵗ) | 0 ile 1 arası (olasılık gibi) | Çıkış katmanında |
| **ReLU** | max(0, t) | 0 veya pozitif | Modern derin ağların gizli katmanlarında |

### ✏️ Örnek 21.4: Üç aktivasyon

`sigmoid(t)` (`math.exp` kullan) ve `relu(t)` fonksiyonlarını yaz. t = −5'ten 5'e kadar üç fonksiyonun grafiğini aynı şekilde, yan yana üç alt grafik olarak çiz (`plt.subplot(1, 3, i)`).

### ✏️ Örnek 21.5: Sigmoid ile karar

21.1'deki nöronda basamak yerine sigmoid kullan. Çıktılar artık "dışarı çıkma **olasılığı**" gibi okunabilir. 4 durumu yüzde olarak yazdır.

💡 **Beklenen çıktı:**
```text
hava=1 ödev=1 → %64.6 ihtimalle dışarı
```

## 3. Mantık Kapıları 🔌

Ders 5'teki `and` ve `or`'u hatırla! Bir nöron uygun ağırlıklarla bunları taklit edebilir.

### ✏️ Örnek 21.6: VE kapısı

`noron` fonksiyonunu `w1 = 1`, `w2 = 1`, `b = -1.5` ile 4 girdi durumu için çalıştır ve Python'un `and` sonucuyla karşılaştır.

💡 **Beklenen çıktı:**
```text
0 VE 0 = 0  (Python: 0)
0 VE 1 = 0  (Python: 0)
1 VE 0 = 0  (Python: 0)
1 VE 1 = 1  (Python: 1)
```

### ✏️ Örnek 21.7: VEYA kapısı

Sadece eşiği değiştirerek VEYA kapısını elde et. Hangi `b` değeri işe yarar?
