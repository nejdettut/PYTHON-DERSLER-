## 1. Piksel Nedir? 🟦

Ekrandaki her görüntü küçük karelerden (**piksel**) oluşur. Bilgisayar için bir görüntü, her pikselin değerini tutan bir **sayı tablosudur**.

| Görüntü türü | Yapı | Piksel değeri |
|---|---|---|
| Siyah-beyaz (gri) | 2 boyutlu tablo: `(yükseklik, genişlik)` | 0 (siyah) – 255 (beyaz) |
| Renkli (RGB) | 3 boyutlu tablo: `(yükseklik, genişlik, 3)` | Her piksel için [Kırmızı, Yeşil, Mavi], her biri 0–255 |

**Bilgisayarlı görü (computer vision):** Bilgisayarların bu sayı tablolarından anlam çıkarmasıdır: yüz tanıma, plaka okuma, otonom araçlar, tıbbi görüntü analizi.

Görüntüleri **numpy** dizileriyle tutacağız:
```text
import numpy as np
resim = np.zeros((8, 8))        # 8x8, tamamı 0 (siyah)
resim[2, 3] = 255               # 2. satır, 3. sütun → beyaz
```

### ✏️ Örnek 31.1: İlk resmin

`np.zeros((5, 5))` ile 5×5 siyah bir resim oluştur, ortadaki pikseli ve dört köşeyi 255 yap. Sayı tablosunu yazdır ve `plt.imshow(resim, cmap="gray")` ile göster.

### ✏️ Örnek 31.2: Piksel sanatı: gülen yüz 🙂

8×8'lik bir listeyle (0 ve 1'lerden) gülen yüz çiz, `np.array` ile diziye çevir ve göster. Sonra resmi `resim[::-1]` ile ters çevir. Dilimleme (Ders 6) görüntülerde de çalışıyor!

### ✏️ Örnek 31.3: Gri tonlar

1×10'luk bir dizide 0'dan 255'e eşit aralıklarla artan değerler oluştur (`np.linspace(0, 255, 10)`) ve göster. Sonra 50×256'lık yatay bir gri geçiş (gradyan) resmi oluştur.

## 2. Renkli Görüntüler: RGB 🎨

Her pikselde 3 sayı var: **K**ırmızı, **Y**eşil, **M**avi (Red, Green, Blue). Işık renkleri karıştırılır:

| Renk | [R, G, B] |
|---|---|
| Kırmızı | [255, 0, 0] |
| Yeşil | [0, 255, 0] |
| Sarı | [255, 255, 0] |
| Beyaz | [255, 255, 255] |
| Siyah | [0, 0, 0] |

> ⚠️ Renkli görüntülerde `dtype=np.uint8` kullan (0–255 tam sayı).

### ✏️ Örnek 31.4: Renk kutuları

`np.zeros((50, 150, 3), dtype=np.uint8)` oluştur. Sol üçte birini kırmızı, ortayı yeşil, sağı mavi boya ve göster.

### ✏️ Örnek 31.5: Bayrak çiz

Kırmızı-beyaz yatay iki şeritli bir bayrak (60×90) çiz. Bonus: Ortaya beyaz bir kare ekle.

### ✏️ Örnek 31.6: Kendi rengini karıştır

Kullanıcıdan R, G, B değerlerini al ve o renkte 50×50'lik bir kare göster. Turuncu, mor ve pembe için hangi değerleri buldun?

💡 **Beklenen çıktı:**
```text
R: 255
G: 165
B: 0
```

Şimdi daha büyük bir resimle çalışalım. Bu hazır hücre bir manzara resmi üretir:

````python
# 60x90 piksellik renkli bir manzara resmi üretelim. Sadece çalıştır!
import numpy as np
import matplotlib.pyplot as plt

Y, G = 60, 90
resim = np.zeros((Y, G, 3), dtype=np.uint8)
for s in range(Y):                                    # gökyüzü: yukarıdan aşağı açılan mavi
    resim[s, :] = [80 + s, 150 + s, 255]
resim[40:, :] = [60, 160, 60]                         # çimen
yy, xx = np.ogrid[:Y, :G]
resim[(yy - 14) ** 2 + (xx - 70) ** 2 <= 49] = [255, 220, 0]   # güneş
resim[28:40, 15:35] = [180, 80, 40]                   # ev gövdesi
for i in range(10):                                   # çatı (üçgen)
    resim[18 + i, 25 - i:26 + i] = [200, 30, 30]
resim[33:40, 22:27] = [90, 50, 20]                    # kapı

plt.imshow(resim)
plt.axis("off")
plt.show()
print("Boyut:", resim.shape)
````
