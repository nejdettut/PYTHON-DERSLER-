## 1. Fonksiyon Nedir?

Belirli bir işi yapan, **adı olan** küçük kod parçasıdır. Bir kez yazılır, istenildiği kadar **çağrılır**.

**Neden kullanırız?**
- Aynı kodu tekrar tekrar yazmaktan kurtarır.
- Programı **okunur** ve **düzenli** yapar.
- Hata bulmayı ve düzeltmeyi kolaylaştırır.

| Tür | Açıklama | Örnek |
|---|---|---|
| **Gömülü** (built-in) | Python'la hazır gelir | `print()`, `input()`, `len()`, `type()`, `int()` |
| **Modül fonksiyonu** | `import` ile eklenir | `math.sqrt()`, `random.randint()` |
| **Özel** (custom) | **Biz** yazarız | `def selam(): ...` |

### ✏️ Örnek 21.1: Gömülü fonksiyonları görelim

`dir(__builtins__)` yazıp çalıştır. Tanıdığın fonksiyonları bul!

### ✏️ Örnek 21.2: math modülü

`import math` ile modülü ekle. `math.factorial(5)`, `math.sqrt(16)` ve `math.pow(2, 3)` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
120
4.0
8.0
```

### ✏️ Örnek 21.3: Karekök hesaplayıcı

Kullanıcıdan sayı al, karekökünü yazdır.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 49
Karekök: 7.0
```

## 2. `def` ile Fonksiyon Tanımlama

```text
def fonksiyon_adi():      # 1) def + ad + () + :
    yapılacak işler       # 2) girintili gövde

fonksiyon_adi()           # 3) adıyla ÇAĞIR
```

`def`, İngilizce **define** (tanımla) kelimesinden gelir.

> ⚠️ Fonksiyonu **tanımlamak** onu çalıştırmaz. Çalışması için **çağırmak** gerekir.

### ✏️ Örnek 21.4: İlk fonksiyonumuz

`selam()` adında, ekrana iki satır mesaj yazan bir fonksiyon tanımla. Sonra fonksiyonu **3 kez** çağır.

💡 **Beklenen çıktı:**
```text
Merhaba!
Python'a hoş geldin.
Merhaba!
...
```

### ✏️ Örnek 21.5: Tanımla ama çağırma

Önce sadece bir fonksiyon tanımla ve çalıştır, hiç çıktı yok mu? Sonra ikinci hücrede, **tanımlamadığın** bir fonksiyonu çağırmayı dene.

> ⚠️ İkinci hücre `NameError` verecek.

### ✏️ Örnek 21.6: İçinde input olan fonksiyon

`tanis()` fonksiyonu kullanıcıdan ad ve yaş alsın, `Merhaba Ayşe, yaşınız: 15` yazsın.

💡 **Beklenen çıktı:**
```text
Adınız: Ayşe
Yaşınız: 15
Merhaba Ayşe, yaşınız: 15
```

### ✏️ Örnek 21.7: Fonksiyona değer gönderelim

Parantez içine bir değişken yazarsak fonksiyona değer gönderebiliriz. `cift(sayi)` fonksiyonu sayının tek mi çift mi olduğunu yazsın. Fonksiyonu 7, 10 ve 7+10 için çağır.

💡 **Beklenen çıktı:**
```text
7 tek sayıdır.
10 çift sayıdır.
17 tek sayıdır.
```

### 🎯 Kendini Dene 1: Karenin Alanı

`kare_alan()` fonksiyonu kullanıcıdan bir kenar uzunluğu alsın ve karenin alanını yazsın. Fonksiyonu iki kez çağır.

💡 **Örnek çıktı:**
```text
Bir kenar giriniz: 4
Karenin alanı: 16
```

## ✅ Bugün Öğrendiklerimiz
- Fonksiyon, adı olan ve tekrar kullanılabilen kod bloğudur.
- `import modul` ile hazır modüller eklenir: `math.sqrt()`.
- `def ad():` ile tanımlanır, `ad()` ile çağrılır; tanımlamak çalıştırmak değildir.
