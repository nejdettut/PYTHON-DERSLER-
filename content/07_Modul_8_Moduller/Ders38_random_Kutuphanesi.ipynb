## 1. random Modülü

| Fonksiyon | Ne üretir? |
|---|---|
| `random.random()` | 0 ile 1 arası ondalıklı sayı (1 hariç) |
| `random.randint(a, b)` | a ile b arası **tam** sayı (ikisi dahil) |
| `random.uniform(a, b)` | a ile b arası **ondalıklı** sayı |
| `random.choice(liste)` | Listeden **bir** eleman |
| `random.sample(liste, k)` | Listeden **k tane farklı** eleman |
| `random.shuffle(liste)` | Listeyi **karıştırır** (yerinde) |
| `random.seed(n)` | Aynı rastgele diziyi tekrar üretmek için başlangıç |

> 🎲 Bu derste her çalıştırmada farklı çıktılar alacaksın, bu normal!

### ✏️ Örnek 38.1: random() ve round

`random.random()` ile iki sayı üret, toplamlarını yazdır. Sonra bir sayıyı `round(sayi, 3)` ile 3 basamağa yuvarla.

### ✏️ Örnek 38.2: Zar at

`randint` ile iki zar at, sonuçları ve toplamı yazdır. Çift gelirse `Çift attın!` yaz.

💡 **Beklenen çıktı:**
```text
Zar 1: 4
Zar 2: 4
Toplam: 8
Çift attın!
```

### ✏️ Örnek 38.3: seed ile aynı sonuç

`random.seed(42)` yazıp ardından 3 tane `randint(1, 100)` üret. Hücreyi birkaç kez çalıştır. Sonuç değişiyor mu?

### ✏️ Örnek 38.4: Üç sayının ortalaması

1–10 arası 3 rastgele sayı üret, listeye ekle, listeyi ve ortalamayı (2 basamak) yazdır.

### ✏️ Örnek 38.5: Tekrarsız sayılar

1–10 arası rastgele sayı üret; listede **yoksa** ekle, varsa `Tekrar: x` yaz. Listede 3 sayı olunca dur.

> ⚠️ Bu problemin tek satırlık çözümü de var: `random.sample(range(1, 11), 3)`

## 2. Listeden Rastgele Seçim

### ✏️ Örnek 38.6: Rastgele çiçek

`cicekler = ['gül', 'karanfil', 'papatya', 'begonya', 'menekşe', 'lale', 'sümbül', 'zakkum']` listesinden önce `choice` ile 3 kez seç (tekrar edebilir mi?), sonra `sample` ile 3 farklı çiçek seç.

### ✏️ Örnek 38.7: Karıştır

1–10 arası sayılardan liste oluştur, `shuffle` ile karıştırıp yazdır.

### ✏️ Örnek 38.8: Sınıf çekilişi 🎁

En az 10 isimlik bir listeden **asil** ve **yedek** kazanan seç. Aynı kişi iki kez seçilmesin. İki yolla dene: (1) `choice` + `remove`, (2) `sample(liste, 2)`.

💡 **Beklenen çıktı:**
```text
Asil kazanan: @elif
Yedek kazanan: @kaan
```

## ✅ Bugün Öğrendiklerimiz
- `randint(a, b)` tam sayı, `uniform(a, b)` ondalıklı rastgele sayı üretir.
- `choice` bir eleman seçer; `sample` tekrarsız birden fazla seçer; `shuffle` karıştırır.
- `seed(n)` ile rastgele sonuçlar tekrarlanabilir hâle gelir.
