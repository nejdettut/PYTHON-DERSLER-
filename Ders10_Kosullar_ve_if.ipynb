## 1. Program Nasıl Karar Verir?

Şimdiye kadar yazdığımız kodlar **yukarıdan aşağıya**, satır satır çalıştı. Ama gerçek hayatta kararlar duruma göre değişir:

- **Eğer** hava yağmurluysa → şemsiye al
- **Eğer** sınavdan 50 ve üzeri aldıysan → geçtin
- **Eğer** şifre doğruysa → giriş yap

Python'da bu kararlar **koşullu ifadelerle** verilir:

| Durum | Yapı |
|---|---|
| Tek koşul | `if` |
| İki durum | `if – else` |
| Çok durum | `if – elif – else` |

> 📌 Bir koşulun sonucu mutlaka **`True`** ya da **`False`** olmalıdır. (Ders 5'i hatırla!)

## 2. Girinti (Indentation) ve Kod Bloğu

Python'da kod blokları süslü parantez `{ }` ile değil, **girinti** ile oluşturulur.

- Satır başında bırakılan boşluğa **girinti** denir. Standart: **4 boşluk** (veya 1 Tab).
- **Aynı hizada** başlayan satırlar **aynı bloğa** aittir ve birlikte çalışır.
- Girinti `if`, döngüler (`for`, `while`) ve fonksiyonlarda (`def`) kullanılır.

> ⚠️ Girinti hatası Python'da **en sık yapılan** hatadır ve programın hiç çalışmamasına neden olur.

### ✏️ Örnek 10.1: Kod bloğunu inceleyelim

`yas = 20` tanımla. `if yas >= 18:` altına **girintili** iki `print()` yaz. En alta **girintisiz** bir `print()` ekle. Sonra `yas = 15` yapıp tekrar çalıştır. Hangi satırlar kayboldu?

💡 **Beklenen çıktı:**
```text
Yaşınız 18 ve üzeri
Bu satır if bloğunun içinde
Bu satır if bloğunun dışında
```

### ✏️ Örnek 10.2: Gereksiz girinti hatası

Önce `print("Blok Yapısı")` yaz, alt satıra başında **3 boşluk** bırakarak `a = 5` yaz ve çalıştır. Sonra ikinci hücrede düzelt.

> ⚠️ İlk hücre `IndentationError` verecek. Bloğu olmayan yerde girinti kullanılmaz.

## 3. `if` Komutu

`if` Türkçede **"eğer"** demektir.

```text
if koşul:
    koşul True ise çalışacak kodlar
```

- Koşul **True** ise blok çalışır, **False** ise blok **atlanır**.
- `if` satırının sonundaki **iki nokta `:`** zorunludur. "...ise" anlamı taşır.

### ✏️ Örnek 10.3: İlk if kodumuz

`sayi = 10` için `sayi > 5` ise ekrana mesaj yazdır. İkinci hücrede `sayi = 3` ile aynı kodu çalıştır. Neden hata almadığın halde çıktı yok?

💡 **Beklenen çıktı:**
```text
Sayı 5'ten büyüktür
```

### ✏️ Örnek 10.4: İki noktayı unutursak?

Örnek 10.3'teki kodu `:` olmadan yazıp çalıştır.

> ⚠️ `SyntaxError: expected ':'` hatası alacaksın.

### ✏️ Örnek 10.5: Pozitif mi, negatif mi, sıfır mı?

Kullanıcıdan bir sayı al. **Üç ayrı `if`** ile sayının pozitif, negatif veya sıfır olduğunu yazdır.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: -4
Girilen sayı negatiftir
```

### ✏️ Örnek 10.6: Yaşa göre selam

Kullanıcıdan yaşını al ve uygun selamı yazdır. `and` kullan.

| Yaş | Mesaj |
|---|---|
| 14'ten küçük | Merhaba Çocuk |
| 14–17 | Merhaba Liseli |
| 18–23 | Merhaba Üniversiteli |
| 24 ve üzeri | Merhaba Çalışan İnsan |

💡 **Beklenen çıktı:**
```text
Yaşınızı girin: 15
Merhaba Liseli
```

### ✏️ Örnek 10.7: Kullanıcı adı ve şifre kontrolü

Kullanıcı adı `admin`, şifre `123456` olsun. Dört ayrı `if` ile şu durumları kontrol et: ikisi doğru, sadece şifre yanlış, sadece kullanıcı adı yanlış, ikisi de yanlış.

💡 **Beklenen çıktı:**
```text
Kullanıcı adınızı girin: admin
Şifrenizi girin: 1111
Şifre yanlış
```

### 🎯 Kendini Dene 1: Termometre

Kullanıcıdan hava sıcaklığını al. 30'dan büyükse `Çok sıcak, bol su iç!`, 10'dan küçükse `Soğuk, mont giy!` yazdır.

💡 **Örnek çıktı:**
```text
Sıcaklık: 34
Çok sıcak, bol su iç!
```

## 📝 Ödev 3 – if Komutu

Klavyeden girilen sınav notu **50 veya üzeriyse** `Geçti`, **50'den küçükse** `Kaldı` mesajını yazdıran kodu sadece `if` kullanarak yazınız.

💡 **Örnek çıktı:**
```text
Sınav notunu giriniz: 50
Geçti
```
