## 3. Turing Testi 🤖❓🧑

1950'de **Alan Turing** şöyle bir oyun önerdi:
- Bir **hakem**, görmediği iki kişiyle yazışır: biri **insan**, biri **makine**.
- Hakem hangisinin makine olduğunu **ayırt edemezse**, makine testi geçmiş sayılır.

> 🔹 Turing'e göre önemli olan makinenin "gerçekten düşünmesi" değil, **düşünüyormuş gibi davranabilmesidir**.

### ✏️ Örnek 3.5: İnsan mı, makine mi?

Aşağıdaki cevapları bir listede tut (`(cevap, "insan"/"makine")`). Her cevabı gösterip kullanıcıya `İnsan mı, makine mi? [i/m]` diye sor ve sonunda puanı göster.

- "Bilmiyorum ya, canım sıkılıyor bugün 😅" → insan
- "Sorunuzu anlamadım. Lütfen tekrar ifade eder misiniz?" → makine
- "Pizza mı? Kesinlikle ananaslı olmaz!" → insan
- "İstanbul'un nüfusu yaklaşık 15,6 milyondur." → makine

💡 **Beklenen çıktı:**
```text
Cevap: "Pizza mı? Kesinlikle ananaslı olmaz!"
İnsan mı, makine mi? [i/m]: i
Doğru!
Puan: 4/4
```
