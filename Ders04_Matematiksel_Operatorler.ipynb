## 1. Matematiksel Operatörler

| Operatör | Anlamı | Örnek | Sonuç |
|---|---|---|---|
| `+` | Toplama | `7 + 2` | `9` |
| `-` | Çıkarma | `7 - 2` | `5` |
| `*` | Çarpma | `7 * 2` | `14` |
| `/` | Bölme | `7 / 2` | `3.5` |
| `//` | Tam sayı bölme (bölüm) | `7 // 2` | `3` |
| `%` | Mod (kalan) | `7 % 2` | `1` |
| `**` | Üs alma | `7 ** 2` | `49` |

> 🔹 `/` her zaman **ondalıklı** sonuç verir: `4 / 2` → `2.0`
> 🔹 İşlem önceliği matematikteki gibidir. Emin değilsen **parantez** kullan.

### ✏️ Örnek 4.1: Hesap makinesi yapalım

`A = 17` ve `B = 5` tanımla. Yedi operatörün hepsiyle işlem yap ve sonuçları yazdır.

💡 **Beklenen çıktı:**
```text
17 + 5 = 22
17 - 5 = 12
17 * 5 = 85
17 / 5 = 3.4
17 // 5 = 3
17 % 5 = 2
17 ** 5 = 1419857
```

### ✏️ Örnek 4.2: 6 sayısının karesini bulalım

En az iki farklı yolla çöz.

💡 **Beklenen çıktı:**
```text
6'nın karesi: 36
```

### ✏️ Örnek 4.3: Parantez fark eder mi?

`2 + 3 * 4` ve `(2 + 3) * 4` işlemlerini yazdır. Sonuçlar neden farklı?

💡 **Beklenen çıktı:**
```text
14
20
```

### ✏️ Örnek 4.4: Değişkenin kendi değerini artıralım

`puan = 10` tanımla. Önce `puan = puan + 5`, sonra kısa yazımla `puan += 5` yap. Her adımda yazdır.

💡 **Beklenen çıktı:**
```text
15
20
```

## 2. Metinlerde `+` ve `*`

- `+` iki metni **birleştirir**: `"Mustafa " + "Kemal"` → `Mustafa Kemal`
- `*` metni **tekrarlar**: `"Ha" * 3` → `HaHaHa`
- Ama `"5" + "6"` sonucu `11` değil, **`56`** olur!

### ✏️ Örnek 4.5: İsmini 5 kez alt alta yazdıralım

İpucu: metnin sonuna `\n` ekle ve `*` kullan.

### ✏️ Örnek 4.6: Metinleri birleştirelim

`metin = "Mustafa " + "Kemal "` tanımla. Yazdırırken sonuna `"Atatürk"` ekle.

💡 **Beklenen çıktı:**
```text
Mustafa Kemal Atatürk
```

### ✏️ Örnek 4.7: Üçü arasındaki fark

Aşağıdakileri yazdır, sonucu önceden tahmin et:

`A = 5 + 6` · `B = "5+6"` · `C = "5" + "6"`

💡 **Beklenen çıktı:**
```text
11
5+6
56
```
