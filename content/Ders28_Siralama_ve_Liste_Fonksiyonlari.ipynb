## 1. Sıralama

| Yazım | Ne yapar? |
|---|---|
| `liste.sort()` | Listeyi **kendisini** küçükten büyüğe/alfabetik sıralar |
| `liste.sort(reverse=True)` | Büyükten küçüğe sıralar |
| `liste.reverse()` | Sırayı **ters çevirir** (sıralamaz!) |
| `sorted(liste)` | Orijinali bozmadan **yeni** sıralı liste döndürür |

### ✏️ Örnek 28.1: İsimleri sırala

`isimler = ["ziya", "yahya", "elif", "ayşe", "kemal", "çetin"]` listesini sırala. `çetin` nereye yerleşti? Neden?

💡 **Beklenen çıktı:**
```text
['ayşe', 'elif', 'kemal', 'yahya', 'ziya', 'çetin']
```

### ✏️ Örnek 28.2: Sayıları sırala

`say = [65, 45, 65.5, 4, 0.15, 156, 98.5, 1]` listesini önce küçükten büyüğe, sonra büyükten küçüğe sırala.

💡 **Beklenen çıktı:**
```text
[0.15, 1, 4, 45, 65, 65.5, 98.5, 156]
[156, 98.5, 65.5, 65, 45, 4, 1, 0.15]
```

### ✏️ Örnek 28.3: reverse sıralamaz!

`sayilar = [30, 10, 50, 20]` listesine `reverse()` uygula. Sonuç büyükten küçüğe mi?

💡 **Beklenen çıktı:**
```text
[20, 50, 10, 30]
```

### ✏️ Örnek 28.4: sorted orijinali bozmaz

`notlar = [70, 95, 40, 85]` için `sirali = sorted(notlar)` yap. Hem `notlar`'ı hem `sirali`'yı yazdır.

💡 **Beklenen çıktı:**
```text
[70, 95, 40, 85]
[40, 70, 85, 95]
```

## 2. Liste Özeti: `len`, `sum`, `min`, `max`

| Fonksiyon | Sonuç |
|---|---|
| `len(liste)` | Eleman sayısı |
| `sum(liste)` | Sayıların toplamı |
| `min(liste)` / `max(liste)` | En küçük / en büyük |

**Ortalama = `sum(liste) / len(liste)`**

### ✏️ Örnek 28.5: Liste karnesi

`sayilar = [10, 5, 20, 8, 15, 6, 10, 45, 52, 65]` için en büyük, en küçük, toplam, eleman sayısı ve ortalamayı yazdır.

💡 **Beklenen çıktı:**
```text
En büyük: 65
En küçük: 5
Toplam: 236
Eleman sayısı: 10
Ortalama: 23.6
```

### ✏️ Örnek 28.6: Tek sayılar listesi

1–15 arası tek sayılardan bir liste oluştur. Küçükten büyüğe ve büyükten küçüğe yazdır.

💡 **Beklenen çıktı:**
```text
[1, 3, 5, 7, 9, 11, 13, 15]
[15, 13, 11, 9, 7, 5, 3, 1]
```

## 3. `enumerate()`: Sıra Numarasıyla Gezinmek

```text
for sira, eleman in enumerate(liste, start=1):
    print(sira, eleman)
```
Her turda hem **sıra numarasını** hem **elemanı** verir.

### ✏️ Örnek 28.7: Haftanın günleri

`gun` listesindeki günleri `1. gün: Pazartesi` şeklinde numaralandırarak yazdır.

💡 **Beklenen çıktı:**
```text
1. gün: Pazartesi
2. gün: Salı
...
```

### ✏️ Örnek 28.8: Numaradan güne

Kullanıcıdan gün numarası (1–7) al ve günün adını yazdır. `enumerate` gerekmeden, indeksle tek satırda çözebilir misin?

💡 **Beklenen çıktı:**
```text
Gün numarası giriniz: 3
Çarşamba
```

## 📝 Ödev 11 – Listeler

Kullanıcının girdiği bir cümleyi harf listesine dönüştürüp içindeki **sesli harfleri çıkaran** ve kalan listeyi yazdıran kodu yazınız.

💡 İpucu: Bir listeyi `for` ile dolaşırken aynı listeden eleman silmek elemanların atlanmasına yol açabilir. Kalanları **yeni bir listeye** eklemeyi dene.

💡 **Örnek çıktı:**
```text
Bir cümle giriniz: saat
['s', 't']
```

## ✅ Bugün Öğrendiklerimiz
- `sort()` listeyi değiştirir, `sorted()` yeni liste döndürür; `reverse()` sadece ters çevirir.
- Ortalama = `sum(liste) / len(liste)`.
- `enumerate(liste, start=1)` sıra numarası ve elemanı birlikte verir.
