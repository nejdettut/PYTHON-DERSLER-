## 2. Öneri Sistemi Nasıl Çalışır?

**Fikir:** "Senin gibi zevkleri olan kişiler bunu da sevdi."

```text
1) Her kullanıcının neyi beğendiğini kaydet          (veri)
2) Sana en çok benzeyen kullanıcıyı bul               (benzerlik)
3) Onun beğendiği ama senin henüz izlemediğin şeyleri öner
```

Benzerliği ölçmenin en basit yolu: **Ortak beğenilen film sayısı** → iki kümenin **kesişimi** (Ders 35)!

Önce şu hazır hücreyi çalıştır:

````python
begeniler = {
    "Ali":    {"Yüzüklerin Efendisi", "Harry Potter", "Matrix", "Star Wars"},
    "Ayşe":   {"Harry Potter", "Karlar Ülkesi", "Coco", "Ratatuy"},
    "Can":    {"Matrix", "Başlangıç", "Yıldızlararası", "Dune"},
    "Deniz":  {"Coco", "Ratatuy", "Oyuncak Hikâyesi", "Karlar Ülkesi"},
    "Elif":   {"Yüzüklerin Efendisi", "Harry Potter", "Narnia", "Hobbit"},
}
print("Kullanıcılar:", list(begeniler.keys()))
````

### ✏️ Örnek 5.1: Ortak filmler

Ali ile Elif'in ve Ali ile Ayşe'nin ortak beğendiği filmleri ve ortak film sayısını yazdır.

💡 **Beklenen çıktı:**
```text
Ali & Elif: {'Harry Potter', 'Yüzüklerin Efendisi'} → 2 ortak
Ali & Ayşe: {'Harry Potter'} → 1 ortak
```

### ✏️ Örnek 5.2: En benzer kullanıcı

`en_benzer(kisi)` fonksiyonu, verilen kişiye en çok ortak filmi olan **diğer** kullanıcıyı döndürsün. Herkes için dene.

💡 **Beklenen çıktı:**
```text
Ali → Elif
Ayşe → Deniz
Can → Ali
Deniz → Ayşe
Elif → Ali
```

### ✏️ Örnek 5.3: Öneri üret

`oner(kisi)` fonksiyonu en benzer kullanıcının beğendiği ama kişinin **henüz izlemediği** filmleri döndürsün (küme farkı!).

💡 **Beklenen çıktı:**
```text
Ali için öneriler: {'Hobbit', 'Narnia'}
```

### ✏️ Örnek 5.4: Yeni kullanıcı

Kullanıcıdan adını ve virgülle ayrılmış en az 3 film al, sözlüğe ekle ve ona öneri yap.

💡 **Beklenen çıktı:**
```text
Adın: Mert
Beğendiğin filmler: Coco, Ratatuy, Cars
Mert için öneriler: {'Harry Potter', 'Karlar Ülkesi'}
```
