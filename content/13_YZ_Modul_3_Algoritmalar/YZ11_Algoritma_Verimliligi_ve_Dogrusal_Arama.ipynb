## 1. Algoritma ve Verimlilik

**Algoritma**, bir problemi çözmek için adım adım talimatlardır (Ders 1'deki çay demleme algoritmasını hatırla!). Yapay zekâ modelleri de birer algoritmadır: Veriden nasıl öğrenileceğini adım adım tarif ederler.

Aynı problemi çözen iki algoritmadan biri saniyede, diğeri saatlerce çalışabilir. Bu yüzden şunu sorarız:

> 🔹 **Veri 10 kat büyürse algoritmam kaç kat yavaşlar?**

| Gösterim | Anlamı | Örnek |
|---|---|---|
| O(1) | Veri büyüse de süre aynı | Listenin ilk elemanına bakmak |
| O(log n) | Çok yavaş artar | İkili arama (YZ-12) |
| O(n) | Veriyle **aynı oranda** artar | Doğrusal arama |
| O(n²) | Veri 10 kat → süre **100 kat** | Kabarcık sıralama (YZ-13) |
| O(2ⁿ) | Patlar! | Tüm olasılıkları denemek (YZ-14) |

## 2. Doğrusal Arama (Linear Search)

Listenin başından başlayıp aradığın elemanı bulana kadar **tek tek** bakmak. Bulursan indeksini, bulamazsan `-1` döndür.

### ✏️ Örnek 11.1: Doğrusal arama fonksiyonu

`dogrusal_ara(liste, hedef)` fonksiyonunu `for` ve `range(len(liste))` ile yaz. `[3, 5, 2, 9]` içinde 2'yi ve 7'yi ara.

💡 **Beklenen çıktı:**
```text
2
-1
```

### ✏️ Örnek 11.2: Adımları say

Fonksiyonu, kaç karşılaştırma yaptığını da döndürecek şekilde değiştir: `return indeks, adim`. 1–100 arası sayılardan oluşan listede 1'i, 50'yi ve 100'ü ara.

💡 **Beklenen çıktı:**
```text
1 → 1 adım
50 → 50 adım
100 → 100 adım
```

### ✏️ Örnek 11.3: En kötü durum

Listede **olmayan** bir sayıyı aradığında kaç adım gerekir? 10, 100, 1000 ve 10000 elemanlı listelerle dene.

💡 **Beklenen çıktı:**
```text
10 eleman → 10 adım
100 eleman → 100 adım
1000 eleman → 1000 adım
10000 eleman → 10000 adım
```
