## 1. Veri Tipleri

Her değişkenin bir **veri tipi** vardır. Python tipi **kendisi belirler**, bizim yazmamız gerekmez.

| Tip | Anlamı | Örnek |
|---|---|---|
| `int` | Tam sayı | `3`, `-12` |
| `float` | Ondalıklı sayı | `5.5`, `2.0` |
| `complex` | Karmaşık sayı | `3 + 4j` |
| `bool` | Mantıksal değer | `True`, `False` |
| `str` | Metin | `"Merhaba"` |

veritipleri.webp

Bir verinin tipini öğrenmek için `type()` kullanılır.

### ✏️ Örnek 5.1: Tam sayı ve ondalıklı sayı

`A = 3` ve `B = 5.5` tanımla, ikisinin de tipini yazdır.

💡 **Beklenen çıktı:**
```text
<class 'int'>
<class 'float'>
```

### ✏️ Örnek 5.2: Sınıfa soru: `4 // 2` ile `4 / 2` aynı tip mi?

İki işlemin sonucunu ve tipini yazdır. Önce tahmin et!

💡 **Beklenen çıktı:**
```text
2 <class 'int'>
2.0 <class 'float'>
```

### ✏️ Örnek 5.3: Virgül mü, nokta mı?

`s1 = 10,56` ve `s2 = 10.56` tanımla. İkisini ve tiplerini yazdır. Sonra ikisini de `* 2` ile çarp.

> ⚠️ Ondalık ayracı Python'da **nokta**dır. Virgül kullanırsan iki ayrı değerden oluşan bir **demet (tuple)** olur.

### ✏️ Örnek 5.4: Karmaşık sayı (kısaca)

`z = 3 + 4j` tanımla. `z`, `z.real` ve `z.imag` değerlerini yazdır.

💡 **Beklenen çıktı:**
```text
(3+4j)
3.0
4.0
```

## 2. Bool Tipi ve Karşılaştırma Operatörleri

Karşılaştırma yapınca sonuç **sayı değil**, ya **`True`** (doğru) ya da **`False`** (yanlış) olur.

| Operatör | Soru |
|---|---|
| `>` | Büyük mü? |
| `<` | Küçük mü? |
| `>=` | Büyük veya eşit mi? |
| `<=` | Küçük veya eşit mi? |
| `==` | Eşit mi? |
| `!=` | Eşit değil mi? |

> ⚠️ **En çok karıştırılan nokta:**
> `=` → **atama** yapar (değer koyar)
> `==` → **karşılaştırma** yapar (eşit mi diye sorar)

### ✏️ Örnek 5.5: İki sayıyı karşılaştıralım

`A = 3`, `B = 5.5` tanımla. `A < B`, `A > B`, `A == B` ve `A != B` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
True
False
False
True
```

### ✏️ Örnek 5.6: Sonucun tipi ne?

`print(type(3 < 5))` yazıp çalıştır.

💡 **Beklenen çıktı:**
```text
<class 'bool'>
```

### ✏️ Örnek 5.7: Not geçer mi?

`not_ortalamasi = 62` tanımla. Ortalamanın 50'ye eşit veya büyük olup olmadığını yazdır.

💡 **Beklenen çıktı:**
```text
Geçti mi? True
```

### ✏️ Örnek 5.8: Metinler de karşılaştırılır

`sifre = "python123"` tanımla. Girilen `"Python123"` ile eşit mi? Sonucu yazdır.

💡 **Beklenen çıktı:**
```text
False
```

## 3. Mantıksal Operatörler

Birden fazla karşılaştırmayı **birlikte** değerlendirmek için kullanılır.

| Operatör | Ne zaman `True`? |
|---|---|
| `and` (ve) | **Hepsi** doğruysa |
| `or` (veya) | **En az biri** doğruysa |
| `not` (değil) | Sonucu **tersine** çevirir |

> 📌 Bunların gerçek gücünü **Modül 3: Koşullu İfadeler (if)** konusunda göreceğiz.

### ✏️ Örnek 5.9: Ehliyet alabilir mi?

`yas = 19` ve `sinav_gecti = True` tanımla. Ehliyet için **yaş 18 veya üstü VE sınavı geçmiş** olmak gerekir. Sonucu yazdır.

💡 **Beklenen çıktı:**
```text
Ehliyet alabilir: True
```

### ✏️ Örnek 5.10: Hafta sonu mu?

`gun = "Pazar"` tanımla. Gün Cumartesi **veya** Pazar mı? Sonucu yazdır.

💡 **Beklenen çıktı:**
```text
Hafta sonu: True
```

### ✏️ Örnek 5.11: `not` ile tersine çevirelim

`yagmur_yagiyor = False` tanımla. `not yagmur_yagiyor` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
True
```

## 4. Hangi Değerler `False` Sayılır?

`bool()` fonksiyonu bir değerin doğru mu yanlış mı sayıldığını gösterir:
- **`0`** → `False`, sıfır dışındaki tüm sayılar → `True`
- **Boş metin `""`** → `False`, dolu metin → `True`
- Boş liste, demet, sözlük ve küme de `False` sayılır *(ileride göreceğiz)*.

### ✏️ Örnek 5.12: Doğru mu, yanlış mı?

`bool(10)`, `bool(0)`, `bool(-5)`, `bool("")` ve `bool("Merhaba")` sonuçlarını yazdır. Önce tahmin et!

💡 **Beklenen çıktı:**
```text
True
False
True
False
True
```

💡 **Örnek çıktı:**
