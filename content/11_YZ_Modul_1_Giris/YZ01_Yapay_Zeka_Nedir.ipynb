## 1. Yapay Zekâ Nedir?

**Yapay zekâ (YZ)**, bilgisayarların normalde insan zekâsı gerektiren işleri yapabilmesini sağlayan bilim ve teknoloji dalıdır: **öğrenmek, problem çözmek, karar vermek, algılamak.**

> 🔹 Yapay zekâ "düşünmez". Veriyi analiz eder, içindeki **kalıpları** bulur ve buna göre en olası sonucu üretir.

| Günlük hayattan örnek | Yapay zekâ ne yapıyor? |
|---|---|
| Sesli asistanlar (Siri, Google Asistan) | Sesi yazıya çevirip komutu anlıyor |
| Çeviri uygulamaları | Cümleyi başka dile çeviriyor |
| Netflix / YouTube önerileri | İzleme geçmişine göre içerik öneriyor |
| Telefonda yüz tanıma | Yüzünü tanıyıp kilidi açıyor |
| E-posta spam filtresi | İstenmeyen e-postaları ayırıyor |

## 2. Kural Tabanlı Program

Bugüne kadar yazdığımız programlar **kural tabanlıydı**: Kuralları **biz** yazdık, bilgisayar uyguladı.

```text
Kurallar (biz yazarız) + Veri  →  Program  →  Cevap
```

Hadi kural tabanlı bir **spam filtresi** yazalım!

### ✏️ Örnek 1.1: Şüpheli kelime listesi

`supheli = ["kazandınız", "bedava", "tıkla", "acil", "ödül"]` listesini oluştur. Bir mesajın içinde bu kelimelerden biri geçiyorsa `SPAM`, geçmiyorsa `Normal` yazdır. Mesajı `.lower()` ile küçük harfe çevirmeyi unutma.

💡 **Beklenen çıktı:**
```text
Mesaj: Tebrikler! Ödül KAZANDINIZ, hemen tıkla
SPAM
```

### ✏️ Örnek 1.2: Filtreyi fonksiyona çevir

`spam_mi(mesaj)` fonksiyonu `True`/`False` döndürsün. Aşağıdaki mesaj listesini filtreden geçir ve sonuçları yazdır:

```text
mesajlar = ["Yarın matematik sınavı var", "BEDAVA telefon için tıkla",
            "Annen aradı, eve gel", "Acil! Hesabınız kapanacak",
            "Bu akşam maç var mı?"]
```

💡 **Beklenen çıktı:**
```text
Yarın matematik sınavı var → Normal
BEDAVA telefon için tıkla → SPAM
Annen aradı, eve gel → Normal
Acil! Hesabınız kapanacak → SPAM
Bu akşam maç var mı? → Normal
```

### ✏️ Örnek 1.3: Filtreyi kandır!

Filtrenin **kaçırdığı** bir spam mesajı ve yanlışlıkla spam dediği **normal** bir mesaj bul. Örnek: `"Bedava"` yerine `"b3dava"` yazmak, ya da `"Acil toplantı: öğretmenler odasına gel"`.

## 3. Kurallar Neden Yetmez?

Spamcılar her gün yeni hileler bulur. Her yeni hile için **elle kural eklemek** imkânsız hâle gelir.

**Yapay zekâ yaklaşımı (makine öğrenmesi):**
```text
Veri (örnekler) + Cevaplar (etiketler)  →  Öğrenme  →  Kurallar (model)
```
Binlerce "spam" ve "normal" e-posta örneği gösterilir; sistem kuralları **kendisi** çıkarır.

| | Kural tabanlı | Veriden öğrenen (YZ) |
|---|---|---|
| Kuralları kim yazar? | Programcı | Sistem veriden çıkarır |
| Yeni durumlara uyum | Zor, elle güncellenir | Yeni veriyle yeniden öğrenir |
| Örnek | Hesap makinesi, Ders 13'teki karar ağacı | Yüz tanıma, çeviri, öneri sistemleri |

### ✏️ Örnek 1.4: Filtre veriden öğrensin (önizleme)

Aşağıdaki **etiketli** mesajlardan, spam mesajlarda geçen ama normal mesajlarda **hiç geçmeyen** kelimeleri bulup yeni bir şüpheli kelime listesi oluştur. İpucu: `split()` ve kümeler (Ders 35)!

```text
spamlar = ["bedava hediye kazan", "hemen tıkla kazan", "bedava tatil fırsatı"]
normaller = ["yarın okulda görüşürüz", "ödevi hemen gönder", "tatil planı yapalım"]
```

💡 **Beklenen çıktı:**
```text
Öğrenilen şüpheli kelimeler: {'bedava', 'fırsatı', 'hediye', 'kazan', 'tıkla'}
```

> 🎉 Tebrikler: Kuralları sen yazmadın, program **veriden çıkardı**. Bu, makine öğrenmesinin en basit fikri!

### 🎯 Kendini Dene 1: Bu Yapay Zekâ mı?

Aşağıdaki sistemleri bir sözlükte tut (`"sistem": True/False`). Kullanıcıya her birini sorup `E/H` cevabını sözlükle karşılaştır, sonunda puanını göster.

Hesap makinesi (H) · Yüz tanıma (E) · Trafik ışığı zamanlayıcısı (H) · Sesli asistan (E) · Çeviri uygulaması (E) · Alarm saati (H)

💡 **Örnek çıktı:**
```text
Hesap makinesi yapay zekâ mı? [E/H]: h
Doğru!
...
Puanın: 5/6
```
