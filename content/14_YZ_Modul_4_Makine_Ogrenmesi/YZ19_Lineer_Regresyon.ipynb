## 2. Lineer Regresyon

Veriye en iyi uyan **doğruyu** bulur:
```text
fiyat = eğim × büyüklük + kesişim
y     =   a  ×    x     +    b
```
Model `fit` sırasında hatayı en aza indiren `a` ve `b` değerlerini bulur (YZ-15'teki **gradyan inişini** hatırla!).

Önce ev verisini yükle:

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Ev büyüklüğü (m²) ve fiyatı (TL)
X = np.array([50, 70, 90, 120, 150]).reshape(-1, 1)   # scikit-learn 2 boyutlu tablo ister
y = np.array([1500000, 2000000, 2500000, 3200000, 4000000])
print("Veri hazır:", X.shape, y.shape)
````

### ✏️ Örnek 19.1: Veriyi çiz

Ev büyüklüğü ile fiyatı saçılım grafiğinde göster. İlişki doğrusal mı?

### ✏️ Örnek 19.2: Modeli eğit

`LinearRegression()` modelini oluştur, `fit(X, y)` ile eğit. Öğrenilen eğimi (`model.coef_[0]`) ve kesişimi (`model.intercept_`) yazdır.

💡 **Beklenen çıktı:**
```text
Eğim: 24810.13 TL/m²
Kesişim: 258,228 TL
```

> 🔹 **Eğim yorumu:** Ev 1 m² büyüdükçe fiyat ortalama eğim kadar artar.

### ✏️ Örnek 19.3: Doğruyu çiz

Saçılım grafiğinin üzerine modelin tahmin doğrusunu (`model.predict(X)`) kırmızı çizgiyle ekle.

### ✏️ Örnek 19.4: Kullanıcının evi

Kullanıcıdan ev büyüklüğünü al, tahmini fiyatı binlik ayraçla yazdır (`{fiyat:,.0f}`). Mesajda girilen büyüklük görünsün!

💡 **Beklenen çıktı:**
```text
Evin büyüklüğü (m²): 100
100 m² evin tahmini fiyatı: 2,739,241 TL
```

### ✏️ Örnek 19.5: Elle hesapla

Aynı tahmini `a * x + b` formülüyle **elle** hesapla. `predict` ile aynı mı?

## 3. Çalışma Saatinden Puan Tahmini 📚

YZ Modülü 2'deki öğrenci veri setimizde çalışma saati ile puan arasında güçlü bir ilişki bulmuştuk. Şimdi bu ilişkiyi bir **modele** çevirelim!

````python
import pandas as pd
veri = {"calisma_saati": [2, 6, 4, 8, 1, 7, 3, 6, 3, 9, 2, 4, 6, 7, 3, 8, 1, 5],
        "puan":          [55, 88, 70, 95, 40, 90, 62, 85, 58, 98, 50, 75, 82, 92, 60, 93, 35, 78]}
df = pd.DataFrame(veri)
print(df.shape)
````

### ✏️ Örnek 19.6: Eğit ve test et

`X = df[["calisma_saati"]]`, `y = df["puan"]`. Veriyi %25 test olacak şekilde böl (`random_state=1`), modeli eğit, test setindeki tahminlerin **R² skorunu** hesapla (`from sklearn.metrics import r2_score`).

**R²:** 1'e ne kadar yakınsa model o kadar iyi açıklıyor demektir.

💡 **Beklenen çıktı:**
```text
R² skoru: 0.97
```

### ✏️ Örnek 19.7: Yorumla

Modelin eğimini yorumla: Haftada 1 saat fazla çalışmak puanı ortalama kaç artırıyor? 12 saat çalışan biri için tahmin ne? Bu mantıklı mı?

💡 **Beklenen çıktı:**
```text
Her ek çalışma saati ≈ 7.7 puan
12 saat için tahmin: 128.3 puan ← 100'ü aştı!
```

### 💬 Tartışalım

19.7'de model 100'den büyük puan tahmin etti. Neden? Model, eğitim verisinde görmediği bir aralıkta (12 saat) tahmin yaparken nelere dikkat etmeliyiz?
