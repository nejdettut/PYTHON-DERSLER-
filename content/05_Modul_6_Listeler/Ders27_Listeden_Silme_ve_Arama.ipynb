## 1. Silme Yöntemleri

| Yöntem | Neye göre siler? | Örnek |
|---|---|---|
| `pop()` | Son elemanı | `liste.pop()` |
| `pop(i)` | **İndekse** göre | `liste.pop(1)` |
| `remove(x)` | **Değere** göre (ilk bulduğunu) | `liste.remove("nar")` |
| `del liste[i]` | İndekse veya dilime göre | `del liste[0:2]` |
| `clear()` | Tüm elemanları siler, liste **boş** kalır | `liste.clear()` |
| `del liste` | Listeyi **tamamen** yok eder | |

### ✏️ Örnek 27.1: pop

`meyveler = ["ayva", "nar", "kiraz", "kayısı", "üzüm"]` listesinde önce `pop()`, sonra `pop(1)` uygula. Her adımda yazdır. `pop()` sildiği elemanı **geri verir**; onu da yazdır.

💡 **Beklenen çıktı:**
```text
['ayva', 'nar', 'kiraz', 'kayısı']
Silinen: üzüm
['ayva', 'kiraz', 'kayısı']
```

### ✏️ Örnek 27.2: remove

Aynı listeden `"nar"` değerini sil. Sonra listede olmayan `"elma"`yı silmeyi dene.

> ⚠️ İkinci deneme `ValueError` verecek.

### ✏️ Örnek 27.3: del ve clear

`takimlar = ["FB", "GS", "BJK", "TS"]` listesinde: `del takimlar[2]`, ardından `del takimlar[0:2]`, ardından `clear()`. Her adımda yazdır.

💡 **Beklenen çıktı:**
```text
['FB', 'GS', 'TS']
['TS']
[]
```

## 2. Sayma ve Arama

- `liste.count(x)` → `x` listede **kaç kez** geçiyor?
- `liste.index(x)` → `x`'in **ilk** bulunduğu indeks (yoksa hata verir, önce `in` ile kontrol et!)

### ✏️ Örnek 27.4: count

`sehir = ["konya", "karaman", "kocaeli", "karaman", "kars", "kayseri", "kars", "karaman"]` listesinde karaman, kars ve konya kaç kez geçiyor?

💡 **Beklenen çıktı:**
```text
3
2
1
```

### ✏️ Örnek 27.5: index

Aynı listede `"kars"`ın ilk indeksini, sonra 5. indeksten itibaren aramaya başlayarak `index("kars", 5)` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
4
6
```

### ✏️ Örnek 27.6: Sayı arama

`A = list(range(1, 21))` oluştur. Kullanıcıdan sayı al; listede varsa indeksini, yoksa `bulunamadı` yazdır.

💡 **Beklenen çıktı:**
```text
Aranacak sayı: 15
Aranan sayı 14. indiste bulundu.
```

## 3. Tümünü Silmek

### ✏️ Örnek 27.7: Bütün karaman'ları sil

`sehir` listesindeki **tüm** `"karaman"`ları sil. İki yol dene: (1) `count` kadar `remove` yap, (2) `while "karaman" in sehir:` döngüsü kur.

💡 **Beklenen çıktı:**
```text
Listede 3 tane karaman var.
['konya', 'kocaeli', 'kars', 'kayseri', 'kars']
```

### ✏️ Örnek 27.8: Cümledeki a harflerini sil

Kullanıcıdan cümle al, `list()` ile harf listesine çevir. Kaç tane `a` olduğunu yaz, sonra hepsini sil ve listeyi yazdır.

💡 **Beklenen çıktı:**
```text
Bir cümle giriniz: ankara
Metinde 3 adet a harfi vardır
['n', 'k', 'r']
```

### 🎯 Kendini Dene 1: Kara Liste

`davetliler` listesine 6 isim yaz. Kullanıcıdan gelmeyecek kişinin adını al; listede varsa sil ve `X davetliler listesinden çıkarıldı` yaz, yoksa `Böyle biri listede yok` yaz. Son listeyi göster.

---
## ✅ Bugün Öğrendiklerimiz
- `pop(i)` indekse, `remove(x)` değere göre siler; `clear()` boşaltır, `del` yok eder.
- `count(x)` sayar, `index(x)` yerini bulur; `index` öncesinde `in` ile kontrol iyi alışkanlıktır.
- Tekrar eden elemanların hepsini silmek için `while x in liste: liste.remove(x)` kalıbı kullanılır.
