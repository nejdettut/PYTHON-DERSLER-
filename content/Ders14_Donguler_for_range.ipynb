## 1. Neden Döngü?

1'den 100'e kadar sayıları yazdırmak için 100 tane `print()` mı yazmalıyız? Hayır!
**Döngü**, bir kod bloğunun istenen sayıda **tekrar** edilmesidir.

| Döngü | Ne zaman? |
|---|---|
| `for` | Tekrar sayısı **baştan belli** ise |
| `while` | Tekrar bir **koşula bağlı** ise (Ders 17) |

### ✏️ Örnek 14.1: Döngüsüz yol

`"Merhaba Dünyalı"` yazısını **5 ayrı `print()`** ile 5 kez yazdır. Peki 1000 kez yazdırman gerekseydi?

## 2. `for` Döngüsü

```text
for değişken in sıra:
    tekrarlanacak kodlar
```

- `for` satırının sonunda **`:`** ve altında **girinti** vardır (tıpkı `if` gibi).
- Döngü değişkeni (`i`, `harf` vb.) her turda sıradaki elemanı alır.

### ✏️ Örnek 14.2: for ile 5 kez yazdıralım

`for i in range(5):` kullanarak aynı yazıyı 5 kez yazdır. Sonra döngünün içine `print(i)` ekle. `i` hangi değerleri aldı?

💡 **Beklenen çıktı:**
```text
Merhaba Dünyalı
0
Merhaba Dünyalı
1
...
```

### ✏️ Örnek 14.3: Metnin harfleri üzerinde döngü

`for harf in "PYTHON":` ile her harfi alt alta yazdır.

💡 **Beklenen çıktı:**
```text
P
Y
T
H
O
N
```

## 3. `range()` Fonksiyonu

`range(başlangıç, bitiş, adım)` bir **sayı dizisi** üretir.

| Yazım | Üretilen sayılar | Not |
|---|---|---|
| `range(7)` | 0, 1, 2, 3, 4, 5, 6 | Başlangıç verilmezse **0** |
| `range(1, 7)` | 1, 2, 3, 4, 5, 6 | **Bitiş dahil değil!** |
| `range(0, 12, 2)` | 0, 2, 4, 6, 8, 10 | İkişer artar |
| `range(7, 0, -1)` | 7, 6, 5, 4, 3, 2, 1 | **Geriye** sayar |

> 🔹 Dilimlemedeki `[başlangıç:bitiş:adım]` kuralıyla aynı mantık!

### ✏️ Örnek 14.4: `range(7)` ve `range(1, 7)`

İkisini ayrı döngülerle yazdır. Farkı ne?

### ✏️ Örnek 14.5: Adımlı range

0'dan 12'ye kadar ikişer ikişer yazdır. 12 neden yok?

💡 **Beklenen çıktı:**
```text
0
2
4
6
8
10
```

### ✏️ Örnek 14.6: Geriye doğru sayalım

Önce `range(7, 0, -1)`, sonra `range(15, -5, -3)` ile yazdır. Sonuçları önce tahmin et!

💡 **Beklenen çıktı:**
```text
15
12
9
6
3
0
-3
```

### ✏️ Örnek 14.7: Tahmin et

`range(0)` ve `range(-5, 5)` döngülerinin çıktısını önce tahmin et, sonra çalıştır.

### ✏️ Örnek 14.8: Kullanıcının seçtiği aralık

Kullanıcıdan başlangıç, bitiş ve artış değerlerini al, bu aralıktaki sayıları yazdır.

💡 **Beklenen çıktı:**
```text
Başlangıç: 3
Bitiş: 15
Artış: 4
3
7
11
```

## ✅ Bugün Öğrendiklerimiz
- `for` döngüsü, tekrar sayısı belli olduğunda kullanılır.
- `range(başlangıç, bitiş, adım)` → bitiş **dahil değildir**; negatif adım geriye sayar.
- `for harf in metin:` metnin her karakterini sırayla verir.
