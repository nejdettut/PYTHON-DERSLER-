## 2. Deney: Önyargılı İşe Alım Modeli ⚖️

Bir şirket geçmiş 10 yıllık işe alım kararlarıyla bir model eğitiyor. Ama geçmişte bir gruptan neredeyse hiç kimse işe alınmamış. Model ne öğrenir?

> 2018'de büyük bir teknoloji şirketi, geçmiş verilerle eğittiği işe alım modelinin kadın adaylara karşı önyargılı olduğunu fark edip sistemi kullanmaktan vazgeçti. Biz de benzer bir durumu **hayalî** verilerle canlandıracağız.

Önce hayalî veriyi oluştur: `deneyim` (yıl), `puan` (mülakat puanı), `grup` (A=0, B=1) ve geçmiş karar `alindi`.

````python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
n = 400
deneyim = rng.integers(0, 11, n)
puan = rng.integers(40, 101, n)
grup = rng.choice([0, 1], n)               # 0 = A grubu, 1 = B grubu (hayalî)
yetkin = (deneyim >= 4) & (puan >= 70)      # gerçekten işe uygun mu?

# Geçmişteki ÖNYARGILI kararlar: B grubundan yetkin adayların sadece %20'si alınmış
alindi = np.where(grup == 0, yetkin, yetkin & (rng.random(n) < 0.2)).astype(int)

gecmis = pd.DataFrame({"deneyim": deneyim, "puan": puan, "grup": grup, "alindi": alindi})
print(gecmis.head())
````

### ✏️ Örnek 20.1: Veriyi incele

Her grupta işe alınma oranını hesapla (`groupby("grup")["alindi"].mean()`). Sonra sadece **yetkin** adaylar (deneyim ≥ 4 ve puan ≥ 70) için aynı oranı hesapla. Ne fark ediyorsun?

### ✏️ Örnek 20.2: Modeli eğit

`deneyim`, `puan` ve `grup` sütunlarıyla bir `DecisionTreeClassifier(max_depth=4, random_state=0)` modeli eğit (`from sklearn.tree import DecisionTreeClassifier`). Eğitim doğruluğunu yazdır.
