## 1. Duygu Analizi 😊😞

Bir metnin **olumlu mu olumsuz mu** olduğunu otomatik bulmak. Şirketler müşteri yorumlarını, sosyal medya paylaşımlarını bu yolla analiz eder.

YZ-6'da bunu kelime listeleriyle (kural tabanlı) denemiştik ve "fena değil" gibi ifadelerde zorlanmıştık. Şimdi model **örneklerden öğrenecek**.

**Tarif:**
```text
yorumlar ──CountVectorizer──▶ sayı vektörleri ──MultinomialNB──▶ olumlu/olumsuz
```
**Naive Bayes**, her kelimenin olumlu ve olumsuz yorumlarda ne sıklıkla geçtiğini öğrenen, metinler için hızlı ve basit bir modeldir.

````python
# Okul kantini hakkında öğrenci yorumları (1 = olumlu, 0 = olumsuz). Sadece çalıştır!
yorumlar = [
    ("tostlar çok lezzetli ve sıcak", 1), ("fiyatlar çok pahalı", 0),
    ("çalışanlar çok güler yüzlü", 1), ("sıra çok uzun bekledim", 0),
    ("ayran soğuk ve taze", 1), ("simit bayattı hiç beğenmedim", 0),
    ("kantin temiz ve düzenli", 1), ("masalar kirli ve dağınık", 0),
    ("poğaça harika olmuş", 1), ("çay soğuktu berbat", 0),
    ("hizmet çok hızlı", 1), ("personel kaba davrandı", 0),
    ("menü çok çeşitli güzel", 1), ("porsiyonlar çok küçük", 0),
    ("meyve suyu taze ve lezzetli", 1), ("tost yanmıştı beğenmedim", 0),
    ("fiyatlar uygun ve makul", 1), ("ekmek bayat ve kuru", 0),
    ("sandviç çok doyurucu", 1), ("çok gürültülü ve kalabalık", 0),
    ("kurabiyeler çok güzel", 1), ("su bile pahalı", 0),
    ("çalışanlar yardımsever", 1), ("yemekler soğuk geldi", 0),
    ("temizlik harika", 1), ("bekleme süresi çok uzun", 0),
    ("lezzetli ve sağlıklı seçenekler var", 1), ("kek kuru ve tatsızdı", 0),
    ("ortam çok ferah ve güzel", 1), ("para üstünü yanlış verdiler", 0),
    ("salata taze ve güzel", 1), ("çorba tuzluydu beğenmedim", 0),
    ("hızlı ve güler yüzlü hizmet", 1), ("kantin çok küçük ve sıkışık", 0),
    ("pizza dilimleri harika", 1), ("içecekler sıcaktı kötü", 0),
    ("fiyat performans çok iyi", 1), ("tezgah kirliydi", 0),
    ("öğrencilere indirim yapıyorlar güzel", 1), ("menüde sağlıklı yiyecek yok", 0),
]
print(len(yorumlar), "yorum yüklendi.")
````

### ✏️ Örnek 28.1: Veriyi ayır

Yorumları ve etiketleri iki ayrı listeye ayır. Kaç olumlu, kaç olumsuz yorum var? `train_test_split` ile %25 test olacak şekilde böl (`random_state=1`, `stratify=etiketler`).

### ✏️ Örnek 28.2: Vektörleştir ve eğit

`CountVectorizer` ile eğitim metinlerini dönüştür (`fit_transform`), test metinlerini sadece `transform` et. `MultinomialNB` modeli eğit ve test doğruluğunu yazdır.

> ⚠️ Test metinleri için asla `fit_transform` kullanma! Sözlük sadece eğitim verisinden kurulmalı.
