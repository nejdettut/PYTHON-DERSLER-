## 1. İki Tür Hata

| Tür | Ne zaman ortaya çıkar? | Örnek |
|---|---|---|
| **Yazım hatası** (SyntaxError) | Kod **hiç çalışmadan** önce | `:` unutmak, tırnağı kapatmamak |
| **Çalışma zamanı hatası** (Exception) | Kod **çalışırken**, beklenmedik bir durumda | Sıfıra bölmek, sayı yerine harf girmek |

### Hata Mesajı Nasıl Okunur?

```text
ZeroDivisionError: division by zero
└── hata türü ──┘  └── açıklama ──┘
```
> 🔹 Hata mesajının **son satırına** bak: hatanın **türünü** ve **nedenini** söyler. Üstündeki satırlar hatanın **nerede** olduğunu gösterir.

## 2. Sık Karşılaşılan Hatalar

| Hata | Nedeni | Örnek |
|---|---|---|
| `SyntaxError` | Yazım kuralı ihlali | `if 3 > 2` (`:` yok) |
| `IndentationError` | Girinti hatası | `if` altında girintisiz satır |
| `NameError` | Tanımsız isim | `print(x)` (x yok) |
| `TypeError` | Uyumsuz tipler | `"5" + 3` |
| `ValueError` | Tip doğru ama değer uygun değil | `int("merhaba")` |
| `ZeroDivisionError` | Sıfıra bölme | `10 / 0` |
| `IndexError` | Olmayan indeks | `[1, 2, 3][5]` |
| `KeyError` | Sözlükte olmayan anahtar | `{"a": 1}["b"]` |
| `FileNotFoundError` | Olmayan dosya | `open("yok.txt")` |

### ✏️ Örnek 40.1: Girinti hatasını bul

Aşağıdaki kodu yaz, çalıştır, hatayı oku ve ikinci hücrede düzelt:
```text
while True:
    x = input("Bir sayı giriniz: ")
    if not x:
    break
    print(float(x) / 2)
```

### ✏️ Örnek 40.2: Yazım hataları

Her birini ayrı hücrede çalıştırıp hatanın türünü not al: `if 3 > 2 print("evet")` · `7isim = "zafer"` · `print("Merhaba)`

### ✏️ Örnek 40.3: Gömülü fonksiyonu ezmek

`print = "Hata ayıklama"` yaz, sonra `print("Merhaba")` çalıştır. Ne oldu? İkinci hücrede `del print` yazarak düzelt.

> ⚠️ `TypeError: 'str' object is not callable`. `print` artık bir metin! Değişkenlere gömülü fonksiyon adı verme.

### ✏️ Örnek 40.4: Çalışma zamanı hataları

Her birini ayrı hücrede dene ve hata türünü yaz: `10 / 0` · `int("merhaba")` · `"5" + 3` · `[1, 2, 3][5]`

## 3. Hatayı Önceden Engellemek

Kullanıcının ne gireceğini bilemeyiz. Ama olası hataları **öngörüp** `if` ve `while` ile engelleyebiliriz.

### ✏️ Örnek 40.5: Sıfıra bölmeyi engelle

İki sayı al ve böl. İkinci sayı 0 girilirse `0 dışında bir değer giriniz!` deyip **tekrar sor**.

💡 **Beklenen çıktı:**
```text
Birinci sayı: 10
İkinci sayı: 0
0 dışında bir değer giriniz!
İkinci sayı: 4
Sonuç: 2.5
```

### ✏️ Örnek 40.6: Geçersiz indeksi engelle

`takim = ["Galatasaray", "Fenerbahçe", "Beşiktaş", "Trabzonspor"]`. Kullanıcıdan indeks al; geçerli değilse uyarıp tekrar sor. İpucu: geçerli aralık `-len(takim)` ile `len(takim) - 1` arası.

💡 **Beklenen çıktı:**
```text
Bir indeks giriniz: 7
Geçersiz indeks! 0-3 arası giriniz.
Bir indeks giriniz: 2
Beşiktaş
```
