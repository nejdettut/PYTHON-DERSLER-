## 1. Veri: Yapay Zekânın Yakıtı ⛽

Yapay zekâ modelleri **veriden öğrenir**. Veri ne kadar **doğru, temiz ve çeşitliyse** model o kadar başarılı olur.

> 🍳 **Benzetme:** Veri, yemeğin malzemesidir. Bozuk malzemeyle iyi yemek yapılmaz.
> 🗑️ Bilgisayar biliminde bunun bir adı var: **Garbage In, Garbage Out** (Çöp girerse çöp çıkar).

## 2. Veri Türleri

| Tür | Özellik | Örnek |
|---|---|---|
| **Yapısal** | Satır ve sütunlu tablo düzeni | Excel, CSV, e-Okul not tablosu |
| **Yarı yapısal** | Etiketli ama esnek | JSON, XML (web servislerinden gelen veriler) |
| **Yapısal olmayan** | Belirli bir düzeni yok | Metin, fotoğraf, ses, video |

> 🔹 Dünyadaki verinin büyük çoğunluğu **yapısal olmayan** veridir. Bu yüzden görüntü ve dil işleme bu kadar önemli.

### ✏️ Örnek 6.1: Aynı bilgi, üç biçim

Aynı öğrencinin bilgisini üç biçimde tut:
- **Yapısal:** `["Ayşe", 16, 88]` gibi bir satır listesi (başlıklar ayrı listede)
- **Yarı yapısal:** sözlük (JSON benzeri)
- **Yapısal olmayan:** `"Ayşe 16 yaşında ve sınavdan 88 aldı."`

Her birinden öğrencinin **notunu** çekip yazdır. Hangisi en kolaydı?

### ✏️ Örnek 6.2: JSON ile tanışma

`import json` ile 6.1'deki sözlüğü `json.dumps(..., ensure_ascii=False)` ile JSON metnine çevir, yazdır. Sonra `json.loads()` ile tekrar sözlüğe çevir.

💡 **Beklenen çıktı:**
```text
{"isim": "Ayşe", "yas": 16, "puan": 88}
```

### ✏️ Örnek 6.3: Metinden veri çıkarmak zordur

`yorumlar = ["Film harikaydı, bayıldım!", "Çok sıkıcıydı.", "Fena değil ama uzundu", "Muhteşem bir yapım"]` listesindeki her yorumun **olumlu mu olumsuz mu** olduğunu, olumlu/olumsuz kelime listeleriyle tahmin et (YZ-1'deki spam filtresi gibi). Hangi yorumda zorlandın?

💡 **Beklenen çıktı:**
```text
Film harikaydı, bayıldım! → olumlu
Çok sıkıcıydı. → olumsuz
Fena değil ama uzundu → olumsuz   ← gerçekten öyle mi?
Muhteşem bir yapım → olumlu
```

## 3. Çöp Girerse Çöp Çıkar 🗑️

### ✏️ Örnek 6.4: Hatalı veriyle ortalama

`yaslar = [15, 16, 15, 17, -5, 16, 150, 15]` listesinin ortalamasını al. Sonra 0–100 aralığı dışındaki değerleri **ayıklayıp** tekrar hesapla. Fark ne kadar?

💡 **Beklenen çıktı:**
```text
Hatalı veriyle ortalama: 29.88
Temiz veriyle ortalama: 15.67
```

### ✏️ Örnek 6.5: Eksik veri

`puanlar = [70, None, 85, 90, None, 60]` listesinde `sum()` çalışır mı? Dene. Sonra `None` olmayanlarla ortalamayı hesapla.

> ⚠️ `sum()` `TypeError` verecek. Eksik veriler hesaplamayı bozar!

### ✏️ Örnek 6.6: Dengesiz veri

Bir kedi/köpek tanıma modeli için eğitim verisi: `["kedi"] * 95 + ["köpek"] * 5`. Her şeye `"kedi"` diyen "tembel" bir modelin doğruluğunu hesapla. Bu model iyi mi?

💡 **Beklenen çıktı:**
```text
Her şeye 'kedi' diyen model doğruluğu: %95.0
```

### 💬 Tartışalım

Örnek 6.6'daki model %95 doğrulukla çalışıyor ama hiç köpek tanıyamıyor. Bir yüz tanıma sistemi bazı ten renklerinden çok az örnekle eğitilirse ne olur? (İpucu: **önyargı / bias**)
