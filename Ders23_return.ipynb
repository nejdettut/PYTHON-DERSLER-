## 1. `print` mi, `return` mı?

| `print()` | `return` |
|---|---|
| Sonucu **ekrana yazar** | Sonucu **geri verir** (döndürür) |
| Sonuç başka yerde kullanılamaz | Sonuç değişkene atanabilir, işlemde kullanılabilir |
| `x = f()` → `x` **None** olur | `x = f()` → `x` sonucu tutar |

> 🔹 `int("5")` de bir değer döndürür; bu yüzden `sayi = int("5")` yazabiliyoruz.
> 🔹 `return` çalışınca fonksiyon **orada biter**.

### ✏️ Örnek 23.1: Farkı görelim

İki fonksiyon yaz: `topla_yaz(a, b)` sonucu **print** etsin, `topla_don(a, b)` sonucu **return** etsin. İkisinin sonucunu değişkenlere ata ve değişkenleri yazdır.

💡 **Beklenen çıktı:**
```text
8
x: None
y: 8
```

### ✏️ Örnek 23.2: Faktöriyel fonksiyonu

`faktoriyel(sayi)` fonksiyonu sonucu **döndürsün**. `a = faktoriyel(5)` ve `b = faktoriyel(5) + 120` değerlerini yazdır.

💡 **Beklenen çıktı:**
```text
120
240
```

### ✏️ Örnek 23.3: Tarla: return olmadan

`alan(u, g)` ve `cevre(u, g)` fonksiyonları sonuçları **ekrana yazsın**. Değerleri kullanıcıdan al.

💡 **Beklenen çıktı:**
```text
Uzunluk: 10
Genişlik: 5
Tarlanın alanı: 50
Tarlanın çevresi: 30
```

### ✏️ Örnek 23.4: Tarla: return ile

Aynı fonksiyonları sonucu **döndürecek** şekilde yeniden yaz ve `print()` içinde çağır.

### ✏️ Örnek 23.5: İki değer birden döndür

Tek bir `alan_cevre(u, g)` fonksiyonu `return alan, cevre` ile iki değer döndürsün. Sonuçları `a, c = alan_cevre(10, 5)` ile al.

💡 **Beklenen çıktı:**
```text
Alan: 50
Çevre: 30
```

### ✏️ Örnek 23.6: Tek mi çift mi? (return)

`tek_cift(sayi)` fonksiyonu `"Tek"` ya da `"Çift"` metnini **döndürsün**. Sonucu f-string içinde kullan.

💡 **Beklenen çıktı:**
```text
Bir sayı girin: 14
Girdiğiniz 14 Çift sayıdır.
```

## 📝 Ödev 10 – Fonksiyonlar

Klavyeden girilen Fahrenayt değerini Santigrat dereceye çeviren `cevir(f)` fonksiyonunu yazınız. Fonksiyon sonucu **döndürsün**.

**Derece = (Fahrenayt − 32) / 1.8**

💡 **Örnek çıktı:**
```text
Fahrenayt değerini giriniz: 212
212 fahrenayt 100.0 derecedir.
```

### 🎯 Kendini Dene 1: Karne Asistanı

İki fonksiyon yaz: `ortalama(n1, n2)` ortalamayı döndürsün, `harf_notu(ort)` ortalamaya göre `AA/BB/CC/DD/FF` döndürsün (Ders 13). İkisini birlikte kullan.

💡 **Örnek çıktı:**
```text
1. not: 80
2. not: 92
Ortalama: 86.0 → AA
```

## ✅ Bugün Öğrendiklerimiz
- `print` ekrana yazar, `return` değeri geri verir.
- Değer döndüren fonksiyonun sonucu değişkene atanabilir ve işlemde kullanılabilir.
- `return a, b` ile birden fazla değer döndürülür.
