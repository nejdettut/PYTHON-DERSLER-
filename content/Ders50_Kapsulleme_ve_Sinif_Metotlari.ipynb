## 1. Kapsülleme (Encapsulation)

Bazı bilgilerin dışarıdan **doğrudan** okunmasını veya değiştirilmesini istemeyiz (şifre, kimlik no, bakiye…). Özelliğin adının başına **iki alt çizgi** `__` koyarak erişimi kısıtlarız.

```text
self.ad = ad            # herkese açık
self.__tc_no = tc_no    # gizli (dışarıdan nesne.__tc_no ile erişilemez)
```

Gizli bilgiye sadece sınıfın **kendi metotları** üzerinden, **kontrollü** erişilir.

### ✏️ Örnek 50.1: Gizli kimlik numarası

`Ogrenci(ad, soyad, tc_no)` sınıfı yaz; `tc_no` gizli olsun (`self.__tc_no`). `bilgi_yaz()` metodu tüm bilgileri yazsın ama TC numarasının sadece **ilk 3 hanesini** göstersin. İki öğrenci oluşturup `bilgi_yaz()` çağır.

💡 **Beklenen çıktı:**
```text
Adı: Nejdet
Soyadı: Tut
TC No: 123******
```

### ✏️ Örnek 50.2: Dışarıdan erişmeyi dene

`ogrenci.ad` ve `ogrenci.__tc_no` değerlerini yazdırmayı dene.

> ⚠️ İkincisi `AttributeError` verecek.

### ✏️ Örnek 50.3: Kontrollü erişim

`Hesap` sınıfında bakiyeyi gizli yap (`self.__bakiye`). Bakiyeyi değiştirmek sadece `para_yatir()` ile mümkün olsun; negatif miktar kabul edilmesin. `bakiye_goster()` metodu bakiyeyi döndürsün.

💡 **Beklenen çıktı:**
```text
Negatif miktar yatırılamaz!
Bakiye: 250
```

## 2. Sınıf Metotları: `@classmethod`

Bazı metotlar tek bir nesneyle değil, **sınıfın tamamıyla** ilgilidir. Bunlar nesne türetmeden doğrudan sınıf üzerinden çağrılır.

```text
class Ogrenci:
    liste = []                      # sınıf özelliği

    @classmethod
    def sayi(cls):                  # self yerine cls (class)
        return len(cls.liste)

Ogrenci.sayi()                      # nesne gerekmez!
```

### ✏️ Örnek 50.4: Öğrenci listesi tutan sınıf

`Ogrenci` sınıfı yaz: sınıf özelliği `liste = []`; `__init__` her yeni öğrencinin adını bu listeye eklesin. `@classmethod` ile `ogrenci_sayisi()` ve `listele()` metotlarını yaz. 4 öğrenci oluşturup ikisini de çağır.

💡 **Beklenen çıktı:**
```text
Sınıf mevcudu: 4
1. Ali
2. Ayşe
3. Can
4. Elif
```

### ✏️ Örnek 50.5: Sınıf metoduyla arama

`Ogrenci` sınıfına `@classmethod ara(cls, ad)` ekle: Öğrenci listede varsa `kayıtlı`, yoksa `kayıtlı değil` yazsın.

💡 **Beklenen çıktı:**
```text
Ayşe kayıtlı.
Mert kayıtlı değil.
```

### ✏️ Örnek 50.6: Ders ve puan ekleme

Her öğrencinin kendi `dersler` sözlüğü olsun (`self.dersler = {}`). `not_ekle(ders, puan)` ve `ortalama()` metotlarını yaz. Bir öğrenciye 3 ders notu ekleyip ortalamasını yazdır.

💡 **Beklenen çıktı:**
```text
Ali - Ortalama: 90.0
```

### 🎯 Kendini Dene 1: Şifreli Kullanıcı

`Kullanici(ad, sifre)` sınıfı yaz; şifre gizli olsun. `giris_yap(sifre)` metodu doğruysa `Giriş başarılı`, yanlışsa `Hatalı şifre` desin. `sifre_degistir(eski, yeni)` sadece eski şifre doğruysa değiştirsin.

## ✅ Bugün Öğrendiklerimiz
- `__ozellik` dışarıdan doğrudan erişimi engeller; erişim metotlar üzerinden kontrollü yapılır.
- `@classmethod` ile tanımlanan metotlar `cls` alır ve `Sinif.metot()` şeklinde çağrılır.
- Sınıf özelliği (`liste = []`) tüm nesnelerde ortak, `self.dersler` her nesnede ayrıdır.
