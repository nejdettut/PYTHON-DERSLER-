## Hazırlık

Fotoğraf uygulamalarındaki filtrelerin çoğu aslında **sayılar üzerinde basit işlemlerdir**. YZ-31'deki manzara resmini oluştur:

```python
# 60x90 piksellik renkli bir manzara resmi üretelim. Sadece çalıştır!
import numpy as np
import matplotlib.pyplot as plt

Y, G = 60, 90
resim = np.zeros((Y, G, 3), dtype=np.uint8)
for s in range(Y):                                    # gökyüzü: yukarıdan aşağı açılan mavi
    resim[s, :] = [80 + s, 150 + s, 255]
resim[40:, :] = [60, 160, 60]                         # çimen
yy, xx = np.ogrid[:Y, :G]
resim[(yy - 14) ** 2 + (xx - 70) ** 2 <= 49] = [255, 220, 0]   # güneş
resim[28:40, 15:35] = [180, 80, 40]                   # ev gövdesi
for i in range(10):                                   # çatı (üçgen)
    resim[18 + i, 25 - i:26 + i] = [200, 30, 30]
resim[33:40, 22:27] = [90, 50, 20]                    # kapı

plt.imshow(resim)
plt.axis("off")
plt.show()
print("Boyut:", resim.shape)
```

## 1. Kırpma, Çevirme, Parlaklık ✂️

| İşlem | numpy ile |
|---|---|
| Kırpma | `resim[10:40, 20:60]` (satırlar, sütunlar) |
| Yatay çevirme (ayna) | `resim[:, ::-1]` |
| Dikey çevirme | `resim[::-1, :]` |
| Parlaklık | `np.clip(resim.astype(int) + 50, 0, 255).astype(np.uint8)` |

> ⚠️ `uint8` 0–255 arası tutar: 250 + 10 = **4** olur (taşma)! Bu yüzden önce `int`'e çevirip `np.clip` ile sınırlarız.
