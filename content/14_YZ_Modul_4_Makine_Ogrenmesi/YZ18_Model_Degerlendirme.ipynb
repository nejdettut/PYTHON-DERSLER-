## Hazırlık

YZ-17'deki modeli tekrar kurmak için şu hazır hücreyi çalıştır. Bu derste test setini biraz büyüttük (`test_size=0.4`) ki farklar görünsün.

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

iris = load_iris()
X, y = iris.data, iris.target
X_egitim, X_test, y_egitim, y_test = train_test_split(X, y, test_size=0.4, random_state=7)
print("Eğitim:", len(X_egitim), "| Test:", len(X_test))
````

## 1. Hata Matrisi (Confusion Matrix)

Doğruluk "kaç tanesini bildim" der ama **hangilerini karıştırdığımı** söylemez. Hata matrisi söyler:

```text
                 Tahmin: setosa  versicolor  virginica
Gerçek: setosa        20            0            0
        versicolor     0           19            2     ← 2 versicolor'ı virginica sandı
        virginica      0            1           18
```
**Köşegen** = doğru tahminler.

### ✏️ Örnek 18.1: Hata matrisi

k=3 ile modeli eğit, test tahminlerini al. `from sklearn.metrics import confusion_matrix` ile hata matrisini yazdır. Hangi türler karıştırılıyor?

### ✏️ Örnek 18.2: Hatalı tahminleri listele

Tahmin ile gerçeğin farklı olduğu test örneklerini bul ve `Gerçek: versicolor → Tahmin: virginica` şeklinde yazdır.


## 2. En İyi k Hangisi?

`n_neighbors` (k) bir **hiperparametredir**: Modelin öğrenmediği, **bizim** seçtiğimiz ayar.

### ✏️ Örnek 18.3: k denemesi

k = 1'den 30'a kadar her değer için modeli eğit ve test doğruluğunu bir listeye kaydet. En iyi k ve doğruluğu yazdır.

💡 **Beklenen çıktı:**
```text

```

### ✏️ Örnek 18.4: k grafiği

## 3. Ezberleyen Model: Aşırı Uyum

| Durum | Eğitim başarısı | Test başarısı | Benzetme |
|---|---|---|---|
| **Aşırı uyum** (overfitting) | Çok yüksek | Düşük | Soruları **ezberleyen** öğrenci: Kitaptaki soruları bilir, yeni soruda takılır |
| **Eksik uyum** (underfitting) | Düşük | Düşük | Konuyu hiç **anlamamış** öğrenci |
| **İyi model** | Yüksek | Yüksek | Konuyu **kavramış** öğrenci |

### ✏️ Örnek 18.5: Eğitim mi, test mi?

k = 1, 15 ve 75 için hem **eğitim** hem **test** doğruluğunu hesapla ve tablo hâlinde yazdır. k=1'de eğitim doğruluğu neden 1.0?

💡 **Beklenen çıktı:**
```text
k=1  → eğitim: 1.00  test: ...
k=15 → eğitim: ...   test: ...
k=75 → eğitim: ...   test: ...
```
