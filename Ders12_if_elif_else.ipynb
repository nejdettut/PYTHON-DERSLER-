## 1. `if – elif – else` Yapısı

`elif` Türkçede **"değilse eğer"** demektir.

```text
if koşul1:
    ...
elif koşul2:
    ...
elif koşul3:
    ...
else:
    hiçbiri doğru değilse
```

- Koşullar **yukarıdan aşağıya** kontrol edilir.
- **İlk doğru** koşulun bloğu çalışır, kalan koşullara **hiç bakılmaz**.
- `else` isteğe bağlıdır ve en sona yazılır.

### ✏️ Örnek 12.1: Not değerlendirme

Kullanıcıdan notu al.

| Not | Mesaj |
|---|---|
| 85 ve üzeri | Pekiyi |
| 70–84 | İyi |
| 50–69 | Geçti |
| 50'den az | Kaldı |

İpucu: `elif notu >= 70` yazman yeterli. Neden `notu < 85` eklemene gerek yok?

💡 **Beklenen çıktı:**
```text
Notunuzu giriniz: 77
İyi
```

### ✏️ Örnek 12.2: Renk seçici

1–5 arası sayı al: 1 → Kırmızı, 2 → Mavi, 3 → Yeşil, 4 → Sarı, 5 → Mor. Başka bir sayıysa `Geçersiz seçim` yazdır.

💡 **Beklenen çıktı:**
```text
1-5 arasında bir sayı giriniz: 3
Yeşil
```

### ✏️ Örnek 12.3: Bay / Bayan / Hatalı giriş

Ders 11'deki karşılama programını `elif` ile geliştir: Erkek → Bay, Kadın → Bayan, başka bir şey → `Hatalı giriş`.

💡 **Beklenen çıktı:**
```text
Adınızı giriniz: Ayşe
Cinsiyetiniz [Kadın/Erkek]: kadın
Bayan Ayşe, hoş geldiniz
```

### ✏️ Örnek 12.4: Boy yorumu (elif ile)

Boyu cm olarak al.

160'tan kısa → kısa · 170'ten kısa → orta · 180'den kısa → uzun · 190'dan kısa → çok uzun · diğer → aşırı uzun

💡 **Beklenen çıktı:**
```text
Boyunuz kaç cm?: 150
Boyunuz kısa
```

## 2. `if` mi, `elif` mi?

| Art arda `if` | `if – elif` |
|---|---|
| **Tüm** koşullar tek tek kontrol edilir | İlk doğru koşulda **durur** |
| Birden fazla blok çalışabilir | **Sadece bir** blok çalışır |

### ✏️ Örnek 12.5: Farkı görelim

Örnek 12.4'ü bu sefer `elif` yerine **sadece `if`** kullanarak yaz. 150 girince kaç satır çıktı aldın? Neden?

💡 **Beklenen çıktı:**
```text
Boyunuz kaç cm?: 150
Boyunuz kısa
Boyunuz orta
Boyunuz uzun
Boyunuz çok uzun
```

## 3. Problemler

### ✏️ Örnek 12.6: Otopark ücreti

Otoparkta kalınan saati al.

| Süre | Ücret |
|---|---|
| 2 saate kadar | 200 TL |
| 3–6 saat | 300 TL |
| 7–12 saat | 400 TL |
| 13–24 saat | 500 TL |
| 24 saatten fazla | 750 TL |

💡 **Beklenen çıktı:**
```text
Otoparkta kalınan saat: 8
Ödemeniz gereken ücret: 400 TL
```

### ✏️ Örnek 12.7: Hangisi büyük?

İki sayı al. Birinci büyükse, ikinci büyükse veya eşitse uygun mesajı yazdır.

💡 **Beklenen çıktı:**
```text
Birinci sayı: 12
İkinci sayı: 12
Bu sayılar eşittir
```

## 📝 Ödev 4 – if elif else

Tam bilet ücreti **100 TL** olan tiyatro için, klavyeden girilen yaşa göre indirimli bilet ücretini hesaplayıp yazdırınız.

| Yaş | İndirim |
|---|---|
| 0–6 | Ücretsiz |
| 7–12 | %50 |
| 13–26 | %20 |
| 27–59 | İndirim yok |
| 60–64 | %20 |
| 65 ve üzeri | %50 |

💡 **Örnek çıktı:**
```text
Yaşınızı giriniz: 15
Normal bilet fiyatı: 100 TL
Bilet ücretiniz: 80.0 TL
```

> 📌 **Hangisini ne zaman kullanırım?**
> Tek koşul → `if` · İki durum → `if – else` · İkiden fazla durum → `if – elif – else`

## ✅ Bugün Öğrendiklerimiz
- `elif` = "değilse eğer". İlk doğru koşul çalışır, diğerleri atlanır.
- Art arda `if`'lerde tüm koşullar kontrol edilir; birden fazla blok çalışabilir.
- Aralık kontrolünde koşulları **büyükten küçüğe** (veya tersi) sıralamak kodu kısaltır.
