## 1. Neden Dosya?

Şimdiye kadar değişkenlerde tuttuğumuz her şey program kapanınca **kayboldu**. Dosyalar verileri **kalıcı** olarak saklar.

**Üç adım:** ① Dosyayı aç → ② Oku / yaz → ③ Kapat

```text
dosya = open("deneme.txt", "w", encoding="utf-8")
dosya.write("Merhaba")
dosya.close()
```

> 📁 Colab ve JupyterLite'ta dosyalar defterle aynı klasöre kaydedilir. Sol taraftaki **dosya simgesinden** görebilirsin.
> ⚠️ JupyterLite'ta dosyalar **tarayıcıda** saklanır; tarayıcı verisi silinirse kaybolur.

## 2. Dosya Açma Kipleri

| Kip | Anlamı | Dosya yoksa | Dosya varsa |
|---|---|---|---|
| `"r"` | Okuma (read) | ❌ Hata | Okur |
| `"w"` | Yazma (write) | Oluşturur | ⚠️ **İçini siler**, baştan yazar |
| `"a"` | Ekleme (append) | Oluşturur | **Sonuna** ekler |
| `"r+"` | Okuma + yazma | ❌ Hata | Okur ve yazar |

> 🔹 Türkçe karakterler bozulmasın diye her zaman `encoding="utf-8"` yaz.

### ✏️ Örnek 43.1: İlk dosyamız

`deneme.txt` dosyasını `"w"` kipinde aç, içine `Bu basit bir dosyadır.` yaz ve kapat. Sonra soldaki dosya panelinden dosyayı bul ve aç.

### ✏️ Örnek 43.2: Dosyayı oku

### ✏️ Örnek 43.3: w kipi her şeyi siler!

💡 **Beklenen çıktı:**

### ✏️ Örnek 43.4: a kipi ile ekle

## 3. `with` ile Otomatik Kapatma

`close()` yazmayı unutmak sık yapılan bir hatadır. `with` bloğu bitince dosya **kendiliğinden kapanır**:

```text
with open("deneme.txt", "r", encoding="utf-8") as dosya:
    icerik = dosya.read()
print(icerik)        # blok bitti, dosya kapandı
```

## ✅ Bugün Öğrendiklerimiz
- Dosya işlemleri: aç → oku/yaz → kapat.
- `"w"` dosyayı sıfırlar, `"a"` sonuna ekler, `"r"` sadece okur.
- `with open(...) as dosya:` dosyayı otomatik kapatır; `encoding="utf-8"` Türkçe için şarttır.
