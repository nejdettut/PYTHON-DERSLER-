## 1. Parametre mi, Argüman mı?

```text
def selamla(ad):         # ad  → PARAMETRE (tanımlarken yazılan)
    print("Merhaba", ad)

selamla("Ayşe")          # "Ayşe" → ARGÜMAN (çağırırken gönderilen)
```

- Fonksiyon birden çok parametre alabilir: `def kayit(isim, soyisim, yas):`
- Argümanlar **aynı sırayla** gönderilir.

### ✏️ Örnek 22.1: Selamlama

`selamla(ad)` fonksiyonunu tanımla ve 3 farklı isimle çağır.

💡 **Beklenen çıktı:**
```text
Merhaba, NEJDET
Merhaba, ÖZGE
Merhaba, DÜNYA
```

### ✏️ Örnek 22.2: Fonksiyonsuz kayıt

Bir kişinin isim, soyisim, doğum yeri ve yaşını 4 değişken ve 4 `print()` ile yazdır. 3 kişi olsa kaç satır yazman gerekirdi?

### ✏️ Örnek 22.3: Fonksiyonlu kayıt

`kayit(isim, soyisim, dogum, yas)` fonksiyonunu yaz ve 3 farklı kişi için çağır.

💡 **Beklenen çıktı:**
```text
İsim: Arda
Soyisim: Güler
Doğum Yeri: Ankara
Yaşı: 18

İsim: Eda
...
```

### ✏️ Örnek 22.4: Eksik argüman

`kayit("Ali", "Kaya")` şeklinde sadece 2 argümanla çağır.

> ⚠️ `TypeError: missing 2 required positional arguments` hatası alacaksın.

## 2. Varsayılan (Default) Değer

Parametreye tanımlarken bir değer verirsek, argüman gönderilmediğinde **o değer** kullanılır:

```text
def selamla(ad="Misafir"):
    print("Merhaba,", ad)

selamla()          # Merhaba, Misafir
selamla("Ayşe")    # Merhaba, Ayşe
```

### ✏️ Örnek 22.5: Varsayılan değerli selam

Yukarıdaki fonksiyonu yaz ve hem argümanlı hem argümansız çağır.

💡 **Beklenen çıktı:**
```text
Merhaba, Misafir
Merhaba, Ayşe
Merhaba, Misafir
```

### ✏️ Örnek 22.6: Metni n kez yaz

`yaz(metin, kac_kere)` fonksiyonu metni istenen sayıda yazsın. Değerleri kullanıcıdan al.

💡 **Beklenen çıktı:**
```text
Metin giriniz: Python
Kaç kere: 3
Python
Python
Python
```

### ✏️ Örnek 22.7: Sınırı kontrol et

Örnek 22.6'yı geliştir: Kullanıcı 1–100 dışında bir sayı girerse `1-100 arası bir sayı giriniz` deyip **tekrar sorsun** (`while True`).

💡 **Beklenen çıktı:**
```text
Kaç kere: 500
1-100 arası bir sayı giriniz.
Kaç kere: 2
...
```

### ✏️ Örnek 22.8: Yılbaşı ağacı 🎄

`agac_ciz(yukseklik, karakter="*")` fonksiyonu yazdır. Her satırda `(yukseklik - satir)` kadar boşluk ve `(2*satir - 1)` kadar karakter olsun. Bir kez sadece yükseklikle, bir kez farklı karakterle çağır.

💡 **Beklenen çıktı:**
```text
    *
   ***
  *****
 *******
*********
```

### 🎯 Kendini Dene 1: Döngülü Kayıt

Örnek 22.3'teki `kayit()` fonksiyonunu kullan. Program `Veri girmek istiyor musunuz? [E/H]` diye sorsun; `E` ise bilgileri alıp `kayit()`'ı çağırsın, `H` ise `Kayıt işlemi bitti` deyip çıksın.

## ✅ Bugün Öğrendiklerimiz
- Tanımlarken yazılan **parametre**, çağırırken gönderilen **argüman**dır.
- Argümanlar parametrelerle **aynı sırada** ve **aynı sayıda** gönderilir.
- `def f(x=5):` varsayılan değer verir; argüman gönderilmezse 5 kullanılır.
