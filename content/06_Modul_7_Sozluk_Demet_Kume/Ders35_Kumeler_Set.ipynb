## 1. Küme (Set) Nedir?

Matematikteki kümeler gibi:
- **Tekrar yoktur**: Aynı eleman bir kez bulunur.
- **Sıra yoktur**: İndeksle erişilemez (`kume[0]` ❌).

```text
meyveler = {"elma", "armut", "çilek"}
sayilar = set([1, 2, 2, 3])        →  {1, 2, 3}
```

> ⚠️ `{}` boş **sözlük** oluşturur! Boş küme için `set()` kullanılır.

### ✏️ Örnek 35.1: Küme oluşturalım

`meyveler = {"elma", "armut", "çilek"}` ve `sayilar = set([1, 2, 3, 4, 5])` oluştur. İkisini ve `meyveler`'in tipini yazdır. Elemanların sırası yazdığın gibi mi?

### ✏️ Örnek 35.2: Tekrarlar gider

`deneme = {1, 2, 3, 4, 4, 5, 5, 6}` tanımla. Kümeyi ve uzunluğunu yazdır.

💡 **Beklenen çıktı:**
```text
{1, 2, 3, 4, 5, 6}
6
```

### ✏️ Örnek 35.3: Boş küme tuzağı

`a = {}` ve `b = set()` tiplerini yazdır.

💡 **Beklenen çıktı:**
```text
<class 'dict'>
<class 'set'>
```

### ✏️ Örnek 35.4: Listedeki tekrarları sil

`yoklama = ["Ali", "Ayşe", "Ali", "Can", "Ayşe", "Ali"]` listesindeki tekrarları `set()` ile temizle, sonra `sorted()` ile sıralı listeye çevir.

💡 **Beklenen çıktı:**
```text
['Ali', 'Ayşe', 'Can']
```

## 2. Küme İşlemleri

| İşlem | Metot | Kısa yazım | Anlamı |
|---|---|---|---|
| Ekleme | `k.add(x)` | | |
| Çıkarma | `k.remove(x)` | | |
| Birleşim | `a.union(b)` | `a \| b` | İkisinde olanların hepsi |
| Kesişim | `a.intersection(b)` | `a & b` | **Ortak** olanlar |
| Fark | `a.difference(b)` | `a - b` | `a`'da olup `b`'de olmayanlar |

> Aşağıdaki örneklerde `kume1 = {1, 2, 3, 4, 5}` ve `kume2 = {3, 5, 6, 7, 8}` kullan. Önce tanımla.

### ✏️ Örnek 35.5: Ekle ve çıkar

`kume1`'e 6 ekle, sonra 3'ü çıkar. Her adımda yazdır.

💡 **Beklenen çıktı:**
```text
{1, 2, 3, 4, 5, 6}
{1, 2, 4, 5, 6}
```

### ✏️ Örnek 35.6: Birleşim ve kesişim

İki kümenin birleşimini ve kesişimini bul.

💡 **Beklenen çıktı:**
```text
Birleşim: {1, 2, 3, 4, 5, 6, 7, 8}
Kesişim: {5, 6}
```

### ✏️ Örnek 35.7: Fark

`kume1 - kume2` ve `kume2 - kume1` farklarını bul. Neden aynı değil?

💡 **Beklenen çıktı:**
```text
{1, 2, 4}
{3, 7, 8}
```
