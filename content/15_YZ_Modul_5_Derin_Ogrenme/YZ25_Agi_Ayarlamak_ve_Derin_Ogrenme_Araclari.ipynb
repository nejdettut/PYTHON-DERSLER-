## 1. Hiperparametreler 🎛️

Sinir ağının **öğrendiği** şeyler ağırlıklardır. **Bizim seçtiğimiz** ayarlara ise **hiperparametre** denir:

| Hiperparametre | MLPClassifier'da | Etkisi |
|---|---|---|
| Katman ve nöron sayısı | `hidden_layer_sizes` | Ağın kapasitesi |
| Öğrenme oranı | `learning_rate_init` | Adım büyüklüğü (YZ-15!) |
| Tur sayısı | `max_iter` | Ne kadar uzun eğitileceği |
| Aktivasyon | `activation` | `"relu"`, `"tanh"`, `"logistic"` (sigmoid) |

Önce rakam verisini hazırla:

````python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
import matplotlib.pyplot as plt
import time, warnings
warnings.filterwarnings("ignore")      # "yakınsamadı" uyarılarını gizle

rakamlar = load_digits()
X = rakamlar.data / 16
X_egitim, X_test, y_egitim, y_test = train_test_split(X, rakamlar.target, test_size=0.25, random_state=0)
print("Hazır:", X_egitim.shape, X_test.shape)
````

### ✏️ Örnek 25.1: Öğrenme oranı deneyi

`learning_rate_init` değerini 0.0001, 0.001, 0.01, 0.1 ve 1.0 yaparak (`hidden_layer_sizes=(64,)`, `max_iter=200`, `random_state=0`) beş ağ eğit. Test doğruluklarını yazdır ve kayıp eğrilerini aynı grafikte çiz.

### ✏️ Örnek 25.2: Tur sayısı deneyi

`max_iter` değerini 5, 20, 50, 200 yaparak dört ağ eğit. Her biri için **eğitim** ve **test** doğruluğunu yazdır.

### ✏️ Örnek 25.3: Aktivasyon deneyi

`activation` için `"relu"`, `"tanh"` ve `"logistic"` değerlerini karşılaştır.

### ✏️ Örnek 25.4: Erken durdurma

`early_stopping=True` ekleyerek eğit. Model eğitim verisinin bir kısmını **doğrulama** için ayırır ve doğrulama başarısı artmayı bırakınca durur. Kaç turda durdu (`model.n_iter_`)? Test doğruluğu nasıl?

### ✍️ Deney Raporu

25.1–25.4'teki sonuçlara göre rakam tanıma için **en iyi ayarları** seç ve 3–4 cümleyle nedenini açıkla.

## 2. Gerçek Derin Öğrenme Araçları 🛠️

`MLPClassifier` öğrenmek için harika ama büyük projelerde şu kütüphaneler kullanılır:

| Araç | Geliştiren | Öne çıkan |
|---|---|---|
| **TensorFlow / Keras** | Google | Kolay katman yazımı, mobil ve web'e aktarım |
| **PyTorch** | Meta | Araştırmalarda en popüler, esnek |
| **Hugging Face** | Topluluk | Hazır dil ve görüntü modelleri |

Bu kütüphaneler büyük olduğu için **tarayıcıda (JupyterLite) çalışmaz**; **Google Colab**'de hazır gelir. Aynı rakam tanıma ağının Keras ile yazılışı:

```text
# ⚠️ Sadece Colab'de çalışır
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation="relu", input_shape=(64,)),   # gizli katman
    tf.keras.layers.Dense(10, activation="softmax"),                    # 10 rakam için çıkış
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.fit(X_egitim, y_egitim, epochs=20, validation_split=0.1)
```

> 🔹 Gördüğün gibi kavramlar aynı: katmanlar, nöron sayısı, aktivasyon, tur (epoch), doğrulama verisi.
