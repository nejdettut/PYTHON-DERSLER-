## 1. Değişken Nedir?

Değişken, veriyi saklayan **etiketli bir kutu** gibidir. Her değişkenin bir **adı**, bir **değeri** ve bir **veri tipi** vardır.

```text
sehir = "Ankara"
```
- `sehir` → değişkenin adı
- `"Ankara"` → değişkenin değeri
- `=` → **atama operatörü**

> 🔹 Python'da `=` matematikteki "eşittir" değildir. **Sağdaki değeri soldaki değişkene koyar.**
> Bu yüzden `5 = A` yanlış, `A = 5` doğrudur.

### ✏️ Örnek 3.1: Değişken tanımlayıp ekrana yazalım

`sehir` adında bir değişken oluştur ve yaşadığın şehri ata. Sonra yazdır.

💡 **Beklenen çıktı:**
```text
Yaşadığınız şehir: Ankara
```

### ✏️ Örnek 3.2: Atama yönü hatası

Önce `X = 5` yazıp yazdır. Sonra `5 = X` yazıp çalıştır.

> ⚠️ İkinci kod hata verecek. Sayı, değişken adı olamaz.

### ✏️ Örnek 3.3: Değişkenlerle işlem yapalım

`A = 5` ve `B = 6` tanımla. Toplamlarını `C` değişkenine ata ve yazdır.

💡 **Beklenen çıktı:**
```text
Toplam: 11
```

## 2. Değişken İsimlendirme Kuralları

| Kural | ✅ Doğru | ❌ Yanlış |
|---|---|---|
| Harf veya `_` ile başlar | `ad`, `_sayi`, `not1` | `1not` |
| Boşluk içermez | `ogrenci_adi` | `ogrenci adi` |
| Özel karakter içermez | `toplam` | `toplam!`, `%oran` |
| Python'un anahtar kelimesi olamaz | `sayac` | `if`, `for`, `while` |
| Büyük-küçük harf duyarlıdır | `Ali` ile `ali` **farklı** değişkenlerdir | |

> 🔹 Hiç değişmeyecek değerler geleneksel olarak **BÜYÜK HARFLE** yazılır: `PI = 3.14`

### ✏️ Örnek 3.4: Python'un anahtar kelimelerini listeleyelim

`import keyword` satırının altına `print(keyword.kwlist)` yaz. Bu kelimeleri değişken adı olarak kullanamazsın.

### ✏️ Örnek 3.5: Büyük-küçük harf farkı

`ali = 10` ve `Ali = 20` tanımla. İkisini de yazdır.

💡 **Beklenen çıktı:**
```text
10 20
```

## 3. Değişkenin Değeri Değişebilir

Bir değişkene yeni değer atanınca **eski değer silinir**, son atanan değer geçerli olur.

### ✏️ Örnek 3.6: Değişen değeri takip edelim

`A` değişkenine sırayla `"Bilişim"`, `36` ve `"Türkiye Cumhuriyeti"` değerlerini ata. Her atamadan sonra yazdır.

💡 **Beklenen çıktı:**
```text
A'nın değeri: Bilişim
A'nın değeri: 36
A'nın değeri: Türkiye Cumhuriyeti
```

### ✏️ Örnek 3.7: Son atanan değer kazanır

`AD` değişkenine sırayla `"Mustafa"`, `"Kemal"`, `"Atatürk"` ata. **Sadece en sonda** bir kez yazdır. Ne görmeyi bekliyorsun?

💡 **Beklenen çıktı:**
```text
Ad: Atatürk
```

### ✏️ Örnek 3.8: Üç değişkenle tam ad

Bu sefer `AD1`, `AD2`, `AD3` olarak üç ayrı değişken kullan ve tam adı yazdır.

💡 **Beklenen çıktı:**
```text
Ad: Mustafa Kemal Atatürk
```

## 4. Çoklu Atama ve Yer Değiştirme

- Aynı değeri birden fazla değişkene atamak: `A = B = C = 10.5`
- İki değişkenin değerini tek satırda değiştirmek: `A, B = B, A`

### ✏️ Örnek 3.9: Çoklu atama yapalım

`A`, `B`, `C` değişkenlerine tek satırda `10.5` ata ve üçünü yazdır.

💡 **Beklenen çıktı:**
```text
10.5 10.5 10.5
```

### ✏️ Örnek 3.10: Yer değiştirme: yanlış yol

`A = 3`, `B = 5` tanımla. Sonra `A = B` ve `B = A` yaz. Yeni değerleri yazdır. Neden ikisi de 5 oldu?

💡 **Beklenen çıktı:**
```text
Yeni A: 5
Yeni B: 5
```

### ✏️ Örnek 3.11: Yer değiştirme: Python yolu

Aynı değerlerle bu sefer `A, B = B, A` kullan.

💡 **Beklenen çıktı:**
```text
Yeni A: 5
Yeni B: 3
```

### ✏️ Örnek 3.12: `del` ile değişken silelim

`X = Y = Z = "ANITKABİR"` tanımla, üçünü yazdır. Sonra `del X` yaz ve `X`'i tekrar yazdırmayı dene.

> ⚠️ Son satır `NameError` verecek, çünkü X artık yok.

### 🎯 Kendini Dene 1: Öğrenci Bilgi Kartı

`ad`, `soyad`, `sinif`, `okul_no` değişkenlerini oluştur ve hepsini tek `print()` ile yazdır. Sonra `sinif` değişkenini değiştirip tekrar yazdır.

### 🎯 Kendini Dene 2: Bardak Problemi

`bardak1 = "çay"` ve `bardak2 = "süt"` olsun. Üçüncü bir değişken (`bos_bardak`) kullanarak içerikleri yer değiştir. `A, B = B, A` kullanmak yok!

💡 **Örnek çıktı:**
```text
bardak1: süt
bardak2: çay
```

## ✅ Bugün Öğrendiklerimiz
- Değişken veriyi saklar; `=` sağdaki değeri soldaki değişkene atar.
- İsimler harf veya `_` ile başlar, boşluk ve özel karakter içermez.
- Son atanan değer geçerlidir; `A, B = B, A` ile yer değiştirilir.
- `del` değişkeni siler.
