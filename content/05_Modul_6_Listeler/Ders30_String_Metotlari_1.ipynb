## 1. Harf Dönüşüm Metotları

| Metot | `"Python PROGRAMLAMA dili"` için sonuç |
|---|---|
| `lower()` | `python programlama dili` |
| `upper()` | `PYTHON PROGRAMLAMA DILI` ⚠️ |
| `title()` | `Python Programlama Dili` |
| `capitalize()` | `Python programlama dili` |
| `swapcase()` | `pYTHON programlama DILI` ⚠️ |

> 🔹 Metotlar **orijinal metni değiştirmez**, yeni bir metin döndürür. Saklamak için: `metin = metin.upper()`
> ⚠️ **Türkçe uyarısı:** Python `"i".upper()` için `I` verir, `İ` değil. Türkçe karakterlerde sonuçları kontrol et!

### ✏️ Örnek 30.1: Beş metodu deneyelim

`metin = "Python PROGRAMLAMA dili"` için beş metodu da uygulayıp sonuçları yazdır.

### ✏️ Örnek 30.2: Metin değişmedi mi?

`metin.upper()` yazdıktan sonra `print(metin)` yap. Sonra `metin = metin.upper()` ile tekrar dene.

### ✏️ Örnek 30.3: Türkçe harf tuzağı

`"istanbul".upper()` ve `"IŞIK".lower()` sonuçlarını yazdır. Doğru mu?

💡 **Beklenen çıktı:**
```text
ISTANBUL
işik
```

### ✏️ Örnek 30.4: Kullanıcı girişini düzenle

Kullanıcıdan `evet`/`hayır` cevabı al. Kullanıcı `EVET`, `Evet`, `evet` yazsa da `.lower()` ile hepsini kabul et.

💡 **Beklenen çıktı:**
```text
Devam edilsin mi? EvEt
Devam ediliyor...
```

## 2. `replace(eski, yeni, adet)`

Metindeki karakterleri/kelimeleri değiştirir. `adet` verilirse sadece o kadarını değiştirir.

### ✏️ Örnek 30.5: Ankara → Angara

`ad = "ankara"` için sırasıyla: `k`→`g`, tüm `a`→`A`, sadece ilk `a`→`A`, tüm `a`'ları sil.

💡 **Beklenen çıktı:**
```text
angara
AnkArA
Ankara
nkr
```

### ✏️ Örnek 30.6: Kötü kelime filtresi

Kullanıcıdan bir cümle al. İçindeki `"aptal"` kelimesini `"*****"` ile değiştir.

💡 **Beklenen çıktı:**
```text
Cümle: sen aptal mısın
sen ***** mısın
```

## 3. `split()` ve `join()`

| Metot | Yön | Örnek |
|---|---|---|
| `metin.split()` | metin → **liste** | `"a b c".split()` → `['a', 'b', 'c']` |
| `metin.split("-")` | ayraçla böl | `"a-b-c".split("-")` → `['a', 'b', 'c']` |
| `"ayraç".join(liste)` | liste → **metin** | `", ".join(['a', 'b'])` → `"a, b"` |

### ✏️ Örnek 30.7: Cümleyi kelimelere ayır

`cumle = "Python dünyanın bir numaralı yazılım dilidir."` cümlesini kelimelere ayır, listeyi ve kelime sayısını yazdır.

💡 **Beklenen çıktı:**
```text
['Python', 'dünyanın', 'bir', 'numaralı', 'yazılım', 'dilidir.']
Cümlenizde 6 adet kelime vardır
```

### ✏️ Örnek 30.8: Virgülle girilen bilgiler

Kullanıcıdan virgülle ayrılmış bilgiler (ad,soyad,şehir) al ve listeye çevir.

💡 **Beklenen çıktı:**
```text
Bilgileri virgülle giriniz: Ayşe,Yılmaz,Ankara
['Ayşe', 'Yılmaz', 'Ankara']
```

### ✏️ Örnek 30.9: join ile birleştir

`liste = ["Python", "programlama", "çok", "eğlenceli"]` listesini boşluk, `;` ve `" ve "` ile birleştir.

💡 **Beklenen çıktı:**
```text
Python programlama çok eğlenceli
Python;programlama;çok;eğlenceli
Python ve programlama ve çok ve eğlenceli
```

### ✏️ Örnek 30.10: Yarısı büyük, yarısı başlık

`metin = "Python dünyanın en popüler ve kolay yazılım dilidir."` cümlesinin ilk yarı kelimelerini `upper()`, kalanlarını `title()` yap ve `join` ile tekrar cümleye çevir.

💡 **Beklenen çıktı:**
```text
PYTHON DÜNYANIN EN POPÜLER Ve Kolay Yazılım Dilidir.
```

### 🎯 Kendini Dene 1: Kullanıcı Adı Üretici

Kullanıcıdan ad ve soyad al. Boşlukları silip küçük harfe çevirerek `ad.soyad` şeklinde kullanıcı adı üret.

💡 **Örnek çıktı:**
```text
Ad Soyad: Ayşe Nur Yılmaz
Kullanıcı adınız: ayşenur.yılmaz
```
