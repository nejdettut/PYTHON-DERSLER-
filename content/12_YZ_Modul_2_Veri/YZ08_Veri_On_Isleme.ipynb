## 1. Veri Ön İşleme Nedir?

Ham veri doğrudan modele verilemez. Veri bilimcilerin zamanının **büyük kısmı** veriyi temizlemekle geçer!

| Adım | Sorun | Çözüm |
|---|---|---|
| **Eksik veri** | Boş hücreler | Satırı sil **veya** ortalama/ortanca ile doldur |
| **Hatalı veri** | Yaş = −5 | Düzelt veya eksik say |
| **Tekrarlı veri** | Aynı satır iki kez | Tekrarı sil |
| **Kodlama** | Model metin anlamaz ("K", "E") | Sayıya çevir (1, 0) |
| **Normalizasyon** | Sütunların ölçekleri farklı (saat: 0–9, puan: 0–100) | Hepsini 0–1 aralığına getir |
