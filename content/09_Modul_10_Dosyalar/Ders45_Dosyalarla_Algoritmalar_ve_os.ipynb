## 1. Hesapla → Kaydet → Kullan

Bazı hesaplamalar uzun sürer. Sonucu bir kez hesaplayıp **dosyaya kaydedersek**, sonra hesaplamadan **dosyadan okuyarak** kullanabiliriz.

> ⚠️ Dosyaya her şey **metin** olarak yazılır. Sayı yazarken `str()`, okurken `int()` gerekir.

### ✏️ Örnek 45.1: Asal sayıları dosyaya yaz

1–1000 arası asal sayıları bul (Ödev 7'yi hatırla) ve aralarında boşluk olacak şekilde `asal.txt`'ye kaydet. Kaç asal sayı bulduğunu yazdır.

💡 **Beklenen çıktı:**
```text
168 asal sayı asal.txt dosyasına kaydedildi.
```

### ✏️ Örnek 45.2: Dosyadan asal kontrolü

`asal.txt`'yi oku, `split()` ile listeye çevir. Kullanıcının girdiği sayının listede olup olmadığına bakarak asal mı değil mi yazdır.

💡 **Beklenen çıktı:**
```text
Sayı: 997
997 asal sayıdır.
```

### ✏️ Örnek 45.3: Döngülü sorgu

Örnek 45.2'yi kullanıcı `0` girene kadar soracak şekilde geliştir.

💡 **Beklenen çıktı:**
```text
Sayı (çıkış 0): 15
15 asal değildir.
Sayı (çıkış 0): 0
Çıkış yapıldı...
```

## 2. `os` Modülü

İşletim sistemiyle (dosyalar, klasörler) konuşmamızı sağlar.

| Fonksiyon | Ne yapar? |
|---|---|
| `os.path.exists("a.txt")` | Dosya var mı? (`True`/`False`) |
| `os.remove("a.txt")` | Dosyayı **siler** |
| `os.listdir()` | Klasördeki dosyaları listeler |

### ✏️ Örnek 45.4: Dosya var mı?

`["asal.txt", "rapor.csv", "notlar.docx"]` listesindeki her dosya için var mı yok mu yazdır.

💡 **Beklenen çıktı:**
```text
asal.txt mevcut.
rapor.csv bulunamadı.
notlar.docx bulunamadı.
```

### ✏️ Örnek 45.5: Klasörü listele ve sil

`os.listdir()` ile klasördeki dosyaları gör. Sonra `silinecek.txt` adında bir dosya oluştur, var olduğunu kontrol et, `os.remove()` ile sil ve tekrar kontrol et.

💡 **Beklenen çıktı:**
```text
Silmeden önce: True
Sildikten sonra: False
```

## 📝 Ödev 17 – Dosya İşlemleri

1. **100–999** arası **50 rastgele sayı** üretip her birini ayrı satırda `sayi.txt`'ye kaydediniz.
2. `sayi.txt`'yi okuyup sayıları **büyükten küçüğe** sıralayınız ve `sirali.txt`'ye kaydediniz.
3. `sirali.txt`'den en büyük ve en küçük sayıyı okuyup yazdırınız.

💡 Sıralamadan önce satırları `int()`'e çevirmeyi unutma! Metin olarak sıralanırsa ne olur?

💡 **Örnek çıktı:**
```text
sayi.txt oluşturuldu.
sirali.txt oluşturuldu.
En büyük: 987
En küçük: 104
```

## ✅ Bugün Öğrendiklerimiz
- Dosyaya sayılar metin olarak yazılır; okurken `int()` ile geri çevrilir.
- Okunan metni `split()` ile listeye çevirip `in` ile arama yapılabilir.
- `os.path.exists()` dosyanın varlığını kontrol eder, `os.remove()` siler.
