## 1. Neden Sıralama?

- İkili arama sıralı veri ister (YZ-12).
- Sıralı veride en büyük/en küçük, ortanca kolayca bulunur.
- Arama motorları ve öneri sistemleri sonuçları **sıralayarak** gösterir.

Python'da `sort()` ve `sorted()` hazır. Ama içeride nasıl çalıştıklarını anlamak için kendi sıralama algoritmalarımızı yazacağız.

## 2. Kabarcık Sıralama (Bubble Sort) 🫧

Yan yana duran iki elemanı karşılaştır; **soldaki büyükse yer değiştir**. Her turda en büyük eleman bir kabarcık gibi **en sona** yükselir.

```text
[5, 3, 8, 2]
 5>3 değiştir → [3, 5, 8, 2]
 5<8 kalsın   → [3, 5, 8, 2]
 8>2 değiştir → [3, 5, 2, 8]   ← 1. tur bitti, 8 yerinde
...
```

### ✏️ Örnek 13.1: Bir tur

`liste = [5, 3, 8, 2]` için **tek bir tur** yap: `for j in range(len(liste) - 1)` ile yan yana elemanları karşılaştır, gerekirse `A, B = B, A` ile yer değiştir (Ders 3!). Her karşılaştırmadan sonra listeyi yazdır.

💡 **Beklenen çıktı:**
```text
[3, 5, 8, 2]
[3, 5, 8, 2]
[3, 5, 2, 8]
```

### ✏️ Örnek 13.2: Kabarcık sıralama fonksiyonu

Tek turu bir dış döngüye koyarak `kabarcik_sirala(liste)` fonksiyonunu yaz. Her tur sonunda listeyi yazdır. Orijinal listeyi bozmamak için fonksiyonun başında `liste = liste.copy()` yap (Ders 26).

💡 **Beklenen çıktı:**
```text
1. tur: [3, 5, 2, 8]
2. tur: [3, 2, 5, 8]
3. tur: [2, 3, 5, 8]
4. tur: [2, 3, 5, 8]
```

### ✏️ Örnek 13.3: Gereksiz turları atla

13.2'de 4. turda hiçbir değişiklik olmadı. Bir turda hiç yer değiştirme olmazsa liste sıralıdır! `degisti = False` bayrağıyla bu durumda `break` yap. Neredeyse sıralı olan `[1, 2, 4, 3, 5]` listesini önce 13.2'deki fonksiyonla, sonra yeni fonksiyonla sırala. Kaç tur kazandın?

💡 **Beklenen çıktı:**
```text
Bayraksız: 5 tur
Bayraklı: 2 tur (1. tur: [1, 2, 3, 4, 5], 2. turda değişiklik yok → dur)
```

## 3. Seçmeli Sıralama (Selection Sort)

Listenin sıralanmamış kısmındaki **en küçük** elemanı bul ve **başa** koy. Sonra kalan kısım için tekrarla.

```text
[64, 25, 12, 22]  → en küçük 12, başa koy → [12, 25, 64, 22]
[12 | 25, 64, 22] → en küçük 22           → [12, 22, 64, 25]
[12, 22 | 64, 25] → en küçük 25           → [12, 22, 25, 64]
```

### ✏️ Örnek 13.4: Seçmeli sıralama

`secmeli_sirala(liste)` fonksiyonunu yaz ve `[64, 25, 12, 22, 11]` listesini sırala.

💡 **Beklenen çıktı:**
```text
[11, 12, 22, 25, 64]
```

## 4. Hız Yarışı 🏁

### ✏️ Örnek 13.5: Kimin kaç adımı var?

1.000, 2.000 ve 4.000 elemanlı rastgele listeleri kabarcık sıralamayla sırala ve süreleri ölç. Liste **2 katına** çıkınca süre kaç kat arttı?

> ⚠️ 4.000 eleman birkaç saniye sürebilir, sabırlı ol!

### ✏️ Örnek 13.6: Python'un sorted()'ı

Aynı listeleri `sorted()` ile sırala ve süreleri karşılaştır. Python'un algoritması (Timsort) O(n log n) çalışır.

### 🎯 Kendini Dene 1: Notları Sırala

`notlar = [("Ali", 72), ("Ayşe", 95), ("Can", 64), ("Elif", 88)]` listesini kendi yazdığın kabarcık sıralamayı değiştirerek **puana göre büyükten küçüğe** sırala. İpucu: Karşılaştırmayı `liste[j][1] < liste[j+1][1]` ile yap.

💡 **Örnek çıktı:**
```text
[('Ayşe', 95), ('Elif', 88), ('Ali', 72), ('Can', 64)]
```

## ✅ Bugün Öğrendiklerimiz
- Kabarcık sıralama yan yana elemanları karşılaştırıp yer değiştirir: O(n²).
- Seçmeli sıralama en küçüğü bulup başa taşır: O(n²).
- Python'un `sorted()` fonksiyonu O(n log n) çalışır ve çok daha hızlıdır.
