## 1. Neden İhtiyacımız Var?

Şu ana kadar değişkenleri virgülle yazdırdık:

```text
print("Girdiğiniz", a, b, c, "değerlerinin toplamı =", a+b+c)
```

Uzun cümlelerde bu yöntem karışık hâle gelir. Metnin içine değişkenleri **yerleştirmenin** iki yolu var: `.format()` ve **f-string**.

## 2. `.format()` Metodu

Metnin içine `{}` yerleştirilir, değerler `.format()` içine **sırayla** yazılır.

| Yazım | Sonuç |
|---|---|
| `"{} {}".format("Ali", 15)` | `Ali 15` |
| `"{1} {0}".format("Ali", 15)` | `15 Ali` (sıra numarasıyla) |
| `"{a} {b}".format(a="elma", b="armut")` | `elma armut` |
| `"{:.2f}".format(3.14159)` | `3.14` (2 basamak) |

### ✏️ Örnek 9.1: Sırayla yerleştirme

`"Sırasıyla çıktılar: {} {} {} {}"` metnine `1, 2, "PYTHON", "Bilişim"` değerlerini yerleştir.

💡 **Beklenen çıktı:**
```text
Sırasıyla çıktılar: 1 2 PYTHON Bilişim
```

### ✏️ Örnek 9.2: Karışık kelimeleri düzeltelim

`"Yazılım", "Python", "Öğreniyorum", "Dilini"` değerlerini **sırasını değiştirmeden** `.format()`'a ver. `{1} {0} {3} {2}` kullanarak anlamlı cümle kur.

💡 **Beklenen çıktı:**
```text
Python Yazılım Dilini Öğreniyorum
```

### ✏️ Örnek 9.3: İsimle yerleştirme

`{a} {b} {a}` kalıbını `a="elma"`, `b="armut"` ile kullan.

💡 **Beklenen çıktı:**
```text
elma armut elma
```

### ✏️ Örnek 9.4: Basamak sınırlama

`pi = 3.14159` değerini 2 basamakla yazdır.

💡 **Beklenen çıktı:**
```text
Pi sayısı: 3.14
```

## 3. f-string

Metnin başına **`f`** harfi koyulur, değişkenler doğrudan `{}` içine yazılır. En okunabilir yöntemdir.

```text
ad = "Ayşe"
print(f"Merhaba {ad}!")
```

- `{}` içinde **işlem** de yapılabilir: `f"Toplam: {a + b}"`
- Basamak sınırlama: `f"{sayi:.2f}"`

> 🔹 f-string, Python 3.6 (2016) ile geldi.

### ✏️ Örnek 9.5: Üç yöntemi karşılaştıralım

`a = 5`, `b = 6`, `c = 9` tanımla. Toplamı **normal (virgül)**, **`.format()`** ve **f-string** ile üç ayrı hücrede yazdır.

💡 **Beklenen çıktı:**
```text
Girdiğiniz 5, 6 ve 9 değerlerinin toplamı 20'dir.
```

### ✏️ Örnek 9.6: Kullanıcı bilgisi

`ad` ve `sifre` değişkenlerini tanımla. f-string ve `\n` ile iki satırda yazdır.

💡 **Beklenen çıktı:**
```text
Kullanıcı adı: Nejdet
Şifre: 4646
```

### ✏️ Örnek 9.7: Ortalama (f-string ile)

İki yazılı notunu `input()` ile al. Ortalamayı f-string ile, **2 basamak** göstererek yazdır.

💡 **Beklenen çıktı:**
```text
Ders ortalaması: 77.50
```
