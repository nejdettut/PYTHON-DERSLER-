## 1. Üç Yaklaşım

| Yaklaşım | Fikir | Örnek | Artı / Eksi |
|---|---|---|---|
| **Kaba kuvvet** (brute force) | Tüm olasılıkları dene | Şifre kırma | ✅ Kesin sonuç · ❌ Çok yavaş |
| **Böl ve yönet** (divide & conquer) | Problemi küçük parçalara böl | İkili arama | ✅ Hızlı · ❌ Her probleme uymaz |
| **Açgözlü** (greedy) | Her adımda o anki en iyi seçimi yap | Para üstü verme | ✅ Basit, hızlı · ❌ Her zaman en iyiyi bulmaz |

## 2. Kaba Kuvvet 🔨

### ✏️ Örnek 14.1: PIN kırıcı

Gizli PIN `"0742"` olsun. `0000`'dan `9999`'a kadar tüm kombinasyonları deneyerek PIN'i bul ve kaç denemede bulduğunu yazdır. İpucu: `str(i).zfill(4)` sayının başına sıfır ekler.

💡 **Beklenen çıktı:**
```text
PIN bulundu: 0742 (743 denemede)
```

### ✏️ Örnek 14.2: Şifre uzunluğu neden önemli?

Sadece küçük harf ve rakamdan (36 karakter) oluşan şifreler için 4, 6, 8, 10 ve 12 karakterlik olası şifre sayısını hesapla. Saniyede 1 milyar deneme yapan bir bilgisayar en kötü durumda kaç **yıl** uğraşır?

💡 **Beklenen çıktı:**
```text
4 karakter: 1679616 olasılık → 0.0 yıl
...
12 karakter: 4738381338321616896 olasılık → 150.3 yıl
```

> 🔐 Bu yüzden uzun şifreler güvenlidir: Her karakter, kaba kuvvetin süresini **36 katına** çıkarır. O(2ⁿ) gibi üstel büyüme budur!

## 3. Açgözlü Yaklaşım 🤑

### ✏️ Örnek 14.3: Para üstü

Bozuk paralar: `[200, 100, 50, 20, 10, 5, 1]` kuruş. Verilen para üstünü **en az sayıda** madeni parayla ver: Her adımda **sığan en büyük** parayı seç. 387 kuruş için dene.

💡 **Beklenen çıktı:**
```text
200 x 1
100 x 1
50 x 1
20 x 1
10 x 1
5 x 1
1 x 2
Toplam 8 madeni para
```

### ✏️ Örnek 14.4: Açgözlü her zaman en iyi mi?

Paralar `[4, 3, 1]` olsun ve para üstü 6 olsun. Açgözlü algoritma kaç para veriyor? Daha az parayla verilebilir mi?

💡 **Beklenen çıktı:**
```text
Açgözlü: [4, 1, 1] → 3 para
En iyisi: [3, 3] → 2 para
```

## 4. Aynı Problem, Üç Çözüm: En Büyük İki Sayı

### ✏️ Örnek 14.5: Kaba kuvvet

`liste = [4, 2, 9, 7, 5, 6]` için **tüm ikilileri** (iç içe döngü, Ders 20) dene ve toplamı en büyük olan ikiliyi bul. Kaç ikili denedin?

💡 **Beklenen çıktı:**
```text
En büyük iki sayı: (9, 7) | 15 ikili denendi
```

### ✏️ Örnek 14.6: Tek geçişte

Listeyi **bir kez** dolaşarak `enb1` ve `enb2` değişkenlerini güncelle (defterdeki `find_top_two` yöntemi).

💡 **Beklenen çıktı:**
```text
(9, 7)
```

### ✏️ Örnek 14.7: Sıralayarak

YZ-13'teki kabarcık sıralamayla sıralayıp son iki elemanı al. Sonra aynısını `sorted()` ile tek satırda yap.

💡 **Beklenen çıktı:**
```text
(9, 7)
```
