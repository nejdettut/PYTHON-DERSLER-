## 1. Tam Yapı

```text
try:
    hata verebilecek kodlar
except HataTürü:
    hata OLURSA çalışır
else:
    hata OLMAZSA çalışır
finally:
    HER DURUMDA çalışır
```

| Blok | Ne zaman çalışır? |
|---|---|
| `try` | Her zaman denenir |
| `except` | Sadece hata olursa |
| `else` | Sadece hata **olmazsa** |
| `finally` | **Her durumda** (hata olsa da olmasa da) |

### ✏️ Örnek 42.1: else ile adım adım kontrol

Bir sayı al. Sayı değilse `Sayı girmediniz` yaz. Sayıysa (`else` bloğunda) `10 / sayi` sonucunu hesapla; orada da sıfıra bölmeyi ayrıca yakala.

💡 **Beklenen çıktı:**
```text
Sayı: 0
Sayı sıfıra bölünemez
```

### ✏️ Örnek 42.2: finally her zaman çalışır

Bir bölme işlemi yap. `finally` bloğuna `İşlem tamamlandı.` yaz. Hem hatalı hem hatasız çalıştır; mesaj her seferinde çıkıyor mu?

💡 **Beklenen çıktı:**
```text
Bölünen: 10
Bölen: 0
Sıfıra bölünemez!
İşlem tamamlandı.
```

## 2. Hata Alınca Tekrar Sor

`while True` + `try` + `break` = kullanıcı doğru veri girene kadar sormaya devam eden program.

```text
while True:
    try:
        sayi = int(input("Sayı: "))
        break                 # başarılıysa döngüden çık
    except ValueError:
        print("Tekrar dene!")  # başarısızsa döngü devam eder
```

### ✏️ Örnek 42.3: Doğru sayı gelene kadar

Kullanıcı geçerli bir tam sayı girene kadar sormaya devam et, sonra sayının karesini yazdır.

💡 **Beklenen çıktı:**
```text
Sayı: on
Bu bir sayı değil, tekrar dene!
Sayı: 10
Karesi: 100
```

### ✏️ Örnek 42.4: Döngülü hesap makinesi

İki sayı al ve böl. `ValueError` ve `ZeroDivisionError` olursa mesaj verip başa dön (`continue`). Başarılı olursa (`else`) sonucu yaz ve `Devam? (e/h)` diye sor.

💡 **Beklenen çıktı:**
```text
1. sayı: 9
2. sayı: 0
Bir sayı sıfıra bölünemez!
1. sayı: 9
2. sayı: 3
Sonuç: 3.0
Devam? (e/h): h
```

### ✏️ Örnek 42.5: Güvenli sayı fonksiyonu

`sayi_al(mesaj)` fonksiyonu yaz: kullanıcı geçerli bir tam sayı girene kadar sorsun ve sayıyı **döndürsün**. Bu fonksiyonla iki sayı alıp topla.

💡 **Beklenen çıktı:**
```text
Birinci sayı: x
Geçersiz giriş!
Birinci sayı: 4
İkinci sayı: 6
Toplam: 10
```

## 📝 Ödev 16 – Hata Ayıklama

Kullanıcıdan alınan **3 sayının ortalamasını** veren programı yazınız. Kullanıcı sayı dışında bir şey girerse program çökmesin; uyarı verip **o sayıyı tekrar** istesin. (İpucu: Örnek 42.5'teki fonksiyon!)

💡 **Örnek çıktı:**
```text
1. sayı: 70
2. sayı: yetmiş
Lütfen sayısal bir değer giriniz.
2. sayı: 80
3. sayı: 90
Ortalama: 80.0
```
