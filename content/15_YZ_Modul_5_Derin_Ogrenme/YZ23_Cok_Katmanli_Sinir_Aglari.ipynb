## 1. Katmanlar

```text
 Girdi katmanı      Gizli katman(lar)      Çıkış katmanı
   x₁  ●──────────▶  ●  ──────────────▶
                  ╲ ╱                      ●  → tahmin
   x₂  ●──────────▶  ●  ──────────────▶
```

- **Girdi katmanı:** Veriyi alır (her özellik bir nöron).
- **Gizli katman:** Ara hesaplamaları yapar; veriyi yeni bir biçime dönüştürür.
- **Çıkış katmanı:** Sonucu verir.

> 🔹 **Derin** öğrenme = **çok sayıda** gizli katman. Modern görüntü ve dil modellerinde yüzlerce katman olabilir.

### ✏️ Örnek 23.1: XOR'u iki katmanla çöz

YZ-21'deki `noron` fonksiyonunu kullan:
- Gizli nöron 1 = **VEYA** (w = 1, 1; b = −0.5)
- Gizli nöron 2 = **VE DEĞİL** (w = −1, −1; b = 1.5)
- Çıkış nöronu = **VE** (gizli nöronların çıktılarını girdi olarak alır; w = 1, 1; b = −1.5)

4 durum için XOR çıktısını hesapla.

💡 **Beklenen çıktı:**
```text
0 XOR 0 = 0
0 XOR 1 = 1
1 XOR 0 = 1
1 XOR 1 = 0
```

## 2. scikit-learn ile Sinir Ağı: `MLPClassifier`

**MLP** = Multi-Layer Perceptron (Çok Katmanlı Perceptron). Ağırlıkları **gradyan inişiyle** (YZ-15) kendisi öğrenir.

```text
from sklearn.neural_network import MLPClassifier
ag = MLPClassifier(hidden_layer_sizes=(8,), max_iter=2000, random_state=1)
#                  └─ 1 gizli katman, 8 nöron
ag.fit(X, y)
```

`hidden_layer_sizes=(16, 8)` → iki gizli katman: 16 ve 8 nöron.

> 💡 JupyterLite'ta `sklearn` ilk yüklemede 10–20 saniye sürebilir.

### ✏️ Örnek 23.2: XOR'u MLP öğrensin

`X = [[0, 0], [0, 1], [1, 0], [1, 1]]`, `y = [0, 1, 1, 0]`. `MLPClassifier(hidden_layer_sizes=(4,), activation="tanh", max_iter=5000, random_state=1)` ile eğit ve 4 durumu tahmin et.

💡 **Beklenen çıktı:**

[0 1 1 0]

### Ay Şekilli Veri 🌙🌙

Şimdi düz çizgiyle ayrılamayan, iç içe geçmiş iki hilal şeklinde bir veri üretelim:

````python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np

X, y = make_moons(n_samples=300, noise=0.2, random_state=0)
X_egitim, X_test, y_egitim, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

def sinir_ciz(model, X, y, baslik):
    """Modelin karar sınırını renkli bölgeler olarak çizer."""
    xx, yy = np.meshgrid(np.linspace(-2, 3, 200), np.linspace(-1.5, 2, 200))
    z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx, yy, z, alpha=0.3, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k", s=20)
    plt.title(baslik)

plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="k", s=20)
plt.title("Ay şekilli veri")
plt.show()
````

### ✏️ Örnek 23.3: Tek nöron yetmez

`from sklearn.linear_model import Perceptron` ile tek nöronlu bir model eğit. Test doğruluğunu yazdır ve `sinir_ciz(model, X, y, "Perceptron")` ile karar sınırını çiz.

💡 **Beklenen çıktı:**
