## 1. `input()` Fonksiyonu

`input()` kullanıcıdan **klavye ile** veri alır. Parantez içine yazdığın mesaj kullanıcıya gösterilir.

```text
ad = input("Adınızı giriniz: ")
```

Hücreyi çalıştırınca altında bir **kutu** açılır. Değeri yaz ve **Enter**'a bas.

> ⚠️ **Altın kural:** `input()` her zaman **metin (str)** döndürür. Kullanıcı `25` yazsa bile bu `"25"` olur.
> Hesap yapacaksan **mutlaka** `int()` veya `float()` ile çevir.

### ✏️ Örnek 8.1: Hoş geldin mesajı

Kullanıcının adını al ve hoş geldin mesajı yazdır.

💡 **Beklenen çıktı:**
```text
Adınızı giriniz: Ayşe
Hoş geldiniz, Ayşe
```

### ✏️ Örnek 8.2: Girilen verinin tipi

Kullanıcıdan bir sayı al. Sayıyı ve tipini yazdır.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 25
25
<class 'str'>
```

### ✏️ Örnek 8.3: Toplama tuzağı

Kullanıcıdan iki sayı al ve **dönüştürmeden** topla. 5 ve 6 girince ne oldu?

💡 **Beklenen çıktı:**
```text
Toplam: 56
```

### ✏️ Örnek 8.4: Tuzağı düzeltelim

Aynı programı önce `int(input(...))`, sonra `float(input(...))` ile yaz.

💡 **Beklenen çıktı:**
```text
Toplam: 11
```

## 2. Etkileşimli Programlar

### ✏️ Örnek 8.5: Kaç yaşında olacaksın?

Doğum yılını al. 2036 yılında kaç yaşında olacağını hesapla.

💡 **Beklenen çıktı:**
```text
Doğum yılınızı giriniz: 2010
2036 yılında 26 yaşında olacaksınız.
```

### ✏️ Örnek 8.6: Geleceğe yolculuk

Bu sefer hem doğum yılını hem de gelecekteki yılı kullanıcıdan al.

### ✏️ Örnek 8.7: Sınav ortalaması

1. ve 2. yazılı notlarını al, ortalamayı hesapla.

💡 **Beklenen çıktı:**
```text
1. yazılı: 70
2. yazılı: 85
Ortalamanız: 77.5
```

### ✏️ Örnek 8.8: Saniyeyi dakikaya çevirelim

Kullanıcıdan saniye al. Kaç dakika kaç saniye ettiğini bul. İpucu: `//` ve `%` ya da `divmod()`.

💡 **Beklenen çıktı:**
```text
Saniye giriniz: 200
200 saniye = 3 dakika 20 saniye
```

### ✏️ Örnek 8.9: Fatura hesaplama

Abone numarası, dönem ve tüketim miktarını (kWh) al. Birim fiyat **4.5 TL** olsun. Borcu hesapla.

💡 **Beklenen çıktı:**
```text
Sayın 12345 abonemiz, Ekim dönemi borcunuz 450.0 TL'dir.
```
