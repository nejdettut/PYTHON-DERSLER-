## 1. Silme Yöntemleri

| Yöntem | Ne siler? |
|---|---|
| `sozluk.pop("anahtar")` | Belirtilen anahtarı (değerini döndürür) |
| `sozluk.popitem()` | **En son eklenen** elemanı |
| `del sozluk["anahtar"]` | Belirtilen anahtarı |
| `sozluk.clear()` | Hepsini; sözlük **boş** kalır |
| `del sozluk` | Sözlüğü tamamen yok eder |

### ✏️ Örnek 33.1: pop ve popitem

`kisi = {"ad": "Nejdet", "yas": 36, "memleket": "Ankara"}` için önce `pop("yas")` yap, sonra yeniden tanımlayıp `popitem()` yap.

💡 **Beklenen çıktı:**
```text
{'ad': 'Nejdet', 'memleket': 'Ankara'}
{'ad': 'Nejdet', 'yas': 36}
```

### ✏️ Örnek 33.2: Renkleri temizle

`renk = {"black": "siyah", "green": "yeşil", "white": "beyaz"}` sözlüğünden önce `white`'ı sil (`del` ile), sonra tamamen boşalt.

💡 **Beklenen çıktı:**
```text
{'black': 'siyah', 'green': 'yeşil', 'white': 'beyaz'}
{'black': 'siyah', 'green': 'yeşil'}
{}
```

## 2. `get()`: Güvenli Okuma

| Yazım | Anahtar yoksa |
|---|---|
| `sozluk["x"]` | ❌ `KeyError` hatası |
| `sozluk.get("x")` | `None` döner |
| `sozluk.get("x", "Bulunamadı")` | `"Bulunamadı"` döner |

### ✏️ Örnek 33.3: Köşeli parantez vs get

`sayi = {"sıfır": 0, "bir": 1, "iki": 2}` için `sayi.get("bir")`, `sayi.get("dört")` ve `sayi.get("dört", "Yok")` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
1
None
Yok
```

### ✏️ Örnek 33.4: Rakamı yazıya çevir

0–9 rakamlarının yazılışlarını tutan bir sözlük oluştur. Kullanıcıdan rakam al, `get()` ile karşılığını; yoksa `Bu bir rakam değil` yazdır.

💡 **Beklenen çıktı:**
```text
Rakam giriniz: 7
yedi
```

## 3. Sözlükte Gezinme

```text
for anahtar in sozluk:                  # sadece anahtarlar
for deger in sozluk.values():           # sadece değerler
for anahtar, deger in sozluk.items():   # ikisi birlikte ✅
```

### ✏️ Örnek 33.5: items ile gezin

`notlar = {"Matematik": 85, "Fizik": 70, "Bilişim": 100}` sözlüğünü `items()` ile gezerek `Matematik: 85` şeklinde yazdır.

💡 **Beklenen çıktı:**
```text
Matematik: 85
Fizik: 70
Bilişim: 100
```

### ✏️ Örnek 33.6: Not ortalaması

Aynı sözlükteki notların ortalamasını `sum(notlar.values()) / len(notlar)` ile bul.

💡 **Beklenen çıktı:**
```text
Ortalama: 85.0
```

### ✏️ Örnek 33.7: Harf sayacı

Kullanıcıdan bir kelime al. Her harfin kaç kez geçtiğini bir sözlükte say. İpucu: `sayac[harf] = sayac.get(harf, 0) + 1`

💡 **Beklenen çıktı:**
```text
Kelime: ankara
{'a': 3, 'n': 1, 'k': 1, 'r': 1}
```

## 📝 Ödev 13 – Sözlükler

Klavyeden girilen **Türkçe gün adının İngilizce karşılığını** veren programı yazınız. Türkçe günler anahtar, İngilizce karşılıkları değer olsun. Kullanıcı `e` girene kadar program çalışmaya devam etsin; sözlükte olmayan bir giriş için `Hatalı giriş!` yazsın.

💡 **Örnek çıktı:**
```text
Türkçe gün (çıkış e): Pazartesi
Monday
Türkçe gün (çıkış e): pazar
Sunday
Türkçe gün (çıkış e): e
Çıkış yapılıyor...
```

## ✅ Bugün Öğrendiklerimiz
- `pop`, `popitem`, `del`, `clear` ile sözlükten silinir.
- `get(anahtar, varsayılan)` anahtar yoksa hata yerine varsayılanı döndürür.
- `for k, v in sozluk.items():` sözlükte gezmenin en kullanışlı yoludur.
