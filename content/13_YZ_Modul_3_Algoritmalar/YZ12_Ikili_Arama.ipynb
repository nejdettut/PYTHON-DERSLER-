## 1. Sayı Tahmin Oyununu Hatırla 🎲

Ders 18'de bilgisayarın 1–100 arasında tuttuğu sayıyı "daha büyük / daha küçük" ipuçlarıyla bulmuştuk. **En akıllı strateji neydi?**

Her seferinde **ortadaki** sayıyı söylemek! Böylece her tahminde ihtimallerin **yarısı** elenir:
```text
100 → 50 → 25 → 13 → 7 → 4 → 2 → 1      (en fazla 7 tahmin!)
```

## 2. İkili Arama Algoritması

> ⚠️ İkili arama sadece **sıralı** listede çalışır!

```text
1) alt = 0, ust = son indeks
2) alt <= ust olduğu sürece:
     orta = (alt + ust) // 2
     liste[orta] == hedef  → bulundu, orta'yı döndür
     liste[orta] < hedef   → aramaya SAĞ yarıda devam et: alt = orta + 1
     liste[orta] > hedef   → aramaya SOL yarıda devam et: ust = orta - 1
3) Döngü biterse → bulunamadı, -1 döndür
```

### ✏️ Örnek 12.1: Elle izle

`[1, 3, 5, 7, 9, 11, 13, 15]` listesinde 11'i ararken her adımda `alt`, `ust`, `orta` ve `liste[orta]` değerleri ne olur? Önce kâğıtta dene, sonra aşağıdaki fonksiyonla kontrol et.

### ✏️ Örnek 12.2: İkili arama fonksiyonu

Yukarıdaki algoritmayı `ikili_ara(liste, hedef)` fonksiyonu olarak `while` ile yaz. Her adımda `alt`, `ust`, `orta` değerlerini yazdır.

💡 **Beklenen çıktı:**
```text
alt=0 ust=7 orta=3 → 7
alt=4 ust=7 orta=5 → 11
Bulundu: 5. indeks
```

### ✏️ Örnek 12.3: Adım sayacı

Fonksiyonu yazdırma yapmadan, `(indeks, adim)` döndürecek şekilde düzenle. 1–1000 arası sayılardan oluşan listede 1, 500, 999 ve 1001'i ara.

💡 **Beklenen çıktı:**
```text
1 → 9 adım
500 → 1 adım
999 → 9 adım
1001 → 10 adım
```

## 3. Doğrusal mı, İkili mi?

### ✏️ Örnek 12.4: Yarış!

100, 1.000, 100.000 ve 1.000.000 elemanlı sıralı listelerde **en kötü durumda** (listede olmayan sayı) iki aramanın adım sayılarını tablo hâlinde yazdır. YZ-11'deki `dogrusal_ara` fonksiyonunu tekrar tanımlaman gerekecek.

💡 **Beklenen çıktı:**
```text
Eleman sayısı | Doğrusal | İkili
100           | 100      | 6
1000          | 1000     | 9
100000        | 100000   | 16
1000000       | 1000000  | 19
```

### ✏️ Örnek 12.5: Matematik sihri

İkili aramanın adım sayısı yaklaşık **log₂(n)**'dir. `import math` ile `math.log2(1_000_000)` değerini hesapla ve 12.4'teki sonuçla karşılaştır. Dünya nüfusu (8 milyar) için kaç adım gerekirdi?

💡 **Beklenen çıktı:**
```text
1 milyon için log2: 19.93
8 milyar için yaklaşık 33 adım
```

### ✏️ Örnek 12.6: Sıralı değilse?

`ikili_ara([9, 2, 7, 4, 1], 4)` sonucunu dene. Neden bulamadı?

> ⚠️ İkili arama sıralı olmayan listede yanlış sonuç verir!

### 🎯 Kendini Dene 1: Bilgisayar Senin Sayını Bulsun

Kullanıcı 1–1000 arası bir sayı tutsun. Bilgisayar **ikili arama** stratejisiyle tahmin etsin, kullanıcı `b` (büyük), `k` (küçük) veya `d` (doğru) desin. Bilgisayar en fazla 10 tahminde bulmalı!

💡 **Örnek çıktı:**
```text
Tahminim: 500 [b/k/d]: k
Tahminim: 250 [b/k/d]: b
Tahminim: 375 [b/k/d]: d
3 tahminde buldum! 🎉
```
