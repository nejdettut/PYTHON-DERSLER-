## 1. İç İçe Döngü

Bir döngünün içine başka bir döngü yazılabilir. **İç döngü bitmeden dış döngü bir sonraki tura geçmez.**

```text
for ders in range(1, 3):         # dış döngü: 2 tur
    for konu in range(1, 4):     # iç döngü: her dış turda 3 tur
        print(ders, konu)
```
Toplam tekrar: 2 × 3 = **6**

> 🔹 Saat gibi düşün: Akrep (dış döngü) 1 ilerlerken yelkovan (iç döngü) tam tur atar.

### ✏️ Örnek 20.1: Ders ve konular

3 ders ve her derste 3 konu olacak şekilde yazdır. `range()` ve f-string kullan.

💡 **Beklenen çıktı:**
```text
Ders 1 Konu 1
Ders 1 Konu 2
Ders 1 Konu 3
Ders 2 Konu 1
...
Ders 3 Konu 3
```

### ✏️ Örnek 20.2: Aynısı while ile

Örnek 20.1'i iç içe `while` ile yaz. İç sayacı **nerede** sıfırlamalısın?

### ✏️ Örnek 20.3: Çarpım tablosu

1'den 10'a kadar çarpım tablosunu yazdır. Her sayının tablosundan sonra bir boş satır bırak.

💡 **Beklenen çıktı:**
```text
1 x 1 = 1
1 x 2 = 2
...
10 x 10 = 100
```

### ✏️ Örnek 20.4: Dik üçgen deseni

İç içe döngü ile aşağıdaki deseni çiz. İpucu: iç döngü `range(satir)` kadar dönsün, `print("#", end=" ")`

💡 **Beklenen çıktı:**
```text
#
# #
# # #
# # # #
# # # # #
```

### 🎯 Kendini Dene 1: Bilgisayar Senin Sayını Tahmin Etsin (Bonus)

Aklından 1–10 arası bir sayı tut ve gir. Bilgisayarın **3 hakkı** olsun; her seferinde rastgele tahmin yapsın. Bilirse tebrik etsin, bilemezse kalan hakkını söylesin.

💡 **Örnek çıktı:**
```text
1 ile 10 arasında sayı tut: 6
Bilgisayarın 1. tahmini: 7
Bilemedin, 2 hakkın kaldı
Bilgisayarın 2. tahmini: 6
Tebrikler, 2. denemede bildin!
```

## 📝 Ödev 9 – Yıldızlı Bonus Ödev

Kullanıcı 1–20 arası bir sayı tutsun. Bilgisayar 3 hakla tahmin etsin ama **her yanlış tahminden sonra aralığı daraltsın**: sayı tahminden büyükse alt sınır yükselsin, küçükse üst sınır düşsün.

💡 **Örnek çıktı:**
```text
1-20 arası bir sayı tut: 11
Tahminim: 4
Bilemedim, aralık: 5-20
Tahminim: 17
Bilemedim, aralık: 5-16
Tahminim: 11
Tebrikler bana, 3. denemede bildim!
```

## ✅ Bugün Öğrendiklerimiz
- İç döngü, dış döngünün **her turunda baştan sona** çalışır.
- Toplam tekrar = dış tur sayısı × iç tur sayısı.
- `print()` boş çağrılırsa alt satıra geçer; desenlerde satır bitirmek için kullanılır.
