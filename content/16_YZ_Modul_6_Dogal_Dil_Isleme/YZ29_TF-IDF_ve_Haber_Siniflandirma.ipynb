## 1. TF-IDF: Her Kelime Eşit Değildir ⚖️

Kelime torbasında "ve", "bir" gibi her metinde geçen kelimeler de büyük sayılar alır. **TF-IDF** bunu düzeltir:

| Parça | Anlamı | Soru |
|---|---|---|
| **TF** (Term Frequency) | Kelime **bu** metinde kaç kez geçiyor? | Bu metin için önemli mi? |
| **IDF** (Inverse Document Frequency) | Kelime **kaç farklı** metinde geçiyor? (az metinde → yüksek puan) | Ayırt edici mi? |

> 🔹 **TF-IDF yüksek** = Bu metinde sık geçen **ama** diğer metinlerde nadir görülen kelime → metnin **anahtar kelimesi**.
