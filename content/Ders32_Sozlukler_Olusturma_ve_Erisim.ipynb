## 1. Sözlük (Dictionary) Nedir?

Gerçek sözlükte bir **kelimeye** bakarız ve **anlamını** buluruz. Python sözlüğünde de **anahtar (key)** ile **değere (value)** ulaşırız.

```text
ogrenci = {"ad": "Ayşe", "sinif": "9-A", "yas": 15}
           anahtar: değer
```

| Özellik | Liste | Sözlük |
|---|---|---|
| Parantez | `[ ]` | `{ }` |
| Erişim | İndeks: `liste[0]` | Anahtar: `sozluk["ad"]` |
| Elemanlar | Değer | **anahtar: değer** çifti |

- Anahtarlar **benzersizdir** (aynı anahtar iki kez olamaz). Genelde metin veya sayıdır.
- Değerler her tipte olabilir ve tekrar edebilir.

### ✏️ Örnek 32.1: Üç yoldan sözlük

Aynı sözlüğü üç yoldan oluştur: (1) `{"bir": 1, "iki": 2}`, (2) `dict(bir=1, iki=2)`, (3) `dict([("bir", 1), ("iki", 2)])`

💡 **Beklenen çıktı:**
```text
{'bir': 1, 'iki': 2}
```

### ✏️ Örnek 32.2: Boş sözlüğü doldur

`S = {}` oluştur, sonra `S["bir"] = 1` şeklinde 4 eleman ekle.

💡 **Beklenen çıktı:**
```text
{'bir': 1, 'iki': 2, 'üç': 3, 'dört': 4}
```

### ✏️ Örnek 32.3: Değerler her şey olabilir

Anahtarları 1, 2, 3, 4 olan bir sözlüğün değerlerine sırasıyla metin, ondalıklı sayı, liste ve demet koy.

💡 **Beklenen çıktı:**
```text
{1: 'Bir', 2: 2.0, 3: ['elma', 'armut'], 4: ('ali', 'veli')}
```

### ✏️ Örnek 32.4: Plaka sözlüğü

Anahtarları plaka kodu (int), değerleri il adı olan bir sözlük oluştur.

💡 **Beklenen çıktı:**
```text
{6: 'Ankara', 34: 'İstanbul', 35: 'İzmir'}
```

## 2. Erişim, Ekleme ve Değiştirme

```text
kisi["ad"]            → değeri oku
kisi["yas"] = 16      → anahtar VARSA değeri değiştirir
kisi["okul"] = "ABC"  → anahtar YOKSA yeni eleman ekler
kisi.keys()           → tüm anahtarlar
kisi.values()         → tüm değerler
```

### ✏️ Örnek 32.5: Kişi bilgileri

`kisi = {"ad": "Nejdet", "yas": 35, "memleket": "Ankara"}` tanımla. Adı ve yaşı ayrı ayrı yazdır.

💡 **Beklenen çıktı:**
```text
Nejdet
35
```

### ✏️ Örnek 32.6: Anahtarlar ve değerler

`iller = {"Konya": "42", "Ankara": "06", "Kars": "36"}` için `keys()` ve `values()` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
dict_keys(['Konya', 'Ankara', 'Kars'])
dict_values(['42', '06', '36'])
```

### ✏️ Örnek 32.7: Güncelle ve ekle

`kisi` sözlüğünde yaşı 36 yap, sonra `"meslek"` anahtarıyla yeni bir bilgi ekle.

💡 **Beklenen çıktı:**
```text
{'ad': 'Nejdet', 'yas': 36, 'memleket': 'Ankara', 'meslek': 'Öğretmen'}
```

### ✏️ Örnek 32.8: Kullanıcı eklesin

Kullanıcıdan bir anahtar ve değer al, `kisi` sözlüğüne ekle.

💡 **Beklenen çıktı:**
```text
Anahtar: hobi
hobi için değer: satranç
{..., 'hobi': 'satranç'}
```

### ✏️ Örnek 32.9: Olmayan anahtar

`kisi["telefon"]` yazdır.

> ⚠️ `KeyError: 'telefon'` alacaksın. Çözümü Ders 33'te: `get()`

### ✏️ Örnek 32.10: İngilizce–Türkçe sözlük

Boş bir sözlükle başla. Kullanıcı `0` girene kadar İngilizce kelime ve Türkçe karşılığını alıp sözlüğe ekle. Çıkışta sözlüğü yazdır.

💡 **Beklenen çıktı:**
```text
İngilizce kelime (çıkış 0): apple
Türkçe karşılığı: elma
İngilizce kelime (çıkış 0): 0
{'apple': 'elma'}
```

### 🎯 Kendini Dene 1: Kartvizitim

Kendi bilgilerinden (ad, okul, sınıf, hobi) bir sözlük oluştur. Sonra `for anahtar in kartvizit:` ile tüm bilgileri `anahtar: değer` şeklinde yazdır.

## ✅ Bugün Öğrendiklerimiz
- Sözlük `{anahtar: değer}` çiftlerinden oluşur; anahtarlar benzersizdir.
- `sozluk[anahtar]` okur; `sozluk[anahtar] = değer` ekler veya günceller.
- `keys()` anahtarları, `values()` değerleri verir.
