## 1. Toplam Değişkeni

Döngüde sayıları toplamak için **döngüden önce** bir kutu açarız ve içine 0 koyarız. Her turda kutuya ekleme yaparız.

```text
toplam = 0                 # 1) Döngüden ÖNCE sıfırla
for i in range(1, 6):
    toplam = toplam + i    # 2) Her turda ekle  (kısaca: toplam += i)
print(toplam)              # 3) Döngüden SONRA yazdır (girintisiz!)
```

| Tur | `i` | `toplam` |
|---|---|---|
| 1 | 1 | 0 + 1 = 1 |
| 2 | 2 | 1 + 2 = 3 |
| 3 | 3 | 3 + 3 = 6 |
| 4 | 4 | 6 + 4 = 10 |
| 5 | 5 | 10 + 5 = **15** |

### ✏️ Örnek 15.1: Döngüyü adım adım izleyelim

1'den 5'e kadar toplarken her turda `i` ve `toplam` değerini yazdır. Sonuç tablodakiyle aynı mı?

💡 **Beklenen çıktı:**
```text
i: 1 toplam: 1
i: 2 toplam: 3
i: 3 toplam: 6
i: 4 toplam: 10
i: 5 toplam: 15
```

### ✏️ Örnek 15.2: 1'den 100'e kadar toplam

1 ve 100 dahil.

💡 **Beklenen çıktı:**
```text
Toplam: 5050
```

> 💡 Matematikçi **Gauss** bu toplamı 8 yaşındayken saniyeler içinde bulmuş: (1+100) × 50 = 5050

### ✏️ Örnek 15.5: Girilen sayıya kadar toplam
