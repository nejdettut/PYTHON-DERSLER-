## 1. `try – except` Yapısı

```text
try:
    hata verebilecek kodlar
except HataTürü:
    hata olursa çalışacak kodlar
```

- `try` bloğu normal çalışır. **Hata olursa** Python hemen `except` bloğuna atlar.
- Hata **olmazsa** `except` bloğu atlanır.
- Program **çökmez**, kullanıcı anlaşılır bir mesaj görür.

> 🔹 Benzetme: `try` = cambazın ipte yürümesi · `except` = altındaki güvenlik ağı.

### ✏️ Örnek 41.1: Sıfıra bölme hatasını yakala

İki sayı al ve böl. `ZeroDivisionError` olursa `Bir sayı sıfıra bölünemez!` yazdır.

💡 **Beklenen çıktı:**
```text
Birinci sayı: 8
İkinci sayı: 0
Bir sayı sıfıra bölünemez!
```

### ✏️ Örnek 41.2: Harf girilirse?

Kullanıcıdan sayı al. `int()` dönüşümü başarısız olursa (`ValueError`) `Lütfen bir sayı giriniz!` yazdır. Ders 40'taki görevi hatırla!

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: on
Lütfen bir sayı giriniz!
```

### ✏️ Örnek 41.3: Geçersiz indeks

Ders 40'taki takım listesini `try–except IndexError` ile yeniden yaz.

💡 **Beklenen çıktı:**
```text
Bir indeks giriniz: 9
Bu indekste takım yok!
```

## 2. Birden Çok `except`

Farklı hatalar için farklı mesajlar verebiliriz. En sona yazılan genel `except Exception:` beklenmeyen tüm diğer hataları yakalar.

```text
try:
    ...
except ZeroDivisionError:
    ...
except ValueError:
    ...
except Exception:
    ...        # diğer her şey
```

### ✏️ Örnek 41.4: İki hatayı birden yakala

İki sayı al ve böl. Hem `ZeroDivisionError` hem `ValueError` için ayrı mesaj ver, en sona `except Exception` ekle. Üç kez çalıştır: normal sayılarla, 0 ile, harfle.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 12
Bir sayı giriniz: abc
Lütfen bir sayı giriniz!
```

### ✏️ Örnek 41.5: Sözlük hatası

`stok = {"elma": 12, "armut": 5}`. Kullanıcıdan ürün adı al ve stok miktarını yazdır. Ürün yoksa `KeyError`'u yakala.

💡 **Beklenen çıktı:**
```text
Ürün: muz
Bu ürün stokta tanımlı değil.
```

## 3. `except ... as hata`

Python'un orijinal hata mesajını görmek için hatayı bir değişkene alırız:
```text
except ValueError as hata:
    print("Hata:", hata)
```

### ✏️ Örnek 41.6: Orijinal mesajı göster

Örnek 41.4'ü her `except` için `as` kullanarak, Python'un orijinal hata mesajını da gösterecek şekilde yeniden yaz.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 5
Bir sayı giriniz: 0
Orijinal hata mesajı: division by zero
```

### 🎯 Kendini Dene 1: Güvenli Not Ortalaması

Kullanıcıdan virgülle ayrılmış notlar al (`80,90,abc`). Her notu `int()`'e çevir; çevrilemeyenleri `try–except` ile atla ve uyarı ver. Geçerli notların ortalamasını yazdır.

💡 **Örnek çıktı:**
```text
Notlar: 80,90,abc,70
'abc' geçerli bir not değil, atlandı.
Ortalama: 80.0
```

## ✅ Bugün Öğrendiklerimiz
- `try` bloğunda hata olursa program çökmez, `except` bloğu çalışır.
- Birden çok `except` ile her hata türüne farklı tepki verilir; `except Exception` geri kalan her şeyi yakalar.
- `except Hata as h:` ile Python'un orijinal mesajı okunur.
