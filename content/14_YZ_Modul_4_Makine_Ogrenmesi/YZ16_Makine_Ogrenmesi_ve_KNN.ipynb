## 2. K-En Yakın Komşu (KNN) 🏘️

**Fikir:** "Bana arkadaşını söyle, sana kim olduğunu söyleyeyim."

Yeni bir örneğin sınıfını bulmak için:
1. Eğitim verisindeki **her örneğe uzaklığını** hesapla.
2. **En yakın k** tanesini seç.
3. Bu komşular arasında **çoğunlukta olan** sınıfı tahmin et.

**Uzaklık** (iki özellik için, Pisagor!): `√((x1 − x2)² + (y1 − y2)²)`

Önce şu hazır veriyi çalıştır: Her meyvenin **ağırlığı** (gram) ve **pürüzlülüğü** (0 = pürüzsüz, 10 = çok pürüzlü) var.

````python
# (ağırlık, pürüzlülük, etiket)
meyveler = [
    (150, 2, "elma"), (170, 3, "elma"), (140, 1, "elma"), (130, 2, "elma"), (160, 2, "elma"),
    (180, 8, "portakal"), (200, 9, "portakal"), (190, 7, "portakal"), (210, 8, "portakal"), (175, 9, "portakal"),
]
print(len(meyveler), "etiketli meyve yüklendi.")
````

### ✏️ Örnek 16.1: Özellik ve etiket

Veride hangi değerler **özellik**, hangisi **etiket**? Özellikleri `X` listesine (`[agirlik, puruzluluk]`), etiketleri `y` listesine ayır ve yazdır.

### ✏️ Örnek 16.2: Veriyi çiz

Elmaları kırmızı, portakalları turuncu noktalarla saçılım grafiğinde göster. x ekseni ağırlık, y ekseni pürüzlülük olsun. Yeni bir meyveyi `(165, 6)` yıldız (`marker="*"`) ile ekle. Sence o ne?

### ✏️ Örnek 16.3: Uzaklık fonksiyonu

`uzaklik(a, b)` fonksiyonu iki noktanın arasındaki uzaklığı döndürsün (`math.sqrt`). `(165, 6)` ile `(150, 2)` arasındaki uzaklığı hesapla.

💡 **Beklenen çıktı:**
```text
15.52
```

### ✏️ Örnek 16.4: Tüm uzaklıklar

Yeni meyvenin her eğitim örneğine uzaklığını `(uzaklik, etiket)` olarak bir listeye ekle, **sırala** ve ilk 3'ünü yazdır.

💡 **Beklenen çıktı:**
```text
(5.83, 'elma')
(6.4, 'elma')
(10.44, 'portakal')
```

### ✏️ Örnek 16.5: Oylama

En yakın 3 komşunun etiketlerinden en çok geçeni bul (`count` veya sözlükle sayma, Ders 33). Tahmin ne?

💡 **Beklenen çıktı:**
```text
Komşular: ['elma', 'elma', 'portakal']
Tahmin: elma
```

### ✏️ Örnek 16.6: KNN fonksiyonu

16.3–16.5'i birleştirerek `knn_tahmin(veri, yeni, k=3)` fonksiyonunu yaz. `(135, 1)`, `(205, 9)` ve `(165, 6)` için k=1, 3 ve 5 ile tahmin yap.

💡 **Beklenen çıktı:**
```text
(135, 1) → k=1: elma, k=3: elma, k=5: elma
(205, 9) → k=1: portakal, k=3: portakal, k=5: portakal
(165, 6) → k=1: elma, k=3: elma, k=5: elma
```

### 💬 Tartışalım

`(165, 6)` meyvesi grafikte portakalların arasında gibi duruyordu; çok pürüzlü olduğu hâlde k ne olursa olsun neden hep **elma** çıktı? Ağırlık 100'lerle, pürüzlülük 0–10 arasında ölçülüyor. Bu, uzaklık hesabında hangi özelliği daha **önemli** yapıyor? (İpucu: YZ-8'deki **normalizasyon**!)
