## 1. time Modülü

| Fonksiyon | Ne yapar? |
|---|---|
| `time.time()` | 1 Ocak 1970'ten (**Epoch**) beri geçen saniye |
| `time.ctime()` | Okunabilir tarih-saat: `Tue Oct 13 10:55:00 2026` |
| `time.localtime()` | Tarih-saat bilgisini parça parça verir |
| `time.strftime("%d/%m/%Y")` | İstediğin biçimde tarih-saat metni |
| `time.sleep(2)` | Programı 2 saniye **bekletir** |

**Sık kullanılan biçim kodları:** `%Y` yıl · `%m` ay · `%d` gün · `%H` saat · `%M` dakika · `%S` saniye

`time.time()`, `time.ctime()` ve `time.localtime()` çıktılarını yazdır.

### ✏️ Örnek 39.2: Tarihi biçimlendir

`strftime` ile bugünün tarihini üç biçimde yazdır: `13/10/2026`, `2026-10-13 10:55:00` ve sadece saat `10:55`.

### ✏️ Örnek 39.3: Geri sayım

### ✏️ Örnek 39.4: Kronometre

## 2. Oyunlar 🎮

### ✏️ Örnek 39.5: Çarpım tablosu oyunu

Program 1–10 arası iki rastgele sayının çarpımını sorsun. Kullanıcı `0` girene kadar devam etsin. Sonunda doğru ve yanlış sayısını ve bir değerlendirme yazsın.

💡 **Beklenen çıktı:**
```text
Haydi hesapla 9x2= 18
Doğru Cevap!
Haydi hesapla 4x10= 2
Yanlış Cevap!
Haydi hesapla 8x5= 0
Çıkış yapıldı.
Doğru: 1  Yanlış: 1
Çarpman ortalarda
```

### ✏️ Örnek 39.6: 10 soruluk puanlı yarışma (Bonus)

Toplam 10 soru sorulsun; doğruya **+2**, yanlışa **−1** puan. Puana göre: >15 çok iyi · >10 iyi · >5 orta · >0 zayıf · diğer `ezber yapman lazım`.

## 📝 Ödev 15 – Bonus: Taş, Kağıt, Makas

Bilgisayara karşı **Taş–Kağıt–Makas** oyunu yazınız. Bilgisayarın seçimi `random.choice` ile yapılsın. Kimin ne seçtiğini ve kazananı yazdırın. Kullanıcı `0` girene kadar oyun devam etsin; sonunda skoru gösterin.

- Taş, makası kırar · Kağıt, taşı sarar · Makas, kağıdı keser

💡 **Örnek çıktı:**
```text
Seçiminiz (taş/kağıt/makas, çıkış 0): kağıt
Bilgisayar: taş
Kağıt taşı sarar, sen kazandın!
Seçiminiz (taş/kağıt/makas, çıkış 0): 0
Skor → Sen: 1  Bilgisayar: 0
```
