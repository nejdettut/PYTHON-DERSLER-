## 1. math Modülü

| Fonksiyon | Görevi | Örnek → Sonuç |
|---|---|---|
| `math.pi` | π sayısı | `3.14159...` |
| `math.pow(x, y)` | Üs alma | `pow(5, 3)` → `125.0` |
| `math.sqrt(x)` | Karekök | `sqrt(25)` → `5.0` |
| `math.factorial(x)` | Faktöriyel | `factorial(5)` → `120` |
| `math.ceil(x)` | **Yukarı** yuvarla | `ceil(4.2)` → `5` |
| `math.floor(x)` | **Aşağı** yuvarla | `floor(4.8)` → `4` |
| `math.trunc(x)` | Ondalığı **kes** | `trunc(-4.8)` → `-4` |
| `math.hypot(a, b)` | Hipotenüs | `hypot(3, 4)` → `5.0` |
| `math.cos(x)` | Kosinüs (**radyan** alır!) | `cos(math.radians(60))` → `0.5` |

> 💡 Tümünü görmek için: `print(dir(math))`

### ✏️ Örnek 37.1: Temel fonksiyonlar

`import math` ile 5'in 3. kuvvetini, 5'in faktöriyelini ve 25'in karekökünü yazdır.

💡 **Beklenen çıktı:**
```text
5'in 3. kuvveti: 125.0
5'in faktöriyeli: 120
25'in karekökü: 5.0
```

### ✏️ Örnek 37.2: Yuvarlama

Kullanıcıdan ondalıklı sayı al; `floor`, `ceil`, `trunc` ve gömülü `round` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
Ondalıklı sayı: 4.6
Aşağı: 4
Yukarı: 5
Kesilmiş: 4
Yuvarlanmış: 5
```

### ✏️ Örnek 37.3: Trigonometri tuzağı

`math.cos(60)` ve `math.cos(math.radians(60))` sonuçlarını yazdır. Hangisi beklediğin 0.5?

💡 **Beklenen çıktı:**
```text
-0.9524129804151563
0.5000000000000001
```

### ✏️ Örnek 37.4: Sadece pow'u al

`from math import pow` ile pow'u al ve kullan. Aynı hücrede `factorial(9)` çağırmayı dene. Sonra ikinci hücrede `from math import pow, factorial` ile düzelt.

> ⚠️ İlk hücre `NameError` verecek.

💡 **Beklenen çıktı:**
```text
362880
```

## 2. Geometri Problemleri

### ✏️ Örnek 37.5: Dairenin alanı

Kullanıcıdan yarıçap al. Alan = π × r². Sonucu 2 basamakla yazdır.

💡 **Beklenen çıktı:**
```text
Yarıçap: 4
Dairenin alanı: 50.27
```

### ✏️ Örnek 37.6: Hipotenüs

Dik üçgenin iki kenarını al. Hipotenüsü iki yolla bul: (1) `math.hypot`, (2) `sqrt(a² + b²)`.

💡 **Beklenen çıktı:**
```text
1. kenar: 3
2. kenar: 4
Hipotenüs: 5.0
```

### ✏️ Örnek 37.7: Heron formülü

Üç kenarı bilinen üçgenin alanını bulan `heron(a, b, c)` fonksiyonunu yaz.

`u = (a + b + c) / 2` → `Alan = √(u·(u−a)·(u−b)·(u−c))`

💡 **Beklenen çıktı:**
```text
Kenarlar: 3 4 5
Üçgenin alanı: 6.0
```

### 🎯 Kendini Dene 1: Silindir Hacmi

Yarıçap ve yüksekliği al. Hacim = π × r² × h. Sonucu `math.ceil` ile yukarı yuvarlayarak litre olarak yazdır (1 litre = 1000 cm³).

💡 **Örnek çıktı:**
```text
Yarıçap (cm): 10
Yükseklik (cm): 30
Hacim: 9425 cm³ ≈ 10 litre
```

## ✅ Bugün Öğrendiklerimiz
- `math` sabitler (`pi`) ve fonksiyonlar (`sqrt`, `pow`, `factorial`) içerir.
