## 1. Satır Satır Yazma

`write()` alt satıra **kendiliğinden geçmez**. Her satırın sonuna `\n` eklemelisin.

| Metot | Ne yapar? |
|---|---|
| `write("metin\n")` | Bir satır yazar |
| `writelines(liste)` | Listedeki metinleri **art arda** yazar (`\n`'i sen eklemelisin) |

### ✏️ Örnek 44.1: Alışveriş listesi dosyası

`alisveris.txt` dosyasına 4 ürünü **her biri ayrı satırda** olacak şekilde yaz. Sonra `read()` ile oku.

💡 **Beklenen çıktı:**
```text
süt
ekmek
yumurta
elma
```

### ✏️ Örnek 44.2: writelines

`sehirler = ["Ankara\n", "İzmir\n", "Kars\n"]` listesini `writelines` ile `sehirler.txt`'ye yaz ve oku. Sonra `\n`'leri silip tekrar dene. Fark ne?

## 2. Satır Satır Okuma

| Yöntem | Sonuç |
|---|---|
| `read()` | Tüm dosya **tek metin** |
| `readline()` | **Bir** satır (her çağrıda bir sonrakini verir) |
| `readlines()` | Tüm satırlar **liste** olarak |
| `for satir in dosya:` | Satırları döngüyle tek tek |

> 🔹 Okunan satırların sonunda `\n` vardır. Temizlemek için `satir.strip()`

> Aşağıdaki örneklerde İstiklal Marşı'nın ilk iki kıtasını kullanacağız. Önce şu **hazır** hücreyi çalıştır:

````python
metin = """Korkma, sönmez bu şafaklarda yüzen al sancak;
Sönmeden yurdumun üstünde tüten en son ocak.
O benim milletimin yıldızıdır, parlayacak;
O benimdir, o benim milletimindir ancak.
Çatma, kurban olayım, çehreni ey nazlı hilâl!
Kahraman ırkıma bir gül! Ne bu şiddet, bu celâl?
Sana olmaz dökülen kanlarımız sonra helâl...
Hakkıdır, Hakk'a tapan milletimin istiklâl!
"""
with open("istiklal.txt", "w", encoding="utf-8") as dosya:
    dosya.write(metin)
print("istiklal.txt hazır!")
````
