## 🛒 Senaryo

Bir online mağaza 200 müşterisinin verisini sana verdi ve şu soruların cevabını istiyor:
1. Müşteriler en çok hangi kategoriden alışveriş yapıyor?
2. Hangi kategori ve hangi yaş grubu daha çok harcıyor?
3. Müşterileri **düşük / orta / yüksek** harcayan diye gruplayabilir miyiz?

> 💡 Müşterileri benzerliklerine göre gruplamaya **segmentasyon** denir. Etiket olmadan grup bulmak, **denetimsiz öğrenmenin** temel fikridir.

Önce veri setini oluşturan hücreyi çalıştır:

````python
# Öğretmenin hazırladığı müşteri veri setini oluşturan hücre. Sadece çalıştır!
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 200
kategoriler = ["Elektronik", "Giyim", "Kitap", "Kozmetik", "Spor"]
taban = {"Elektronik": 4000, "Giyim": 1500, "Kitap": 400, "Kozmetik": 900, "Spor": 1200}

yas = rng.integers(18, 71, n).astype(float)
kategori = rng.choice(kategoriler, n)
odeme = [round(taban[k] * rng.uniform(0.5, 1.8) + y * 10, 2) for k, y in zip(kategori, yas)]
yas[rng.choice(n, 8, replace=False)] = np.nan

musteri = pd.DataFrame({
    "musteri_id": range(1001, 1001 + n),
    "yas": yas,
    "cinsiyet": rng.choice(["Kadın", "Erkek"], n),
    "sehir": rng.choice(["Ankara", "İstanbul", "İzmir", "Bursa"], n),
    "urun_kategori": kategori,
    "toplam_odeme": odeme,
})
musteri.to_csv("musteri_veriseti.csv", index=False, encoding="utf-8")
print("musteri_veriseti.csv hazır!", musteri.shape)
````
