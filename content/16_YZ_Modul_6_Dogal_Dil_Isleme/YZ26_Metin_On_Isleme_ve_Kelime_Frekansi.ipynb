## 1. Doğal Dil İşleme (NLP) Nedir? 💬

Bilgisayarların **insan dilini** anlamasını ve üretmesini sağlayan yapay zekâ dalıdır: çeviri, sohbet botları, duygu analizi, sesli asistanlar, otomatik özetleme.

**Neden zor?**
- "Yüz" → sayı mı, surat mı, yüzmek mi?
- "Harika, yine yağmur yağıyor 🙄" → olumlu kelime, olumsuz anlam (ironi)
- Türkçe **sondan eklemeli**: kitap → kitaplarımızdan

**Bilgisayar harf anlamaz, sayı anlar.** NLP'nin ilk adımı metni temizleyip sayılara çevirmektir:

```text
Ham metin → küçük harf → noktalama temizliği → kelimelere ayırma → etkisiz kelimeleri çıkarma → sayma
```

````python
haber = """Yapay zekâ, eğitimde büyük bir dönüşüm başlattı! Öğrenciler artık yapay zekâ ile
kişiselleştirilmiş öğrenme deneyimi yaşıyor. Öğretmenler ise yapay zekâ araçlarını derslerinde
kullanmaya başladı. Uzmanlara göre yapay zekâ, öğrencilerin eksiklerini daha hızlı tespit ediyor;
ancak öğretmenin rolü hâlâ çok önemli. Eğitimde yapay zekâ kullanımı her geçen gün artıyor."""
print(len(haber), "karakter")
````

### ✏️ Örnek 26.1: Türkçe küçük harf

`kucult(metin)` fonksiyonunu yaz: Önce `I` → `ı` ve `İ` → `i` değişimini yap, sonra `.lower()` uygula (YZ-2'deki ipucu). `haber`'i küçült ve ilk 80 karakterini yazdır.

### ✏️ Örnek 26.2: Noktalamayı temizle

`import string` ile `string.punctuation` içindeki her karakteri metinden sil (`replace`). Temizlenmiş metnin ilk 80 karakterini yazdır.

### ✏️ Örnek 26.3: Kelimelere ayır (tokenization)

Temiz metni `split()` ile kelimelere ayır. Kaç kelime var? Kaç **farklı** kelime var (küme!)?

💡 **Beklenen çıktı:**
```text
Toplam kelime: 48
Farklı kelime: 39
```

### ✏️ Örnek 26.4: Etkisiz kelimeleri çıkar

`etkisiz = {"ve", "ile", "bir", "bu", "da", "de", "için", "ise", "çok", "daha", "her", "göre", "ancak", "artık", "hâlâ"}` kümesindeki kelimeleri listeden çıkar. Kaç kelime kaldı?

### ✏️ Örnek 26.5: Hepsi bir arada

## 2. Kelime Frekansı 📊

Bir metinde en sık geçen kelimeler, metnin **konusunu** ele verir. `collections.Counter` sayma işini tek satırda yapar:
```text
from collections import Counter
sayac = Counter(kelimeler)
sayac.most_common(5)
```

### ✏️ Örnek 26.6: En sık kelimeler

`Counter` ile en sık geçen 5 kelimeyi bul. Sonra aynı işi Ders 33'teki sözlükle sayma kalıbıyla yap; sonuçlar aynı mı?

💡 **Beklenen çıktı:**
```text
[('yapay', 5), ('zekâ', 5), ('eğitimde', 2), ...]
```
