## 1. Liste Nedir?

Bir değişken tek bir değer tutar. **Liste** ise **birden çok değeri sırayla** tek bir değişkende tutar.

```text
meyve = ["elma", "armut", "portakal", "muz"]
```

- Köşeli parantez `[ ]` ile yazılır, elemanlar virgülle ayrılır.
- Farklı tipte elemanlar olabilir: `["Zeynep", 15, 3.5, True]`
- Listeler **değiştirilebilir**: eleman eklenir, silinir, değiştirilir.
- Veri tipi `list`'tir.

### ✏️ Örnek 25.1: Farklı listeler oluşturalım

Boş bir liste (`[]` ve `list()` ile iki yoldan), bir sayı listesi, bir isim listesi ve **karışık tipte** bir liste oluştur. Hepsini yazdır.

💡 **Beklenen çıktı:**
```text
[]
[]
[1, 2, 3, 4, 5]
['ali', 'veli', 'ayşe']
['zeynep', 36, 3.16, '?']
```

### ✏️ Örnek 25.2: Liste içinde değişken ve liste

`sayi = 15`, `kelime = "ATATÜRK"` ve `liste1 = [1, 2, 3]` tanımla. Bunları başka bir listenin elemanı yap ve yazdır.

💡 **Beklenen çıktı:**
```text
['a', 'b', 15, 'ATATÜRK', [1, 2, 3]]
```

### ✏️ Örnek 25.3: Metinden liste

`list("Python")` sonucunu yazdır.

💡 **Beklenen çıktı:**
```text
['P', 'y', 't', 'h', 'o', 'n']
```

## 2. İndeks

Tıpkı metinlerde olduğu gibi (Ders 6) liste elemanları da **0'dan** numaralanır.

```text
meyve = ["elma", "armut", "portakal", "muz"]
           0        1         2         3
          -4       -3        -2        -1
```

- `meyve[2]` → `"portakal"`
- Son eleman indeksi = **uzunluk − 1**
- `meyve[1] = "kiraz"` ile eleman **değiştirilir**.

### ✏️ Örnek 25.4: Elemana ulaşalım

`meyve = ["elma", "armut", "portakal", "muz", "karpuz", "kavun", "vişne", "kivi"]` tanımla. 5. indeksi, son elemanı ve sondan 3. elemanı yazdır.

💡 **Beklenen çıktı:**
```text
kavun
kivi
kavun
```

### ✏️ Örnek 25.5: Elemanı değiştirelim

Aynı listede 2. indeksi `"kiraz"`, 0. indeksi `"mandalina"` yap ve listeyi yazdır.

💡 **Beklenen çıktı:**
```text
['mandalina', 'armut', 'kiraz', 'muz', 'karpuz', 'kavun', 'vişne', 'kivi']
```

### ✏️ Örnek 25.6: Elemanın içindeki harf

`meyve[-3]` bir metindir. Onun 3. harfini `meyve[-3][3]` ile yazdır.

💡 **Beklenen çıktı:**
```text
u
```

### ✏️ Örnek 25.7: Var olmayan indeks

8 elemanlı listede `meyve[8]` yazdır.

> ⚠️ `IndexError: list index out of range`. Son indeks 7'dir.

## 3. Dilimleme

Metinlerdeki kural aynen geçerli: `liste[başlangıç : bitiş : adım]` → **bitiş dahil değil**.

### ✏️ Örnek 25.8: Dilimleme alıştırması

`dersler = ["Python", "C++", "Java", "C#", "PHP"]` tanımla ve sırasıyla yazdır: tamamı, ilk 4, 2. indeksten sona, ikişer atlayarak, sondan ikinci, tersten.

💡 **Beklenen çıktı:**
```text
['Python', 'C++', 'Java', 'C#', 'PHP']
['Python', 'C++', 'Java', 'C#']
['Java', 'C#', 'PHP']
['Python', 'Java', 'PHP']
C#
['PHP', 'C#', 'Java', 'C++', 'Python']
```

### ✏️ Örnek 25.9: Aynı dilimi farklı yollarla

`dersler` listesinden `['C++', 'Java']` dilimini **en az iki farklı** yazımla al. (Pozitif ve negatif indeksleri karıştırabilirsin.)

💡 **Beklenen çıktı:**
```text
['C++', 'Java']
```

### ✏️ Örnek 25.10: Eşya listede var mı?

`esya = ["ayna", "televizyon", "perde"]` tanımla. Kullanıcıdan bir eşya al ve listede olup olmadığını yazdır.

💡 **Beklenen çıktı:**
```text
Eşya giriniz: perde
Girdiğiniz perde listede var.
```

### 🎯 Kendini Dene 1: Haftanın Programı

`gunler` listesine haftanın 7 gününü yaz. Hafta içi günleri ve hafta sonu günlerini **dilimleme ile** ayrı ayrı yazdır. Sonra `"Pazar"`ı `"Tatil"` olarak değiştir.

💡 **Örnek çıktı:**
```text
Hafta içi: ['Pazartesi', 'Salı', 'Çarşamba', 'Perşembe', 'Cuma']
Hafta sonu: ['Cumartesi', 'Pazar']
```

## ✅ Bugün Öğrendiklerimiz
- Liste `[ ]` ile yazılır, farklı tipte değerler tutabilir ve değiştirilebilir.
- İndeks 0'dan başlar; `-1` son elemandır.
- `liste[i] = yeni` ile eleman değiştirilir; dilimleme metinlerdeki gibidir.
