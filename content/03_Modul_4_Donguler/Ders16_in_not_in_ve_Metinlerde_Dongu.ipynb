## 1. `in` ve `not in`

| İfade | Sonuç |
|---|---|
| `x in s` | x, s'nin içinde **varsa** `True` |
| `x not in s` | x, s'nin içinde **yoksa** `True` |

Metinlerde, `range()`'de ve listelerde kullanılabilir.

### ✏️ Örnek 16.1: Harf var mı?

`"p" in "python"`, `"a" in "python"` ve `"p" not in "python"` sonuçlarını yazdır.

💡 **Beklenen çıktı:**
```text
True
False
False
```

### ✏️ Örnek 16.2: E-posta kontrolü

Kullanıcıdan e-posta adresi al. İçinde `@` varsa `Geçerli görünüyor`, yoksa `Geçersiz e-posta` yazdır.

💡 **Beklenen çıktı:**
```text
E-posta: ayse.okul.com
Geçersiz e-posta
```

## 2. Listeye İlk Bakış 👀

**Liste**, birden fazla değeri köşeli parantez içinde saklar:

```text
meyve = ["elma", "armut", "muz", "çilek"]
```

Şimdilik bu kadarını bilmen yeterli: listeler `in` ile sorgulanabilir ve `for` ile dolaşılabilir.
Ayrıntılar **Modül 6: Listeler**'de.

### ✏️ Örnek 16.3: Listede portakal var mı?

`meyve` listesini tanımla. `if – else` ve `in` ile portakalın listede olup olmadığını yazdır.

💡 **Beklenen çıktı:**
```text
Meyve listesinde portakal yoktur.
```

### ✏️ Örnek 16.4: Listeyi döngüyle dolaşalım

Listedeki meyveleri **yan yana** yazdır.

💡 **Beklenen çıktı:**
```text
Listedeki meyveler: elma armut muz çilek
```

### ✏️ Örnek 16.5: Kullanıcı sorgulasın

Kullanıcıdan bir meyve adı al ve listede olup olmadığını f-string ile yazdır.

💡 **Beklenen çıktı:**
```text
Meyve adı giriniz: muz
Meyve listesinde muz vardır.
```

## 3. Metinlerde Döngü

### ✏️ Örnek 16.6: Bir harf kaç kez geçiyor?

`cumle = "Karakter dizileri ve listeler üzerinde for döngüsü ile iterasyon işlemleri yapabiliriz."` tanımla. Kullanıcıdan bir harf al, cümlede kaç kez geçtiğini **sayaç** ile bul.

💡 **Beklenen çıktı:**
```text
Harf giriniz: a
Girilen harf cümlede 5 adet bulunmaktadır.
```

### ✏️ Örnek 16.7: Kelime sayısı

Kullanıcıdan bir cümle al. Boşlukları sayarak kelime sayısını bul. (Kelime sayısı = boşluk sayısı + 1)

💡 **Beklenen çıktı:**
```text
Cümleyi giriniz: Python çok eğlenceli bir dil
Cümlede 5 kelime var.
```

### ✏️ Örnek 16.8: Sesli harfleri yazdıralım

`sesli = "aeıioöuüAEIİOÖUÜ"` tanımla. Kullanıcıdan alınan cümledeki sesli harfleri yan yana yazdır. İpucu: `if harf in sesli:`

💡 **Beklenen çıktı:**
```text
Bir cümle giriniz: Bilişim
i,i,i,
```

### ✏️ Örnek 16.9: Sesli harf sayısı

Aynı mantıkla bu sefer sesli harfleri **say**.

💡 **Beklenen çıktı:**
```text
Bir cümle giriniz: Bilişim Teknolojileri
Toplam sesli harf sayısı: 9
```

### 🎯 Kendini Dene 1: 3'ün Katları

`sayilar = [4, 9, 12, 17, 21, 30, 44, 51]` listesindeki 3'ün katlarını yan yana yazdır ve kaç tane olduğunu söyle.

💡 **Örnek çıktı:**
```text
9 12 21 30 51
3'ün katı olan sayı adedi: 5
```

## ✅ Bugün Öğrendiklerimiz
- `in` içinde var mı, `not in` içinde yok mu diye sorar; sonuç `True`/`False`'tur.
- Liste `[ ... ]` ile yazılır; `for` ile dolaşılabilir.
- Metindeki karakterler döngüyle tek tek kontrol edilip sayılabilir.
