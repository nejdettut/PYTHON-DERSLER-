## 1. pandas ve DataFrame 🐼

`pandas`, Python'un en çok kullanılan veri analizi kütüphanesidir. Tabloları **DataFrame** adlı bir yapıda tutar (Excel sayfası gibi).

| Kod | Ne yapar? |
|---|---|
| `pd.read_csv("dosya.csv")` | CSV'yi DataFrame olarak okur |
| `df.head()` / `df.tail()` | İlk / son 5 satır |
| `df.shape` | (satır, sütun) sayısı |
| `df.columns` | Sütun adları |
| `df.info()` | Sütun tipleri ve boş olmayan değer sayıları |
| `df.describe()` | Sayısal sütunların özet istatistikleri |

> 💡 JupyterLite'ta `import pandas` ilk seferde birkaç saniye sürebilir.

## 2. Seçme, Süzme, Sıralama

| Kod | Ne yapar? |
|---|---|
| `df["puan"]` | Tek sütun |
| `df[["isim", "puan"]]` | Birden çok sütun |
| `df["puan"].mean()` | Sütunun ortalaması (`max`, `min`, `sum` de var) |
| `df[df["puan"] >= 85]` | Koşula uyan satırlar |
| `df[(df["cinsiyet"] == "K") & (df["puan"] > 80)]` | Birden çok koşul (`&` = ve, `\|` = veya) |
| `df.sort_values("puan", ascending=False)` | Sıralama |
