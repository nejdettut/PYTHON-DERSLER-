## 1. Telefonundaki Kelime Önerisi 📱

Mesaj yazarken klavyen bir sonraki kelimeyi tahmin eder. Nasıl? Daha önce yazılmış milyonlarca cümlede "**günaydın**"dan sonra en sık hangi kelimelerin geldiğini **saymıştır**.

**Bigram (ikili) modeli:** Her kelimeden sonra hangi kelimelerin kaç kez geldiğini bir sözlükte tutarız.

```text
"küçük kedi bahçede"  →  {"küçük": ["kedi"], "kedi": ["bahçede"]}
```
Sonra bir kelimeden başlayıp, her adımda olası sonraki kelimelerden birini **rastgele** seçerek yeni metin üretiriz.

````python
# Kendi yazdığımız kısa bir metin. Sadece çalıştır!
metin = """
küçük kedi bahçede oynuyordu . küçük kedi bir kelebek gördü .
kelebek çiçeğin üstüne kondu . kedi kelebeği yakalamak istedi .
kelebek uçtu ve ağacın üstüne kondu . kedi ağaca bakıp miyavladı .
küçük kedi annesinin yanına koştu . annesi kediye süt verdi .
kedi sütü içti ve bahçede uyudu . kelebek çiçeğin üstünde dans etti .
"""
kelimeler = metin.split()
print(len(kelimeler), "kelime yüklendi.")
````

### ✏️ Örnek 30.1: Bigram sözlüğü

`sonraki = {}` sözlüğüne her kelimeden sonra gelen kelimeyi listeye ekleyerek bigram modelini kur (Ders 33'teki kalıp). `"kedi"` ve `"kelebek"` kelimelerinden sonra neler gelmiş?

💡 **Beklenen çıktı:**
```text
kedi → ['bahçede', 'bir', 'kelebeği', 'ağaca', 'annesinin', 'sütü']
kelebek → ['gördü', 'çiçeğin', 'uçtu', 'çiçeğin']
```

### ✏️ Örnek 30.2: Olasılıklar

`Counter` ile `"kelebek"` kelimesinden sonra gelen kelimelerin olasılıklarını yüzde olarak yazdır.

💡 **Beklenen çıktı:**
```text
çiçeğin: %50.0
gördü: %25.0
uçtu: %25.0
```

### ✏️ Örnek 30.3: Metin üret ✨

`uret(baslangic, uzunluk=15)` fonksiyonu başlangıç kelimesinden itibaren her adımda `random.choice(sonraki[kelime])` ile kelime seçerek yeni bir metin üretsin. `"küçük"` ile birkaç kez çalıştır.

### ✏️ Örnek 30.4: En olası kelime

Rastgele seçmek yerine her adımda **en sık** gelen kelimeyi seçen `uret_en_olasi` fonksiyonunu yaz. Ne oldu? Neden aynı cümle tekrar etmeye başladı?

### ✏️ Örnek 30.5: Kendi metninle

En az 10 cümlelik kendi metnini yaz (bir masal, şarkı değil kendi cümlelerin!) ve modelini onunla eğitip metin üret.

## 2. Büyük Dil Modelleri (LLM) 🤖

ChatGPT, Gemini, Claude gibi modeller de **temelde aynı şeyi** yapar: **sonraki parçayı (token) tahmin etmek.** Farkları:

| | Bizim bigram modelimiz | Büyük dil modeli |
|---|---|---|
| Baktığı geçmiş | Sadece **1** önceki kelime | Binlerce önceki kelime (bağlam) |
| Eğitim verisi | 10 cümle | İnternetteki dev metin yığınları |
| Yöntem | Sayma | Milyarlarca ağırlıklı derin sinir ağı (Transformer) |
| Parametre sayısı | Birkaç düzine sayı | Milyarlarca |

> ⚠️ Bu modeller "doğru olanı" değil, **en olası görüneni** üretir. Bu yüzden emin bir dille yanlış bilgi verebilirler (**halüsinasyon**). Önemli bilgileri her zaman kaynağından kontrol et!
