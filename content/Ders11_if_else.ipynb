## 1. `if – else` Yapısı

Bazen koşul **sağlanmadığında** da bir şey yapılması gerekir. `else` Türkçede **"aksi hâlde / değilse"** demektir.

```text
if koşul:
    koşul True ise çalışır
else:
    koşul False ise çalışır
```

- Bu yapıda **mutlaka iki bloktan biri** çalışır, ikisi birden asla çalışmaz.
- `else` tek başına kullanılamaz, her zaman bir `if`'ten sonra gelir.
- `else` satırına **koşul yazılmaz**, sadece `else:` yazılır.

### ✏️ Örnek 11.1: Reşit mi?

Kullanıcıdan yaşını al. 18 ve üzeriyse `Reşitsiniz`, değilse `Reşit değilsiniz` yazdır.

💡 **Beklenen çıktı:**
```text
Yaşınızı giriniz: 16
Reşit değilsiniz
```

### ✏️ Örnek 11.2: `else`'e koşul yazarsak?

Örnek 11.1'i `else yas < 18:` yazarak tekrar dene.

> ⚠️ `SyntaxError` alacaksın. `else` koşul almaz.

### ✏️ Örnek 11.3: Tek mi, çift mi?

Kullanıcıdan bir sayı al. İpucu: Bir sayının 2'ye bölümünden kalan 0 ise sayı çifttir → `sayi % 2 == 0`

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 7
Girdiğiniz sayı tek sayıdır
```

### ✏️ Örnek 11.4: Daha kısa giriş kontrolü

Ders 10'daki kullanıcı adı-şifre kontrolünü bu sefer `and` ve `if – else` ile **tek kontrolde** yaz. Giriş başarılıysa üç satır mesaj, değilse tek uyarı göster.

💡 **Beklenen çıktı:**
```text
Kullanıcı Adını Giriniz: admin
Parolayı Giriniz: 123456
Giriş Başarılı...
Hoş geldin, admin
Tüm yetkiye sahipsiniz.
```

### ✏️ Örnek 11.5: Bay / Bayan karşılama

Kullanıcıdan ad ve cinsiyet al. Cinsiyet `Erkek`, `ERKEK` veya `erkek` ise `Bay ..., hoş geldiniz`, değilse `Bayan ..., hoş geldiniz` yazdır. Çıktıyı **f-string** ile oluştur.

💡 **Beklenen çıktı:**
```text
Adınızı giriniz: Ali
Cinsiyetiniz [Kadın/Erkek]: erkek
Bay Ali, hoş geldiniz
```

### ✏️ Örnek 11.6: Araç kullanabilir mi?

Yaşı 18 ve üzeriyse `Araç kullanabilirsiniz`, değilse `Araç kullanamazsınız` yazdır.

💡 **Beklenen çıktı:**
```text
Yaşınızı giriniz: 19
Araç kullanabilirsiniz
```

### ✏️ Örnek 11.7: 5'e bölünür mü?

Kullanıcıdan alınan sayının 5'e tam bölünüp bölünmediğini yazdır.

💡 **Beklenen çıktı:**
```text
Bir sayı giriniz: 35
35 sayısı 5'e tam bölünür.
```

### 🎯 Kendini Dene 1: Güçlü Şifre

Kullanıcıdan bir şifre al. Şifre **en az 8 karakterse** `Şifre kabul edildi`, değilse kaç karakter eksik olduğunu yazdır. İpucu: `len()`

💡 **Örnek çıktı:**
```text
Şifre: abc12
Şifre çok kısa! 3 karakter daha eklemelisin.
```

## ✅ Bugün Öğrendiklerimiz
- `if – else` yapısında iki bloktan **mutlaka biri** çalışır.
- `else` koşul almaz ve tek başına kullanılamaz.
- `sayi % 2 == 0` çift sayı, `sayi % 5 == 0` beşe bölünebilme kontrolüdür.
