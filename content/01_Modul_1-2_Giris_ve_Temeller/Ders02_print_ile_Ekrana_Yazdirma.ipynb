## 1. Tırnak Türleri

| Yazım | Ne zaman? |
|---|---|
| `"Merhaba"` | Normal metin |
| `'Merhaba'` | Normal metin (aynı sonuç) |
| `"""..."""` veya `'''...'''` | **Birden çok satıra** yayılan metin |

İçinde kesme işareti (`'`) olan bir metni **çift tırnakla** yazarız: `"Ankara'nın Bağları"`

### ✏️ Örnek 2.1: Aynı metni tek ve çift tırnakla yazdıralım

💡 **Beklenen çıktı:**
```text
Merhaba Dünya
Merhaba Dünya
```

### ✏️ Örnek 2.2: Üçlü tırnakla çok satırlı metin yazdıralım

Tek bir `print()` ile 4 satırlık bir şiir veya mesaj yazdır.

💡 **Beklenen çıktı:**
```text
Merhaba
Marslı
Dünyalı
Jüpiterli
```

### ✏️ Örnek 2.3: Tırnak içinde tırnak kullanalım

`Ankara'nın Bağları` ve `Atatürk dedi ki: 'İlk hedefiniz Akdenizdir!'` cümlelerini yazdır.

💡 **Beklenen çıktı:**
```text
Ankara'nın Bağları
Atatürk dedi ki: 'İlk hedefiniz Akdenizdir!'
```

## 2. Sayı mı, Metin mi?

Tırnak **içindeki** her şey metindir, Python onu hesaplamaz.
Tırnak **dışındaki** sayılarla Python işlem yapar.

### ✏️ Örnek 2.4: Farkı görelim

Aşağıdaki üç satırı yaz, çalıştırmadan önce sonucu **tahmin et**:

`print(6*4)` · `print("6*4")` · `print("6"*4)`

💡 **Beklenen çıktı:**
```text
24
6*4
6666
```

### ✏️ Örnek 2.5: Virgül ile artı (+) arasındaki fark

`"Python", "Yazılım", "Dili"` kelimelerini önce **virgülle**, sonra **+ ile** yazdır. Boşluklara dikkat et!

💡 **Beklenen çıktı:**
```text
Python Yazılım Dili
PythonYazılımDili
```

## 3. Kaçış Karakterleri

| Karakter | Görevi |
|---|---|
| `\n` | Alt satıra geçer (**n**ew line) |
| `\t` | Tab kadar boşluk bırakır |
| `\\` | Ekrana bir tane `\` yazar |

### ✏️ Örnek 2.6: `\n` ile alt satıra geçelim

Tek bir `print()` ile "Python ile" ve "Yapay Zeka" ifadelerini iki satırda yazdır.

💡 **Beklenen çıktı:**
```text
Python ile
Yapay Zeka
```

### ✏️ Örnek 2.7: `\t` ile boşluk bırakalım

Ad, Soyad ve Sınıf başlıklarını aralarında tab olacak şekilde yazdır.

💡 **Beklenen çıktı:**
```text
Ad	Soyad	Sınıf
```

### ✏️ Örnek 2.8: Ters eğik çizgi yazdıralım

Önce `print("\")` yazıp çalıştır, sonra düzelt.

> ⚠️ İlk kod hata verecek. Tek `\` tırnağı "kaçırır". Doğrusu için iki tane yazmalısın.

## 4. `sep=` ve `end=` Parametreleri

- **`sep=`** → Virgülle ayrılan ifadelerin **arasına** ne konacağını belirler. Varsayılanı bir boşluktur.
- **`end=`** → `print()` bittiğinde **sonuna** ne konacağını belirler. Varsayılanı `\n` (alt satır).

### ✏️ Örnek 2.9: `sep=` ile harfleri birleştirelim

`"P","Y","T","H","O","N"` harflerini önce normal, sonra `sep=""`, sonra `sep="-"` ile yazdır.

💡 **Beklenen çıktı:**
```text
P Y T H O N
PYTHON
P-Y-T-H-O-N
```

### ✏️ Örnek 2.10: Web adresi oluşturalım

`"www", "google", "com"` ifadelerini `sep=` kullanarak birleştir.

💡 **Beklenen çıktı:**
```text
www.google.com
```

### ✏️ Örnek 2.11: `end=` ile aynı satırda kalalım

İki ayrı `print()` yaz ama ikisi **aynı satırda** görünsün.

💡 **Beklenen çıktı:**
```text
Merhaba Dünya
```

### ✏️ Örnek 2.12: Yıldız (*) hilesi

`print(*"NEJDET", sep="\n")` yaz. Sonra kendi adınla dene. Bir de `sep="-"` ile çalıştır.

💡 **Beklenen çıktı:**
```text
N
E
J
D
E
T
```
